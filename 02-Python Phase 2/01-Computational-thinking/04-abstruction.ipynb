```python
# Understand the parts of a function:
def greet(name):
    print("Hello, " + name + "!")

greet("Ada")
greet("Grace")

"""
def tells Python that you are defining a function.
greet is the function's name. It describes its responsibility.
name is a parameter: a placeholder for the input the function needs.
The indented print(...) line is the function body—the instructions that run when the function is called.
greet("Ada") is a function call. It runs the function and supplies "Ada" as the input.
greet("Grace") calls the same function again with different input.
"""
```

```python
# Understand the parts of a function:
def greet(name):
    return "Hello, " + name + "!"

greet("Ada")
greet("Grace")
```

```python
"""
# Function 1 uses print() to display both greetings immediately, whereas Function 2 uses return to send back a value, and because WE don't print or store the returned values, only the last expression's result may be shown by OUR interactive environment.
"""
```

```python
# Parameters and arguments
def greet(name):       # name is the parameter
    print(name)

greet("Ada")           # "Ada" is the argument
```

```python
# Return value versus print:
def add(a, b):
    return a + b

result = add(3, 4)
print(result)
```

```python
# Abstraction can contain selection and iteration:
# A function does not replace if statements or loops. It can organize them under one responsibility.

def is_password_long_enough(password):
    return len(password) >= 8

is_password_long_enough("password123")
```

```python
# Exercise 1 — Write and test is_even(number)
def is_even(number):
    return number % 2 == 0

test_numbers = [4, 7, 10, 13]

for number in test_numbers:
    result = is_even(number)
    print(number, "->", result)
```

```python
# Exercise 4 — Create traffic_light_message(color):
def traffic_light_message(color):
    if color == "red":
        return "Stop"
    elif color == "yellow":
        return "Prepare to stop"
    elif color == "green":
        return "Go"
    else:
        return "Unknown color"

colors = ["green", "yellow", "red", "green"]

for color in colors:
    message = traffic_light_message(color)
    print(message)
```
