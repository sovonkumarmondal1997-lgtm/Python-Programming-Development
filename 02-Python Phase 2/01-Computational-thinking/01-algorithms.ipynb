```python
# calculate grocery limits overbudget or not:
"""
grocry items: rice = 400, milk = 230, vegetable = 300, fruits = 200, meets = 500, snacks = 180
total_budget = 1500
"""
# plan the algorithms in plain english:
"""
1. set the grocery items and their prices
2. set the total budget 
3. calculate the total cost of the grocery items
4. compare the total cost of the grocery items with the total budget 
5. if the total cost is grater than the total budget, then print overbudget,
    otherwise print within budget
"""

# solution 01:
grocery_items = [400, 230, 300, 200, 500, 180]

total_budget = 1500
total_cost = 0

total_cost = total_cost + grocery_items[0]
total_cost = total_cost + grocery_items[1]
total_cost = total_cost + grocery_items[2]
total_cost = total_cost + grocery_items[3]
total_cost = total_cost + grocery_items[4]
total_cost = total_cost + grocery_items[5]

print("Total_cost of grocery items: ", total_cost)

if total_cost > total_budget:
    print("Overbudget")
else:
    print("Within budget")
```

```python
# solution 02:
rice = 400
milk = 230
vegetable = 300
fruits = 200
meets = 500
snacks = 180
total_budget = 1500

total_cost = rice + milk + vegetable + fruits + meets + snacks
print("Total cost of grocery items: ", total_cost)

if total_cost > total_budget:
    print("Overbudget")
else:
    print("Within budget")
```
