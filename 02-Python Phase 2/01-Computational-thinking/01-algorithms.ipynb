```python
"""
# Why Algorithms, Programs, and Data matters:
Every software application works using three things: Data (what information it uses), Algorithm (the steps to solve a problem), and Program (the Python code that executes those steps).

# Key_terms:
Algorithm: A step-by-step plan for solving a problem.
Program: A set of instructions written in a programming language that a computer can execute.
Data: Information that a program uses, processes, or produces.
Source Code: The written Python instructions that tell the computer what to do.
Runtime Behavior: What actually happens when a program runs.
Interpreter: A program that reads Python code and executes its instructions.
Instruction: A single command that tells the computer to perform a specific task.
"""
```

```python
"""
1. An algorithm is not a computer thing — it is a plan:
-> An algorithm is simply a step-by-step plan that explains how to solve a particular problem. You already use algorithms in your daily life without realizing it. For example, when you follow a cooking recipe, you perform a series of steps: boil water, add pasta, wait for 10 minutes, and drain the water. These steps form an algorithm because they guide you toward a specific result. In programming, an algorithm works in the same way: it tells us exactly what steps need to be performed to solve a problem. A good algorithm should have three important qualities: Precise means every step must be clear and understandable; Finite means the steps must eventually come to an end rather than continuing forever; and Correct means following those steps should produce the expected result and solve the problem properly.

2. Inputs and outputs:
-> An input-process-output (IPO) model explains how an algorithm works. First, it receives some information called input, then performs operations on that information through a series of steps called processing, and finally produces a result called output. For example, imagine you want to calculate the total cost of items in a shopping basket. The individual item prices are the input, adding all those prices together is the process, and the final shopping bill is the output. In simple terms, input means the information you provide, processing means what the program does with that information, and output means the result you receive. Not every algorithm needs external input, but this model is a useful starting point for understanding how programs work.

3. A program is written for a computer:
-> A program is a collection of instructions written in a programming language that tells a computer what to do and how to perform a particular task. A computer cannot directly understand ordinary human instructions such as a recipe written in English, so we use programming languages like Python to communicate instructions in a structured way that the computer can process and execute. For example, if you want a computer to calculate the total cost of your shopping, you write Python instructions that tell it how to perform that calculation. These instructions implement an algorithm and allow the computer to solve the problem. In this learning roadmap, we will use Python as our programming language.

4. Data is what the algorithm works on:
-> Data is the information that a program uses to perform a task, such as numbers, names, text, temperatures, or a list of expenses. An algorithm takes this information and performs operations on it to produce a result. For example, suppose you have four expenses: ₹100, ₹200, ₹150, and ₹50. Your algorithm will follow the steps required to add these numbers and calculate the total expense of ₹500. The important thing to understand is that the algorithm does not need to change when the input data changes. You can provide different expense amounts, and the same algorithm can calculate their total by following the same steps. Remember the difference: A program contains the instructions that tell the computer what to do, whereas data is the information on which those instructions operate.

5. Source code versus runtime behavior:
-> Source code is the actual Python instructions that a programmer writes and saves inside a .py file, whereas runtime behavior is what actually happens when the computer executes those instructions. Think of source code as a cooking recipe written on paper and runtime behavior as what happens when someone follows that recipe in the kitchen. For example, you might write Python code to calculate the average of four numbers. The written instructions are your source code, and the actual calculation and displayed result are its runtime behavior. Understanding this difference is important because code can look perfectly correct when you read it but still produce an incorrect result or an error when executed. This can happen because of incorrect input data or unexpected situations. Therefore, as a programmer, you must not only write code but also execute it, test it with different inputs, and check whether it behaves as expected.

6. How Python actually runs your instructions:
-> When you execute a Python file using a terminal command such as python my_program.py, the Python interpreter starts processing your program. First, it reads your Python source code and checks whether the instructions follow Python's grammar rules, which are called syntax. If you write something incorrectly according to Python's syntax rules, Python raises a SyntaxError, and the program cannot proceed with execution. If the syntax is valid, Python processes and executes the instructions, normally in order from top to bottom. Each instruction may perform a calculation, display something on the screen, change existing data, or produce an error. For example, if your program contains three print statements, Python normally executes the first, then the second, and finally the third. However, later you will learn about conditions and loops, which allow Python to skip certain instructions or repeat them instead of always following a simple top-to-bottom order.

-> In CPython, which is the standard Python implementation used in this learning roadmap, the execution process has three important stages: Processing, Compilation, and Execution. During processing, Python reads and checks your source code to understand whether it follows the language rules. During compilation, Python converts the valid code into an internal set of instructions called bytecode, which is a lower-level representation of your Python program. Finally, during execution, the Python Virtual Machine (PVM) executes that bytecode to perform the required operations. You do not need to understand the internal details of bytecode or the PVM right now; for now, remember that Python reads your code, checks it, compiles it into bytecode, and executes it to produce the expected result. You will study these internal mechanisms in more detail in the later topic Python Core Language and Data Types.
"""
```

```python
"""
This is your first Python program, and it contains a single instruction. The print() is a built-in Python function that tells the computer to display something on the screen. The text "Hello, this is my first program." is called a string, which means a piece of text or information (data). The quotation marks tell Python that the content inside them is text, not a command or instruction. When you run this program, the Python interpreter reads the instruction, executes the print() function, and displays the text on your screen. In simple terms, you are telling Python to show a particular message on the screen.
"""
print("Hello, this is my first program.")
```

```python
"""
This Python program demonstrates how to use an algorithm to calculate the total expenses from a list of numbers. First, expenses = [12, 45, 9, 20] creates a list containing four expense amounts: 12, 45, 9, and 20. The variable expenses stores this list so that we can access its values whenever needed. Next, total = 0 creates a variable named total and initializes it with zero because we have not added anything yet. After that, the program adds each expense amount to total, one by one. For example, expenses[0] accesses the first number (12), expenses[1] accesses the second number (45), expenses[2] accesses the third number (9), and expenses[3] accesses the fourth number (20). Python uses zero-based indexing, which means the first position in a list is numbered 0 instead of 1. Each time an expense is added, the updated result is stored back in the total variable. Finally, print(total) displays the calculated result on the screen.
"""

expenses = [12, 45, 9, 20]
total = 0
total = total + expenses[0]
total = total + expenses[1]
total = total + expenses[2]
total = total + expenses[3]
print(total)


"""
The main concept to understand: The algorithm follows a simple plan: start with zero, take each number from the list, add it to the running total, and display the final result. Even if you change the expense amounts, the same algorithm can calculate their total by following the same steps. This demonstrates how an algorithm (the steps), data (the expense amounts), and a program (the Python code) work together.
"""
```
