```python
empty_tuple = ()
coordinates = (3, 5)
colors = "red", "green", "blue"
print(colors)        # ('red', 'green', 'blue')
print(type(colors))  # <class 'tuple'>
```

```python
single_item = (5,)
not_a_tuple = (5)
print(type(single_item))  # <class 'tuple'>
print(type(not_a_tuple))  # <class 'int'>
```

```python
numbers = (10, 20, 10, 30)
print(numbers)
print(numbers[0])    # 10
print(numbers[-1])   # 30
print(numbers[1:3])  # (20, 10)
print(numbers[:])
```

```python
# Immutability: what can and cannot change:
point = (3, 5)
point[0] = 99  # TypeError: tuple does not support item assignment
```

```python
# Reassigning the variable is different: it points the name at a new tuple and is allowed:
point = (3, 5)
point = (99, 5)
print(point)  # (99, 5)
```

```python
container = ([1, 2],)
container[0].append(3)
print(container)  # ([1, 2, 3],)
print(container[0])
print(container[1]) # index position 1 is unavailable, so it's showing error
```

```python
# Packing and unpacking:
coordinates = 4, 9
x, y = coordinates
print(x, y)  # 4 9

# A mismatch such as x, y = (4, 9, 1) raises ValueError.
```

```python
a = 1
b = 2
a, b = b, a
print(a, b)  # 2 1
```

```python
def min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = min_max([4, 8, 1, 9])
print(lowest, highest)  # 1 9
```

```python
# Extended unpacking with *:
numbers = (1, 2, 3, 4, 5)
first, *rest = numbers
print(first, rest)  # 1 [2, 3, 4, 5]

first, *middle, last = numbers
print(first, middle, last)  # 1 [2, 3, 4] 5
```

```python
# Convert between lists and tuples:
coordinates_list = [3, 5]
coordinates_tuple = tuple(coordinates_list)
back_to_list = list(coordinates_tuple)

print(coordinates_tuple)  # (3, 5)
print(type(coordinates_tuple))

print(back_to_list)       # [3, 5]
print(type(back_to_list))
```
