```python
"""
1. Case methods (6 methods)"
These methods change the capitalization of characters.
"""
```

```python
# 1. capitalize()
# Purpose: Converts the first character to uppercase and the remaining characters to lowercase.
# When to use: When we wants a string to begin with a capital letter and have the rest lowercase.
name = "sOVON KUMAR"

result = name.capitalize()

print(result)
```

```python
# 2. casefold():
# Purpose: Converts a string into a normalized lowercase form for case-insensitive comparison.
# When to use: When comparing text without considering capitalization, especially with Unicode text.
name1 = "HELLO"
name2 = "hello"

print(name1.casefold())
print(name2.casefold())
print(name1.casefold() == name2.casefold())
```

```python
# 3. lower():
# Purpose: Converts all alphabetic characters to lowercase.
# When to use: Normalizing user input, email addresses, and text data.
name = "SOVON KUMAR"

print(name.lower())
```

```python
# 4. upper():
# Purpose: Converts all alphabetic characters to uppercase.
# When to use: Displaying text in uppercase or standardizing text for comparison.
name = "sovon kumar"

print(name.upper())
```

```python
# 5. swapcase():
# Purpose: Changes uppercase characters to lowercase and lowercase characters to uppercase.
# When to use: When you need to reverse the existing capitalization of text.
text = "Hello Python"

print(text.swapcase())
```

```python
# 6. title():
# Purpose: Converts the first character of each word to uppercase and the remaining characters to lowercase.
# When to use: Formatting simple titles and headings.
text = "welcome to python programming"

print(text.title())

# title() can capitalize unexpectedly around apostrophes.
# It treats the character after the apostrophe as the beginning of a new word.
# herefore, don't rely on title() for perfect human-name formatting.
text = "they're learning python"

print(text.title())
```

```python
"""
2. Trimming and padding methods (4 methods)
These methods remove unwanted characters or add padding to strings.
"""
```

```python
# 7. strip():
# Purpose: Removes specified characters from both the beginning and the end of a string.
# It removes any combination of the specified characters from both ends until it encounters a character outside that set.
text = "   Hello Python   "
print(text.strip())

text = "###Python###"
print(text.strip("#"))

text = "abcPythoncba"
print(text.strip("abc"))
```

```python
# 8. lstrip():
# Purpose: Removes specified characters from the left side (beginning) of a string.
# When to use: Removing leading spaces or unwanted prefixes made up of particular characters.
text = "   Python"
print(text.lstrip())

text = "###Python"
print(text.lstrip("#"))
```

```python
# 9. rstrip():
# Purpose: Removes specified characters from the right side (end) of a string.
# When to use: Removing trailing whitespace or unwanted ending characters.
text = "Python   "
print(text.rstrip())

text = "Python###"
print(text.rstrip("#"))
```

```python
# 10. zfill():
# Purpose: Adds zeros to the beginning of a string until it reaches the specified width.
# When to use: Formatting numeric identifiers, fixed-width codes, and sequence numbers
number = "42"
print(number.zfill(5))

number = "-42"
print(number.zfill(5))
```

```python
"""
3. Search and count methods (7 methods)
These methods help you locate characters or substrings and count their occurrences.
"""
```

```python
# 11. count()
# Purpose: Counts how many times a substring appears in a string.
text = "python is easy and python is powerful"
print(text.count("python"))

# counts how many times the character "a" appears in "banana" between index 0 and index 4 
text = "banana"
print(text.count("a", 0, 4))
```

```python
# 12. find():
# Purpose: Returns the index of the first occurrence of a substring; returns -1 if it is not found.
# When to use: Searching for text when the substring might not exist and we want to handle that situation without an exception.
text = "Hello Python"
print(text.find("Python"))
print(text.find("Java"))
```

```python
# 13. rfind():
# Purpose: Returns the index of the last occurrence of a substring; returns -1 if it is absent.
# When to use: Finding the last occurrence of a character or substring.
text = "python is easy and python is powerful"
print(text.rfind("python"))

```

```python
# 14. index()
# Purpose: Returns the index of the first occurrence of a substring; raises ValueError if it is not found.
# When to use: When we expect the substring to exist and want Python to raise an exception if it does not.
text = "Hello Python"
print(text.index("Python"))
print(text.index("Java"))
```

```python
# 15. rindex()
# Purpose: Returns the index of the last occurrence of a substring; raises ValueError if it is absent.
# When to use: Finding the last occurrence when absence should be treated as an error.
text = "python is easy and python is powerful"
print(text.rindex("python"))
```

```python
# 16. startswith():
# Purpose: Checks whether a string begins with a particular substring.
# When to use: Checking filename prefixes, URL prefixes, and text patterns.
filename = "data.csv"
print(filename.startswith("data"))
print(filename.startswith("image"))
```

```python
# 17. endswith():
# Purpose: Checks whether a string ends with a particular substring.
# When to use: Checking file extensions, suffixes, and text endings.
filename = "customers.csv"
print(filename.endswith(".csv"))
print(filename.endswith(".json"))
```

```python
'''
4. Replace and remove methods (2 methods)
These methods replace or remove specific text.
'''
```

```python
# 18. replace()
# Purpose: Replaces occurrences of one substring with another substring.
# When to use: Cleaning text, replacing values, and standardizing incoming data.
text = "I like Java"
result = text.replace("Java", "Python")
print(result)

text = "apple apple apple"
print(text.replace("apple", "orange", 2))
```

```python
# 19. removeprefix()
# Purpose: Removes the specified prefix only if the string starts with that exact prefix.
# When to use: Removing known prefixes from filenames, identifiers, and labels
filename = "backup_data.csv"
print(filename.removeprefix("backup_"))

# when the prefix is absent it also work:
filename = "data.csv"
print(filename.removeprefix("backup_"))
```

```python
"""
5. Split, join, and lines methods (4 methods)
These methods are extremely important in Data Engineering because you frequently process text files,
CSV data, logs, and records.
"""
```

```python
# 20. split()
# Purpose: Divides a string into pieces and returns them in a list.
# When to use: Breaking text into fields, processing records, and parsing simple text formats.
text = "Python is easy to learn"
result = text.split()
print(result)

data = "Rahul,25,India"
result = data.split(",")
print(result)

data = "apple,banana,orange"
print(data.split(",", 1))

print("a,,b,".split(","))
```

```python
# 21. rsplit()
# Purpose: Splits a string from the right side.
# When to use: When you need to split from the end, such as extracting the last part of a path or record.
text = "apple,banana,orange"
print(text.rsplit(",", 1))
text.split(",", 1) # compare with split
```

```python
# 22. splitlines()
# Purpose: Splits a string at line boundaries.
# When to use: Processing multiline text, log files, and text documents.
text = "Python is easy\nJava is popular\nC++ is powerful"
result = text.splitlines()
print(result)

text = "Hello\nWorld\n"
print(text.splitlines(keepends=True))
```

```python
# 23. join():
# Purpose: Combines multiple strings using the string on which join() is called as the separator.
# When to use: Combining strings, constructing CSV-like text, and creating formatted output.
words = ["Python", "is", "powerful"]
result = " ".join(words)
print(result)

names = ["Rahul", "Amit", "Sovon"]
print(", ".join(names))
```

```python
"""
6. Character testing methods (12 methods)
These methods check whether a string satisfies a particular character condition.
They return Boolean values: True or False.
"""
```

```python
# 24. isalpha()
# Purpose: Checks whether all characters are alphabetic and the string is not empty.
# When to use: Checking whether input contains only alphabetic characters.
text = "Python"
print(text.isalpha())

print("Python Programming".isalpha())
print("AI-document23@56-processing99#@65".isalpha())
```

```python
# 25. isdigit()
# Purpose: Checks whether all characters are digits according to Python's Unicode digit definition.
# When to use: Checking digit-only input, while remembering that it does not validate decimal numbers.
text = "12345"
print(text.isdigit())

print("123.45".isdigit())
```

```python
# 26. isdecimal()
# Purpose: Checks whether all characters are decimal characters.
# When to use: Checking decimal-digit characters specifically.
text = "12345"
print(text.isdecimal())

# Important difference: isdecimal() is stricter than isdigit().
text = "²"
print(text.isdecimal())
print(text.isdigit())
print(text.isnumeric())
```

```python
# 27. isnumeric()
# Purpose: Checks whether all characters are numeric characters, 
# including a broader range of Unicode numeric characters.
# When to use: Checking Unicode numeric characters beyond ordinary decimal digits.
text = "12345"
print(text.isnumeric())

text = "½"
print(text.isnumeric())
```

```python
# 28. isalnum():
# Purpose: Checks whether all characters are alphabetic or numeric, and the string is not empty.
# When to use: Checking alphanumeric identifiers or input that must not contain spaces or punctuation.
text = "Python123"
print(text.isalnum())

text = "Python 123"
print(text.isalnum())
```

```python
# 29. isspace()
# Purpose: Checks whether all characters are whitespace characters and the string is not empty.
# When to use: Detecting blank or whitespace-only input.
text = "   "
print(text.isspace())

text = "Python "
print(text.isspace())
```

```python
# 30. isupper()
# Purpose: Checks whether all cased characters are uppercase and at least one cased character exists.
# When to use: Checking whether text is uppercase.
text = "PYTHON"
print(text.isupper())
print("PYTHON123".isupper())
```

```python
# 31. islower()
# Purpose: Checks whether all cased characters are lowercase and at least one cased character exists.
# When to use: Checking whether text is lowercase.
text = "python"
print(text.islower())
print("python123".islower())
```

```python
# 32. istitle()
# Purpose: Checks whether the string follows title-case rules.
# When to use: Checking whether text follows title-case formatting.
text = "Hello World"
print(text.istitle())

text = "hello world"
print(text.istitle())
```

```python
# 33. isidentifier()
# Purpose: Checks whether a string is syntactically valid as a Python identifier.
text = "student_name"
print(text.isidentifier())

print("123student".isidentifier())
print("class".isidentifier())
```

```python
# 34. isascii()
# Purpose: Checks whether all characters belong to the ASCII character set.
# When to use: Checking whether text contains only ASCII characters.
text = "Python123"
print(text.isascii())


text = "café"
print(text.isascii())
```

```python
# 35. isprintable()
# Purpose: Checks whether all characters are printable, or the string is empty.
# When to use: Checking whether text contains non-printable characters before displaying or processing it.
text = "Hello Python"
print(text.isprintable())

text = "Hello\nPython" # The newline character is not considered printable.
print(text.isprintable())
```

```python
"""
7. Alignment and formatting methods (6 methods)
These methods help you format strings, align text, and insert values.
"""
```

```python
# 36. center()
# Purpose: Centers a string within a specified width.
# When to use: Creating centered headings and formatted text output.
text = "Python"
print(text.center(12))

print(text.center(12, "*")) # we can specify a fill character
```

```python
# 37. ljust()
# Purpose: Left-aligns a string within a specified width.
# When to use: Formatting left-aligned columns in text output.
text = "Python"
print(text.ljust(12, "*"))
```

```python
# 38. rjust()
# Purpose: Right-aligns a string within a specified width.
# When to use: Formatting right-aligned columns and identifiers.
text = "Python"
print(text.rjust(12, "*"))
```

```python
# 39. format()
# Purpose: Inserts values into replacement fields in a string.
name = "Sovon"
age = 25

result = "My name is {} and I am {} years old.".format(name, age)

print(result)

# example 02:
result = "Name: {name}, Age: {age}".format(
    name="Sovon22",
    age=254
)

print(result)
```

```python
# 40. format_map()
# Purpose: Formats a string using a mapping, such as a dictionary, to supply named values.
# When to use: Formatting strings directly from dictionaries or other mapping objects.
student = {
    "name": "Sovon",
    "age": 25
}

result = "Name: {name}, Age: {age}".format_map(student)

print(result)
```

```python
"""
8. Partition methods (2 methods)
These methods divide a string into three parts based on a separator.
"""
```

```python
# 41. partition()
"""
Purpose: Splits a string at the first occurrence of a separator and returns three parts:
1. Text before the separator.
2. The separator itself.
3. Text after the separator.
"""
# When to use: Extracting a key and value from simple key-value text.
text = "name=Rahul=Developer"
result = text.partition("=")
print(result)

# If the separator does not exist:
print("Python".partition("="))
```

```python
# 42. rpartition()
# Purpose: Splits a string at the last occurrence of a separator and returns three parts.
# When to use: Extracting the final part of a string based on a separator.
text = "name=Rahul=Developer"
result = text.rpartition("=")
print(result)

# If the separator does not exist:
print("Python".rpartition("="))
```

```python
"""
9. Translation and tab methods (3 methods)
These methods help replace multiple characters and format tab characters.
"""
```

```python
# 43. maketrans()
# Purpose: Creates a translation table that maps characters to replacement characters.
# When to use: Creating character-replacement or character-deletion rules.
table = str.maketrans("abc", "123")
print(table)

# Another form allows to specify characters to delete:
table = str.maketrans("", "", "aeiou")
print(table)
```

```python
# 44. translate()
# Purpose: Applies a translation table to a string.
# When to use: Replacing or removing multiple individual characters efficiently.
table = str.maketrans("abc", "123")

text = "abc cab"

result = text.translate(table)

print(result)

# Example of deleting characters:
table = str.maketrans("", "", "aeiou")

text = "education"

print(text.translate(table))
```

```python
# 45. expandtabs()
# Purpose: Replaces tab characters (\t) with spaces according to tab stops.
# When to use: Formatting tab-separated text and aligning text columns.
text = "Name\tAge"
print(text.expandtabs(8))

text = "A\tB"
print(repr(text.expandtabs(4)))
```

```python
# 10. Encoding method (1 method)
# 46. encode()
# Purpose: Converts a Python string (str) into bytes (bytes) using a specified character encoding.
# By default, Python uses UTF-8.
text = "Hello"

result = text.encode()

print(result)
print(type(result))


# Example with a non-ASCII character:
text = "café"

result = text.encode("utf-8")

print(result)
```

```python
text = "Hello"

encoded = text.encode("utf-8")
decoded = encoded.decode("utf-8")

print(decoded)
```

```python
# 11. Important additional concept: repr()
# When to use: Debugging strings, identifying hidden whitespace, and inspecting newline or tab characters.
text = "  Hello Python  "

print(text)
print(repr(text))


# example 02:
text = "Hello\nPython"

print(text)
print(repr(text))
```
