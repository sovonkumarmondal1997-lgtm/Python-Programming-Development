```python
# Boolean operators: and, or, not:
"""
and: both conditions must be true.
or: at least one condition must be true.
not: reverses a Boolean value.
"""

has_ticket = True
is_on_time = False
print(has_ticket and is_on_time)  # False
print(has_ticket or is_on_time)   # True
print(not is_on_time)             # True
```

```python
# Python's and and or operators return operands, not necessarily Boolean values (True or False).
# Python evaluates 0 as falsy, so the and operator immediately returns 0 without evaluating "hello"
print(0 and "hello")          # 0

# Python evaluates "hello" as truthy, so the and operator evaluates the second operand and returns "world"
print("hello" and "world")    # world

# Python evaluates 0 as falsy, so the or operator evaluates the second operand and returns "fallback"
print(0 or "fallback")        # fallback

# Python evaluates "hello" as truthy, so the or operator immediately returns "hello" without evaluating "fallback".
print("hello" or "fallback")  # hello
```

```python
# Membership operators:
print("cat" in "concatenate")       # True
print("dog" in "concatenate")       # False
print("dog" not in "concatenate")   # True
```

```python
# Identity operators and None:
score = None
if score is None:
    print("No score has been recorded yet.")
```
