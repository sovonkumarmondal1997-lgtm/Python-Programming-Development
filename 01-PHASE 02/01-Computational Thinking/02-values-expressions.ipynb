```python
"""
# Why Values, Expressions, Statements, and Variables matters:
-> These four concepts are the building blocks of Python: values are the data, expressions produce results, statements give instructions, and variables store references to data.

# Key_terms:
- Value: A piece of data, such as a number, text, or list.
- Expression: Python code that produces a value when evaluated.
- Statement: A Python instruction that performs an action.
- Variable: A name used to refer to a value so you can use it later.
- Assignment: Giving a value to a variable using =.
- Reassignment: Changing a variable so it refers to a different value.
- Mutation: Changing the contents of an existing object, such as adding an item to a list.
"""
```

```python
# 1. Values: The raw material:
# A value is a piece of information that a Python program can use. For example, 42 is a whole number, 3.14 is a decimal number, "hello" is text, True is a Boolean value, and [1, 2, 3] is a list containing numbers. These are all values, but they represent different kinds of information. Every value has a type, which tells Python what kind of data it is and what operations can be performed on it. Think of values as the raw materials that a Python program works with.

42            # a whole number (int)
3.14          # a number with a decimal point (float)
"hello"       # text (a string)
True          # a Boolean value: True or False
[1, 2, 3]     # a list of values

```

```python
# 2. Expressions: Recipes that produce a value
# An expression is a piece of Python code that produces a value when Python evaluates it. For example, 3 + 4 is an expression that produces 7, and "a" + "b" produces "ab". Expressions can be simple or complex. A single value, such as 42, is also technically an expression because it evaluates to itself. Think of an expression as a small calculation or recipe: you give Python an instruction to calculate something, and Python produces a result.


3 + 4
"a" + "b" 
```

```python
# 3. Statements: Complete instructions
# A statement is a complete instruction that tells Python to perform an action in a program. For example, total = 3 + 4 tells Python to calculate 3 + 4 and assign the result to the name total. Similarly, print(total) tells Python to display the value of total. The important difference is that an expression produces a value, while a statement performs a complete action using that value or other instructions. For example, if you write 3 + 4 alone in a Python script, Python calculates 7, but the result is not automatically saved or displayed. You need another instruction, such as print(3 + 4), to display it.

total = 3 + 4     # an assignment statement (contains the expression 3 + 4)
print(total)      # a statement that calls the print instruction
```

```python
# 4. Variables and assignment: Naming a value
# A variable is a name that refers to a value so that you can use that value again without typing it repeatedly. For example, in age = 25, Python connects the name age to the value 25. This process is called assignment, and the = symbol is used to perform it. Read age = 25 as "age is assigned the value 25", not "age equals 25". In Python, = assigns a value, while == checks whether two values are equal. Once you assign a value to a variable, you can use its name in other calculations. For example, next_year_age = age + 1 uses the value referred to by age, adds 1, and assigns the result (26) to next_year_age.

age = 25
next_year_age = age + 1
print(next_year_age)   # 26
```

```python
# 5. Reassignment: Pointing the name somewhere new
# Reassignment means giving an existing variable name a different value. For example, when you write score = 10, the name score refers to the value 10. Later, if you write score = 20, the same name now refers to 20 instead of 10. The original number 10 has not been changed; only the value that the variable name refers to has changed. Think of a variable as a label: you can move the same label from one object to another. This helps you understand that a variable is not permanently connected to its first assigned value

score = 10
print(score)   # 10
score = 20
print(score)   # 20
```

```python
# 6. Mutation: Changing a value without reassigning the name
# Mutation means changing the contents of an existing object without assigning a new object to the variable. For example, scores = [10, 20, 30] creates a list, and scores.append(40) adds 40 to that same list. The variable scores still refers to the same list, but the list's contents have changed to [10, 20, 30, 40]. This is different from reassignment because we did not write scores = ... again. However, not all Python values can be changed in this way. Lists are mutable, meaning their contents can be changed after creation, while numbers and strings are immutable, meaning they cannot be changed in place. For example, name.upper() creates and returns a new uppercase string, "SAM", but it does not change the original value "sam". To store the uppercase result, you must write name = name.upper(). The key difference is that mutation changes an existing object, whereas reassignment makes a variable name refer to a different object.

scores = [10, 20, 30]
scores.append(40)
print(scores)   # [10, 20, 30, 40]


name = "sam"
name.upper()
print(name)   # still "sam" — upper() did NOT change name

name = "sam"
name = name.upper()
print(name)
```

```python
# This program calculates the total cost of buying 3 items, where each item costs 19.99, and then displays the final amount on the screen. First, price = 19.99 assigns the value 19.99 to the variable price. This is a decimal number, called a float in Python. Next, quantity = 3 assigns the whole number 3 to the variable quantity, which is an integer (int). Then, total_cost = price * quantity tells Python to multiply the price by the quantity. Python takes the values of both variables (19.99 and 3), multiplies them, and gets 59.97. This calculation, price * quantity, is an expression because it produces a value. Python then assigns the result 59.97 to the variable total_cost. Finally, print(total_cost) displays the calculated result on the screen.

price = 19.99
quantity = 3
total_cost = price * quantity
print(total_cost)


# In simple terms: The program takes the price, multiplies it by the quantity, stores the total cost, and displays the result.
```

```python
# This program demonstrates reassignment by calculating how much money remains in a bank balance after making two withdrawals. First, balance = 100 creates a variable named balance and assigns it the value 100. Next, balance = balance - 30 tells Python to take the current balance (100), subtract 30, and assign the new result (70) back to the same variable balance. Then, balance = balance - 15 takes the updated balance (70), subtracts another 15, and reassigns the result (55) to balance. Finally, print(balance) displays the remaining balance, which is 55.

# The important thing to understand is that balance = balance - 30 is not a mathematical equality. Python first calculates the right-hand side (100 - 30) and then assigns the result (70) to the variable on the left. Each reassignment uses the variable's latest value.

balance = 100
balance = balance - 30
balance = balance - 15
print(balance)

#In simple terms: The program starts with 100, subtracts 30, subtracts another 15, and displays the remaining balance of 55.
```

```python
# This program demonstrates the difference between mutation and reassignment using a Python list. First, password_attempts = [] creates an empty list and assigns it to the variable password_attempts. Next, password_attempts.append("qwerty") adds "qwerty" to the list, and password_attempts.append("letmein") adds another item, "letmein". These operations are called mutation because they change the contents of the existing list without creating a new list or reassigning the variable. When print(password_attempts) runs, it displays ['qwerty', 'letmein'].

# Next, backup_attempts = password_attempts creates another variable name, backup_attempts, that refers to the exact same list as password_attempts. Python does not create a copy of the list here. Both variables point to one shared list. Then, backup_attempts.append("hunter2") adds a third item to that shared list. Because both variables refer to the same list, the change is visible through either variable. Therefore, print(password_attempts) displays ['qwerty', 'letmein', 'hunter2'], even though we added the new item using backup_attempts.

# Finally, password_attempts = ["reset"] performs reassignment. It creates a brand-new list containing "reset" and makes password_attempts refer to this new list. It does not change or delete the original three-item list. The variable backup_attempts still refers to the original list because we never reassigned it. Therefore, the final two print() statements display different lists: password_attempts displays ['reset'], while backup_attempts displays ['qwerty', 'letmein', 'hunter2'].

password_attempts = []

password_attempts.append("qwerty")
password_attempts.append("letmein")
print(password_attempts)          # ['qwerty', 'letmein']

backup_attempts = password_attempts
backup_attempts.append("hunter2")
print(password_attempts)          # ['qwerty', 'letmein', 'hunter2']

password_attempts = ["reset"]
print(password_attempts)          # ['reset']
print(backup_attempts)            # ['qwerty', 'letmein', 'hunter2']


# The most important concept to understand: Mutation changes the contents of an existing list, so every variable referring to that list sees the change. Reassignment makes a variable refer to a different list, without changing what other variables refer to.
```
