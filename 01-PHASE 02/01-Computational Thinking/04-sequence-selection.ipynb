```python
# Counting positive, negative, and zero values:
numbers = [10, -5, 0, 3, -1, 0, 7, 2, -4, -9, 6, 0, 8, -2, 1, -3, 4, 5, -6, 0, 9]

positive_count = 0
negative_count = 0
zero_count = 0

for number in numbers:
    if number > 0:
        positive_count = positive_count + 1
    elif number < 0:
        negative_count = negative_count + 1
    else:
        zero_count = zero_count + 1

print("Positive numbers:", positive_count)
print("Negative numbers:", negative_count)
print("Zero numbers:", zero_count)
```

```python
# Exercise 4 — Display a traffic-light instruction:
colors = ["green", "yellow", "red", "green"]

for color in colors:
    if color == "red":
        print(color, "-> Stop")
    elif color == "yellow":
        print(color, "-> Prepare to stop")
    elif color == "green":
        print(color, "-> Go")
    else:
        print(color, "-> Unknown color")
```
