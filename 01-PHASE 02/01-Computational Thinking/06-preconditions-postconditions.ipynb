```
Topic: Preconditions, Postconditions, and Edge Cases

Preconditions, Postconditions, and Edge Cases teach us to check what must be true before code runs, what must be true after it finishes, and how to handle unusual inputs without breaking the program.
```

```
Topic: Key Terms

1. Precondition: A condition that must be satisfied before a function can work correctly.
2. Postcondition: A condition or result that must be true after a function finishes successfully.
3. Invariant: A rule or condition that must always remain true while a program is running.
4. Normal Case: A typical input that a function is expected to handle correctly.
5. Edge Case: An unusual input that may expose hidden bugs in a program.
6. Invalid Input: An input that does not follow the requirements of a function, such as providing text instead of a number.
7. Empty Input: An input that contains no data or elements, such as an empty list or empty string.
8. Duplicate: A value that appears more than once in a collection, such as [10, 20, 10].
9. Boundary Value: A value at the minimum or maximum limit of the allowed input range.
```

```
Topics: 01. Preconditions: What must be true before you start:
- A precondition is something that must be true before a function runs correctly.

def average(numbers):
    return sum(numbers) / len(numbers)

- Precondition: numbers must contain at least one number.
- Why? An empty list causes division by zero.

Remember: Preconditions define what a function expects before execution.
```

```
Topic: 02. Postconditions: What you promise after you finish
- A postcondition defines what result a function guarantees after it finishes successfully.

Example:
def sorted_ascending(numbers):    
    return sorted(numbers)
print(sorted_ascending([5, 2, 8, 1]))

Output:
[1, 2, 5, 8]

- Postcondition: The returned list must contain the same numbers arranged from smallest to largest.

Remember: Postconditions define what a function promises after execution.
```

```
Topic: 03. Invariants: What must never stop being true
- An invariant is a condition that must remain true throughout the execution of a program or operation.

Example:
balance = 1000
balance = balance - 200
balance = balance + 100

- Invariant: If no money is added or removed from the account externally, the balance must follow the account's transaction rules and must not become negative (assuming overdrafts are not allowed).

- Another example: During a bank transfer, the combined money in both accounts must remain unchanged.

Remember: Invariants define the rules that must always remain true.
```

```
Topic: 04. Edge Cases: Where bugs hide
- Edge cases are unusual inputs that might cause a function to behave incorrectly or crash.

There are four important categories:
A. Empty Input
- An input containing no elements.

Example:
numbers = []
print(len(numbers))

Output:
0

Question: What happens if you calculate the average of an empty list?
B. Invalid Input
- An input that does not match what a function expects.

Example:
age = "twenty"
- If a function expects a number, providing text creates invalid input.

C. Duplicates
- When the same value appears multiple times in a collection.

Example:
numbers = [10, 20, 10, 30, 20]
- Here, 10 and 20 are duplicates.

Question: Will your function handle repeated values correctly?

D. Boundary Values
- Values at the minimum or maximum limits of an allowed range, including values just outside those limits.

Example:

Suppose a system accepts ages from 0 to 120.

Age	         Meaning
-1	         Below minimum (invalid)
0	         Minimum boundary
25	         Normal value
120	         Maximum boundary
121	         Above maximum (invalid)

Remember: Always test the minimum, maximum, and values just outside the allowed range.
```

```
Example 1 — A function with an unstated precondition and the crash it causes
1. Understand the code
def average(numbers):
    total = 0    
    for number in numbers:        
        total = total + number    
    return total / len(numbers)
    
print(average([10, 20, 30]))
print(average([]))

2. How does the code work?
First function call:
average([10, 20, 30])

1. total = 0 → Initializes the total.
2. The for loop adds all numbers: 10 + 20 + 30.
3. total becomes 60.
4. len(numbers) returns 3.
5. 60 / 3 returns 20.0.

Output: 20.0

Second function call:
average([])

1. total = 0 → Initializes the total.
2. The for loop does not execute because the list is empty.
3. total remains 0.
4. len(numbers) returns 0.
5. Python tries to calculate 0 / 0.

Error: ZeroDivisionError

Python cannot divide a number by zero, so the program crashes.
```

```
3. What are we learning from this example?

Concept	                 Explanation
Hidden precondition	     The function assumes that the list contains at least one number.
Normal case	             [10, 20, 30] works correctly.
Edge case	             [] causes the function to crash.
Problem	                 The function does not check whether the list is empty.
```

```
4. Important programming lesson:

Before writing a function, always ask:
- What input does my function expect?
- What happens if the input is empty?
- Can the input cause an error?
- Have I handled the unusual cases?

Key takeaway: A function may work perfectly with normal input but crash with unusual input when its preconditions are not satisfied.
```

```
Example 2 — Making the precondition explicit and defining a postcondition:

def average(numbers):
    if len(numbers) == 0:
        return None

    total = 0
    for number in numbers:
        total = total + number

    return total / len(numbers)


print(average([10, 20, 30]))  # 20.0
print(average([]))            # None

2. How does the code work:
- First function call:
average([10, 20, 30])

- len(numbers) == 0 → Checks whether the list is empty.
- The condition is False because the list contains 3 numbers.
- The loop adds all numbers: 10 + 20 + 30 = 60.
- 60 / 3 = 20.0.

Output: 20.0

Second function call:
average([])

- len(numbers) == 0 → Checks whether the list is empty.
- The condition is True.
- return None executes immediately.
- The function stops without performing division.

Output: None
```

```
3. What are we learning from this example:

- Explicit checking: The function checks whether the list is empty.
- None: Represents the absence of an average when the list is empty.
- Postcondition: Returns the numeric average for a non-empty list or None for an empty list.
- Trade-off: The caller must check for None before using the result as a number.
- Edge-case handling: The function no longer crashes when it receives an empty list.
```

```
4. Understanding the trade-off

Suppose you write:
result = average([])
print(result + 10)

- This will cause a TypeError because None cannot be added to an integer.

Therefore, check the result first:
result = average([])

if result is not None:    
    print(result + 10)
else:   
    print("Average is not available")

Output:
Average is not available

5. Important programming lesson
- Normal input should produce the expected result.
- Empty input should have a clearly defined behavior.
- Never assume a function always returns a number if it can also return None.

Key takeaway: Instead of allowing unexpected input to crash your function, explicitly handle it and define what the function should return.
```

```
Example 3 — A function tested against a full checklist of edge cases
1. Understand the code
The purpose of this function is to check whether a password meets the specified length and data-type rules.
```

````python
def validate_password(password):
    if not isinstance(password, str):
        return "invalid: not text"
    if len(password) == 0:
        return "invalid: empty"
    if len(password) < 8:
        return "invalid: too short"
    if len(password) > 64:
        return "invalid: too long"
    return "valid"
````

```
2. How does the function work?
The function checks the password step by step:

Code	                    What it checks
isinstance(password, str)	Is the input text (string)?
len(password) == 0	        Is the password empty?
len(password) < 8	        Is it shorter than 8 characters?
len(password) > 64	        Is it longer than 64 characters?
return "valid"	            If all checks pass, return valid.

Important: Understanding isinstance()
```

````python
isinstance("hello", str)  # True
isinstance(123, str)      # False
````

```
It checks whether a value belongs to a particular data type.
3. Understanding the test cases
```

````python
test_cases = [
    "hunter22",
    "",
    "short",
    "a" * 8,
    "a" * 64,
    "a" * 65,
    12345678,
]
````

```
Each test case represents a different situation.

Test input	             Category	                Expected result
"hunter22"	            Normal case	                    valid
""	                    Empty input	                invalid: empty
"short"	              Below minimum boundary	    invalid: too short
"a" * 8	              Minimum boundary	                valid
"a" * 64	          Maximum boundary	                valid
"a" * 65	          Above maximum boundary	    invalid: too long
12345678	           Invalid input	            invalid: not text

What does "a" * 8 mean?

Python repeats the string "a" eight times.
```

````python
print("a" * 8)
````

```
Similarly, "a" * 64 creates a string containing 64 characters.

4. Understanding the for loop and repr()
```

````python
for case in test_cases:
    print(repr(case), "->", validate_password(case))
````

```
- for case in test_cases: → Takes each test case one by one.
- repr(case) → Displays the value in a way that makes its representation clear.
- validate_password(case) → Sends that test case to the function.
- print() → Displays the input and its result.

Why use repr()?
```

````python
print(repr(""))
print(repr(12345678))
````

```
It helps distinguish an empty string from other values.

5. Understanding the final output:
'hunter22' -> valid
'' -> invalid: empty
'short' -> invalid: too short
'aaaaaaaa' -> valid
'aaaaaaaa...(64 characters)' -> valid
'aaaaaaaa...(65 characters)' -> invalid: too long
12345678 -> invalid: not text

The important point is that every input produces a defined response instead of an unexpected crash.

6. Important programming lessons

- Normal case: Test ordinary, expected inputs.
- Empty input: Check what happens when no data is provided.
- Invalid input: Check whether the input has the correct data type.
- Boundary values: Test the minimum and maximum allowed values.
- Postcondition: The function returns a defined text message for each tested input.
- Edge-case testing: Test unusual inputs deliberately, not just normal inputs.

- Important security note: This function only checks data type and length. It does not determine whether a password is secure. Never print or log real passwords.

Key takeaway: Before declaring a function complete, test normal inputs, empty inputs, invalid inputs, duplicates where relevant, and boundary values to ensure it behaves as expected.
```
