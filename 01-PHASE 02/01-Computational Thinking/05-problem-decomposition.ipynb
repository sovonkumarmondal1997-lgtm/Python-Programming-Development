```python
#  Example 3 — A decomposed password report
"""
This program checks a list of passwords against two conditions:
1. Does the password contain at least 8 characters?
2. Does the password contain at least one digit (0-9)?
"""
                                                 
def has_minimum_length(password):
    return len(password) >= 8

def has_a_digit(password):
    for character in password:
        if character.isdigit():
            return True
    return False

def check_password(password):
    if not has_minimum_length(password):
        return "too short"
    if not has_a_digit(password):
        return "needs a digit"
    return "This Password is Valid and Strong"

def print_password_report(passwords):
    for password in passwords:
        result = check_password(password)
        print(password, "->", result)

passwords_to_check = ["hunter2", "abc", "correcthorsebattery9", "adsfshdjty"]
print_password_report(passwords_to_check)
```

```python
# Exercise 2 — Implement and compose the functions:
def read_scores():
    scores = []

    number_of_students = int(input("How many student's scores wants to enter: "))

    for student_count in range(number_of_students):
        score = int(input("Please enter the exam score: "))
        scores.append(score)  

    return scores

def count_pass(scores):
    pass_count = 0

    for score in scores:
        if score >= 60:
            pass_count = pass_count + 1

    return pass_count

def calculate_average(scores):
    if len(scores) == 0:
        return 0
    
    total = 0

    for score in scores:
        total = total + score

    return total / len(scores)

def print_report(scores, pass_count, average):
    print("Exam Report")
    print("Number of students:", len(scores))
    print("Students passed:", pass_count)
    print("Class average:", average)


def main():
    scores = read_scores()
    pass_count = count_pass(scores)
    average = calculate_average(scores)
    print_report(scores, pass_count, average)
```

```python
main()
```

```python
# Exercise 3 — Add a no-spaces password rule:
def has_no_spaces(password):
    for character in password:
        if character == " ":
            return False
    return True


def has_minimum_length(password):
    return len(password) >= 8


def has_a_digit(password):
    for character in password:
        if character.isdigit():
            return True
    return False


def check_password(password):
    if not has_minimum_length(password):
        return "too short"
    if not has_a_digit(password):
        return "needs a digit"
    if not has_no_spaces(password):
        return "must not contain spaces"
    return "OK"


print(has_no_spaces("secure9pass"))
print(has_no_spaces("bad pass9"))
print(check_password("secure9pass"))
print(check_password("bad pass9"))
```
