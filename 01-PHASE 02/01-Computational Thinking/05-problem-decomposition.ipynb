```
Topic: Problem Decomposition

Problem Decomposition: Breaking one big programming problem into smaller, manageable tasks, where each task is handled by a separate function with one clear responsibility.
```

```
- Decomposition: Breaking one big problem into smaller, easier-to-solve parts.
- Responsibility: A specific job that a function is designed to perform.
- Single Responsibility: A function should do only one clear job.
- Composition (of Functions): Combining multiple small functions to perform a bigger task.
- Coupling: How much one function or piece of code depends on another.
```

```
Step-by-step explanation:

1. Why "one giant block of code" becomes a problem:

- Writing everything in one large block of code makes a program difficult to understand, test, reuse, and fix.

Example:

Imagine an expense management program that performs four tasks:
- Reads expenses.
- Validates expenses.
- Calculates the total.
- Prints a report.

If all four tasks are written together without separate functions:
- Difficult to understand: You must read the entire code to find a particular operation.
- Difficult to test: You cannot easily test the calculation separately.
- Difficult to reuse: You may need to copy and paste code to use the validation logic elsewhere.
- Difficult to debug: A bug in one part can affect other parts of the program.

Main idea: Keep different tasks separate instead of writing everything in one large block.
```

```
2. Decomposition: splitting one problem into small responsibilities:

- Decomposition means identifying the different tasks inside a big problem and creating separate functions to handle them.

Example: Expense management program

Instead of writing everything together, divide it into four functions:

Function	                                 Responsibility
read_expenses()	                             Collect expense data.
validate_expenses()	                     Check whether expenses are valid.
calculate_total()	                     Calculate the total expenses.
print_report()	                             Display the expense report.

Connection to Input → Process → Output:
- Input: Collect expense data.
- Process: Validate expenses and calculate totals.
- Output: Display the report.

Main idea: Break a big problem into smaller tasks, and give each task its own function.
```

```
3. Giving each responsibility a function with one clear job:

- Each function should perform one specific job, and its name should clearly describe that job.

Example: Bad approach
```

````python
def calculate_total_and_print_report(expenses):
    total = sum(expenses)
    print("Total:", total)
````

```
Problem: This function performs two different jobs:
1. Calculates the total.
2. Prints the report.

Better approach:
```

````python
def calculate_total(expenses):
    return sum(expenses)


def print_report(total):
    print("Total:", total)
````

```
Now:
- calculate_total() only calculates the total.
- print_report() only prints the report.
Main idea: One function = One clear responsibility.
```

```
4. Composing small functions into a full program:

- Composition means combining multiple small functions by calling them in the correct order to complete a bigger task.

Example:
```

````python
def read_expenses():
    return [100, 200, 300]


def calculate_total(expenses):
    return sum(expenses)


def print_report(total):
    print("Total expenses:", total)


# Main program
expenses = read_expenses()
total = calculate_total(expenses)
print_report(total)
````

```
Example 1 — One small responsibility
```

````python
def total_of(expenses):
    total = 0
    for expense in expenses:
        total = total + expense
    return total


expenses = [12, 45, 9, 20]
print(total_of(expenses))
````

```
Explanation:

Step 1: Define the function
def total_of(expenses):
Creates a function named total_of() that accepts a list of expenses.

Step 2: Initialize the total
total = 0
Creates a variable total and starts its value at 0.

Step 3: Loop through expenses
for expense in expenses:
Takes each expense from the list, one at a time.

Step 4: Add each expense
total = total + expense
Adds the current expense to the running total.

For example:

Expense	                    Running total
12	                              12
45	                              57
9	                              66
20	                              86

Step 5: Return the result
return total
Sends the final calculated total (86) back to wherever the function was called.

Step 6: Create the expense list
expenses = [12, 45, 9, 20]
Stores four expense values in a list.

Step 7: Call the function
print(total_of(expenses))
Passes the expense list to total_of() and prints the returned result.
```

```
4. Important concepts to understand:

- Single responsibility: total_of() performs only one job: calculating the total.
- Running total: Continuously adds new values to the previous total.
- Return: Sends the calculated result back to the caller.
- Reusability: You can use total_of() with different expense lists.
- Testability: You can test this function independently without running the entire program.
```

````python
# Example of independent testing:
print(total_of([1, 2, 3]))
````

```
Example 2 — Two responsibilities, kept separate

- In this example, we use three separate functions to calculate total expenses, calculate average expenses, and generate a spending summary.
```

````python
def total_of(expenses):
    total = 0
    for expense in expenses:
        total = total + expense
    return total


def average_of(expenses):
    if len(expenses) == 0:
        return 0
    return total_of(expenses) / len(expenses)


def describe_spending(expenses):
    total = total_of(expenses)
    average = average_of(expenses)
    return "Total: " + str(total) + ", Average: " + str(average)


expenses = [12, 45, 9, 20]
print(describe_spending(expenses))
````

```
2. Explanation of each function:

Step 1: total_of(expenses)
def total_of(expenses):    
    total = 0    
    for expense in expenses:        
        total = total + expense    
    return total

- Calculates the sum of all expenses and returns the result.

Example:
total_of([12, 45, 9, 20])

Output: 86

Remember: Its only responsibility is calculating the total.

Step 2: average_of(expenses)
def average_of(expenses):    
    if len(expenses) == 0:        
        return 0    
    return total_of(expenses) / len(expenses)

- Calculates the average expense by dividing the total expenses by the number of expenses.

Let's understand two important things:

A. Checking for an empty list
if len(expenses) == 0:
    return 0

- len(expenses) counts the number of items.
- If the list is empty, its length is 0.
- The function returns 0 instead of attempting division by zero.

B. Calculating the average
return total_of(expenses) / len(expenses)

- total_of(expenses) calculates the total (86).
- len(expenses) counts the expenses (4).
- 86 / 4 = 21.5.

Remember: Its only responsibility is calculating the average. It reuses total_of() instead of writing the addition logic again.
```

```
Step 3: describe_spending(expenses)
def describe_spending(expenses):
    total = total_of(expenses)    
    average = average_of(expenses)    
    return "Total: " + str(total) + ", Average: " + str(average)

- Creates a human-readable spending summary using the total and average calculated by the other functions.

A. Calling other functions:
total = total_of(expenses)average = average_of(expenses)

- Calls total_of() to get the total.
- Calls average_of() to get the average.
- Stores both returned results in variables.

B. Converting numbers into strings:
str(total)
str(average)

- str() converts numbers into text so Python can combine them with other text using +.

For example:
str(86)       # "86"
str(21.5)     # "21.5"

C. Returning the summary
return "Total: " + str(total) + ", Average: " + str(average)

-VCombines everything into one readable message.

Remember: Its only responsibility is preparing the spending summary, not calculating the numbers itself.
```

```
Example 3 — A small decomposed program: Password Checker Report:

In this example, we build a password checker that checks multiple passwords against two rules:
1. Password must contain at least 8 characters.
2. Password must contain at least one digit (0–9).

We divide the program into four separate functions, each having one responsibility.
```

````python
def has_minimum_length(password):
    return len(password) >= 8


def has_a_digit(password):
    for character in password:
        if character.isdigit():
            return True
    return False


def check_password(password):
    if not has_minimum_length(password):
        return "too short"
    if not has_a_digit(password):
        return "needs a digit"
    return "OK"


def print_password_report(passwords):
    for password in passwords:
        result = check_password(password)
        print(password, "->", result)


passwords_to_check = ["hunter2", "abc", "correcthorsebattery9"]
print_password_report(passwords_to_check)
````

```
Step 1: has_minimum_length(password)
def has_minimum_length(password):

- len(password) counts the characters.
- >= 8 checks whether the length is 8 or greater.
- return sends back True or False.

Example:
has_minimum_length("python123")
Output: True

has_minimum_length("abc")
Output: False

Remember: Its only responsibility is checking password length.
```

```
Remember: Its only responsibility is generating the report, not implementing the password validation rules.
```
