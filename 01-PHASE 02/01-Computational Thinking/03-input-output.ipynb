```python
# The Input → Process → Output (IPO) model means a program takes in data, works on it, and produces a result.
```

```python
"""
- Input: Data that a program receives to work with.
- Process: Steps a program performs to work on the input and get a result.
- Output: The result or information produced by a program.
- input(): A Python function that asks the user to enter something and receives it as text.
- print(): A Python function that displays information on the screen.
- Type conversion: Changing a value from one data type to another, such as converting text "5" into the number 5.
"""
```

```python
# 1. The IPO model in one sentence:
# IPO means a program takes input, processes it, and produces an output.

price = 100          # Input
quantity = 3         # Input
total = price * quantity  # Process
print(total)          # Output
```

```python
# 2. Why this model helps you before you code:
# Before writing code, identify what information you need (Input), what work you must perform (Process), and what result you want (Output).

# Input
marks = 80

# Process
result = marks + 5

# Output
print(result)
```

```python
# 3. Input in Python:
# Input is the information a program receives, either from existing variables or directly from a user using input().

name = input("What is your name? ")
print(name)
```

```python
# 4. Process in Python:
# Process means performing calculations or other operations on input data to get the required result

price = 100
quantity = 2

total = price * quantity
```

```python
# 5. Output in Python
# Output is the result produced by a program, which can be displayed on the screen using print()
total = 200
print(total)
```

```python
# 6. The type-conversion trap:
# Python's input() function always receives data as text, so you must convert it into a number using int() or float() before performing numerical calculations.

age_text = input("Enter your age: ")

age_number = int(age_text)

next_year = age_number + 1

print(next_year)
```

```python
# This program asks the user for their name, joins it with a greeting message, and displays a personalized greeting on the screen.
name = input("What is your name? ")
greeting = "Hello, " + name + "!"
print(greeting)
```

```python
# This program asks the user to enter an item's price and quantity using input(), converts the price from text into a decimal number using float() and the quantity into a whole number using int(), multiplies them using total = price * quantity to calculate the total cost, stores the result in total, and finally displays the result using print() (for example, if the price is 19.99 and the quantity is 3, it displays Total cost: 59.97)

price_text = input("Enter the price of the item: ")
quantity_text = input("Enter how many you are buying: ")

price = float(price_text)
quantity = int(quantity_text)

total = price * quantity

print("Total cost:", total)
```

```python
# This program asks the user to enter three expenses, converts each input from text into a decimal number using float(), adds them together to calculate the total spending, divides the total by 3 to calculate the average expense, displays both results using print(), and finally checks whether the total spending is greater than 100 using if total > 100—if it is, it displays a high-spending warning; otherwise, the warning is skipped.

expense_1 = float(input("Enter first expense: "))
expense_2 = float(input("Enter second expense: "))
expense_3 = float(input("Enter third expense: "))

total = expense_1 + expense_2 + expense_3
average = total / 3

print("Total spent:", total)
print("Average expense:", average)

if total > 100:
    print("Warning: total spending is high this period.")
```
