```
Topic: Pseudocode, Flowcharts, and Dry Runs
Why this topic matters: Think through your solution before writing code and manually check its logic to catch mistakes early.

KEY TERMS:

1. Pseudocode: Writing the steps of a program in simple English before writing actual code.
2. Flowchart: A diagram that visually shows the steps and decisions of a program.
3. Dry Run: Manually checking how code executes, step by step, without running it.
4. Trace Table: A table used during a dry run to track how variable values change at each step.
```

```
1. Why plan before coding:
Planning helps you understand the program's logic before writing Python code, reducing mistakes.

Example: First decide how to calculate the average of three numbers, then write the Python code.

2. Writing pseudocode
Writing the program's steps in simple English before converting them into Python code.

Example:

START
  SET total TO 0
  ADD each expense to total
  RETURN total
END

Meaning: Start with zero, add all expenses, and return the final total.

3. Drawing a simple text-based flowchart
Showing the program's steps and decisions visually using boxes, arrows, and branches.

Example:

[Start]
   |
   v
[Set total = 0]
   |
   v
<More expenses?>
   | Yes       | No
   v           v
[Add expense] [Return total]
   |           |
   +-----> [End]

Meaning: Keep adding expenses until none are left, then return the total.

4. Dry runs: tracing code by hand:
Manually following the code line by line and tracking variable values to understand the output or find mistakes.

Example:
```

````python
total = 0
for number in [2, 3]:
    total = total + number
````

```
Step	            number	             total
Start	              —	                   0
1	                  2	                   2
2	                  3	                   5

Final value: total = 5 — without running the code, we manually tracked how the value changed.
```

```
Example 1 — Pseudocode and Python Side by Side
Problem: Count how many numbers in a list are greater than 10.

1. Pseudocode
- START — Begin the algorithm.
- SET count TO 0 — Create a counter and initialize it to 0.
- FOR EACH number IN numbers — Check every number in the list one by one.
- IF number > 10 — Check whether the current number is greater than 10.
- ADD 1 TO count — Increase the counter by 1 if the condition is True.
- END IF — Finish checking the condition.
- END FOR — Finish checking all numbers in the list.
- RETURN count — Return the final count.
- END — Finish the algorithm.
```

````python
def count_checking(numbers):
    count = 0
    for number in numbers:
        if number > 10:
            count = count + 1
    return count 

result = count_checking([3,67,11,10,6,8,9,23,13,15,24,1])
print(result)
````


````python
total = 0
count = 0
for number in [4, 7, 2, 9]:
    total = total + number
    count = count + 1
    if number > 5:
        total = total + 1

average = total / count
print(average)
````

```
Example 3 — Using a Dry Run to Find a Bug
Problem: Count how many positive and negative numbers are present in a list.
```

````python
def positive_negative_count(numbers):
    positive_count = 0
    negative_count = 0
    neutral_count = 0
    for number in numbers:
        if number > 0:
            positive_count = positive_count + 1
        elif number < 0:
            negative_count = negative_count + 1
        else:
            neutral_count = neutral_count + 1
    return positive_count, neutral_count, neutral_count

result = positive_negative_count([12,3,4,-2,0,6,-5,0,-7,-1,-6,0,0,2])
print(result)
````
