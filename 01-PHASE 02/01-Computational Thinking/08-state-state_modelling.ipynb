```
State and State Modelling:

1. State: Information a program remembers between steps. Example: balance = 5000 remembers the current bank balance.

2. State changes over time: A program updates its remembered information when something happens. Example: After depositing ₹1,000, the balance changes from ₹5,000 to ₹6,000.

3. Unexpected state changes: Bugs can happen when a program changes information incorrectly. Example: A bank account balance accidentally becomes negative after a withdrawal.

4. Invalid state changes: Some changes should never be allowed according to the system's rules. Example: A traffic light should not jump directly from red to yellow.

5. State modelling: Planning what information a program must remember and defining which changes are allowed. Example: Designing a traffic light system that follows red → green → yellow → red.

6. Valid vs. invalid changes: A valid change follows the system's rules, while an invalid change breaks them. Example: red → green is valid, but red → yellow is invalid in our simplified traffic-light model.
```

```
Step-by-step explanation:

Topic: 01. What is State:

1. State: Information a program remembers that can change over time and affect future actions. Example: balance = 5000 stores the current bank balance.

2. State persists: Remembered information stays available across multiple program steps. Example: After depositing ₹1,000, the balance remains ₹6,000 until another transaction changes it.

3. State updated by events: An event triggers a change in the current state. Example: Clicking the login button changes logged_in = False to logged_in = True.

4. State affects future behaviour: The program uses its current state to decide what to do next. Example: If balance = 500, the program should reject a withdrawal of ₹1,000.

5. Not State (Recalculatable Value): A value that can be calculated whenever needed from existing data does not necessarily need to be stored separately. Example: Calculate the average of [10, 20, 30] whenever needed instead of storing it.

6. State Modelling Decision: Decide which information must be remembered and which can be recalculated. Example: Store bank balance, but calculate the average transaction amount from transaction records.
```

```
Topic: 02. State transitions: How State Changes

1. State Transition: A change from one current state to another. Example: logged_out → logged_in.

2. Event: An action or occurrence that triggers a state transition. Example: Clicking the login button triggers the login process.

3. Events cause transitions: Different events can move a system into different states. Example: An elevator moves from idle → moving_up when someone requests an upper floor.
```

```
Topic: 03. Valid versus Invalid Transitions

1. Valid Transition: A state change that follows the system's rules. Example: red → green in our traffic-light model.

2. Invalid Transition: A state change that breaks the system's rules. Example: red → yellow in our traffic-light model.

3. Preventing Invalid Transitions: Check whether a requested change is allowed before updating the state. Example: Reject red → yellow instead of changing the traffic light.

4. Precondition Check: A condition that must be satisfied before an operation is allowed. Example: Check whether the current state is red before allowing a transition to green.
```

```
Topic: 04. A State Machine: States Plus Allowed Transitions

1. State Machine: A model that defines all possible states and the transitions allowed between them. Example: A traffic light with red, green, and yellow states.

2. Current State: The state in which the system currently exists. Example: current_state = "red".

3. Allowed Next State: The state a system is permitted to enter next. Example: From red, the only allowed next state is green.

4. Transition Table: A table showing which state changes are allowed.

Current state	             Allowed next state
red	                               green
green	                           yellow
yellow	                             red

5. Rejecting Invalid Transitions: If a requested transition is not in the rules, reject it and keep the current state unchanged. Example: red → yellow is rejected, so the state remains red.
```

```
Topic: 05. Representing State and Transitions in Python

1. Variable for State: Use a variable to store the current state as a string. Example: current_state = "red".

2. Function for Transitions: Use a function to decide which state comes next. Example: advance("red") returns "green".

3. if / elif / else: Use conditional statements to check the current state and decide what action to take. Example: if state == "red": return "green".

4. Checking Before Updating: Verify that a requested transition is allowed before changing the state. Example: Only update current_state if the transition validation function returns True.
```

```
Example 1 — A Single Piece of State: A Simple Counter

1. def increment(current_value): — Defines a function named increment that accepts a number as input.

2. return current_value + 1 — Adds 1 to the input number and returns the new value. Example: 5 becomes 6.

3. counter_value = 0 — Creates a variable that stores the initial state of the counter, which is 0.

4. print(counter_value) — Displays the current counter value, which is 0.

5. counter_value = increment(counter_value) — Calls the function with the current value (0), gets 1, and saves it back into counter_value.

6. counter_value = increment(counter_value) — Calls the function again with 1, gets 2, and saves it back into counter_value.

7. print(counter_value) — Displays the updated counter value, which is 2.
```

````python
def increment(current_value):
    return current_value + 1


counter_value = 0
print(counter_value)                       # 0

counter_value = increment(counter_value)
counter_value = increment(counter_value)
print(counter_value)                       # 2
````

```
Example 2 — A Traffic Light with Valid Transitions Only

1. def advance(state): — Defines a function named advance that accepts the current traffic-light state as input.

2. if state == "red": — Checks whether the current state is "red".

3. return "green" — If the state is "red", returns "green" as the next state.

4. elif state == "green": — Checks whether the current state is "green".

5. return "yellow" — If the state is "green", returns "yellow" as the next state.

6. elif state == "yellow": — Checks whether the current state is "yellow".

7. return "red" — If the state is "yellow", returns "red" as the next state.

8. current_state = "red" — Creates a variable that stores the initial traffic-light state as "red".

9. print(current_state) — Displays the current state: red.

10. current_state = advance(current_state) — Calls advance() with "red", receives "green", and saves it in current_state.

11. print(current_state) — Displays the updated state: green.

12. current_state = advance(current_state) — Calls advance() with "green", receives "yellow", and saves it in current_state.

13. print(current_state) — Displays the updated state: yellow.

14. current_state = advance(current_state) — Calls advance() with "yellow", receives "red", and saves it in current_state.

15. print(current_state) — Displays the updated state: red.
```

````python
def advance(state):
    if state == "red":
        return "green"
    elif state == "green":
        return "yellow"
    elif state == "yellow":
        return "red"


current_state = "red"
print(current_state)                       # red

current_state = advance(current_state)
print(current_state)                       # green

current_state = advance(current_state)
print(current_state)                       # yellow

current_state = advance(current_state)
print(current_state)                       # red
````

```
Example 3 — An Elevator That Rejects Invalid Requested Transitions:

Function 1: Checking whether a transition is allowed

1. def is_transition_allowed(state, requested_state): — Defines a function that accepts the current state and the requested next state. Example: "idle", "moving_up".

2. if state == "idle": — Checks whether the elevator is currently idle.

3. return requested_state == "moving_up" or requested_state == "moving_down" — Returns True if the requested state is either "moving_up" or "moving_down"; otherwise, returns False.

4. elif state == "moving_up": — Checks whether the elevator is currently moving upward.

5. return requested_state == "idle" or requested_state == "door_open" — Allows the elevator to move to "idle" or "door_open".

6. elif state == "moving_down": — Checks whether the elevator is currently moving downward.

7. return requested_state == "idle" or requested_state == "door_open" — Allows the elevator to move to "idle" or "door_open".

8. elif state == "door_open": — Checks whether the elevator's door is currently open.

9. return requested_state == "idle" — Allows the elevator to transition only to "idle".

10. else: — Handles any unrecognized current state.

11. return False — Rejects the transition if the current state is not recognized.

Function 2: Requesting a transition

12. def request_transition(state, requested_state): — Defines a function that checks whether a requested state change is allowed.

13. if is_transition_allowed(state, requested_state): — Calls the first function to validate the requested transition.

14. return True — Returns True if the requested transition is valid.

15. else: — Executes when the requested transition is invalid.

16. return False — Returns False if the requested transition is rejected.

Main program: Updating the elevator state

17. current_state = "idle" — Sets the elevator's initial state to "idle".

18. print(current_state) — Displays the initial state: idle.

19. if request_transition(current_state, "moving_up"): — Checks whether moving upward is allowed from the current state.

20. current_state = "moving_up" — Updates the state to "moving_up" because the transition is valid.

21. print(current_state) — Displays the updated state: moving_up.

22. if request_transition(current_state, "door_open"): — Checks whether opening the door is allowed while moving upward.

23. current_state = "door_open" — Updates the state to "door_open" because the transition is valid.

24. print(current_state) — Displays the updated state: door_open.

25. if request_transition(current_state, "moving_down"): — Checks whether moving downward is allowed while the door is open.

26. current_state = "moving_down" — Would update the state to "moving_down" if the transition were valid, but this line does not execute.

27. print(current_state) — Displays door_open because the invalid transition was rejected and the state remained unchanged.
```

````python
def is_transition_allowed(state, requested_state):
    if state == "idle":
        return requested_state == "moving_up" or requested_state == "moving_down"
    elif state == "moving_up":
        return requested_state == "idle" or requested_state == "door_open"
    elif state == "moving_down":
        return requested_state == "idle" or requested_state == "door_open"
    elif state == "door_open":
        return requested_state == "idle"
    else:
        return False


def request_transition(state, requested_state):
    if is_transition_allowed(state, requested_state):
        return True
    else:
        return False


current_state = "idle"
print(current_state)                                         # idle

if request_transition(current_state, "moving_up"):
    current_state = "moving_up"
print(current_state)                                          # moving_up

if request_transition(current_state, "door_open"):
    current_state = "door_open"
print(current_state)                                          # door_open

if request_transition(current_state, "moving_down"):
    current_state = "moving_down"
print(current_state)                                          # still door_open (rejected)
````
