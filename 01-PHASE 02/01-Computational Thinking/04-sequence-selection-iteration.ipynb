```
Topic: Sequence, Selection, Iteration, and Abstraction

Why this topic matters:

- Sequence: Executing Python statements one after another in the order they are written.
- Selection: Choosing which instructions to execute based on whether a condition is True or False.
- Iteration: Repeating a set of instructions multiple times using loops.
- Abstraction: Grouping a set of instructions under a reusable name, usually a function, so you can use them without rewriting the code.

-- Why this topic matters: Understanding these four concepts helps you recognize how programs are structured and how they perform tasks, make decisions, repeat actions, and reuse code.
```

```
Topic: Key terms

1. Sequence: Executing Python instructions one after another in the order they are written.
2. Selection: Choosing which instructions to execute based on whether a condition is true or false, using if, elif, or else.
3. Condition: An expression that Python checks to determine whether something is True or False.
4. Iteration: Repeating a set of instructions multiple times using a for or while loop.
5. Abstraction: Grouping multiple instructions under one name, usually a function, so you can reuse them without rewriting the code.
6. Function: A named, reusable block of Python code that performs a specific task and can accept inputs and return a result.
7. Parameter: A named placeholder inside a function definition that receives a value when the function is called.
8. Return value: The result that a function sends back to the part of the program that called it.
```

```
Step-by-step explanation: Sequence, Selection, Iteration, and Abstraction

1. Sequence: The default behavior

- Sequence means Python executes instructions one after another, from top to bottom, in the exact order they are written.

- Python normally follows this order unless a decision (if) or a loop (for or while) changes which instructions run or how often they run.

Example:
```

````python
name = "Sovon"
age = 25
next_year_age = age + 1

print(name)
print(next_year_age)
````

```
How does this program work?
1. name = "Sovon" stores the name in the variable name.
2. age = 25 stores the age in the variable age.
3. next_year_age = age + 1 adds 1 to the age and stores the result.
4. print(name) displays the name.
5. print(next_year_age) displays the age for next year.

Key takeaway: Python executes instructions in sequence, meaning one instruction finishes before the next instruction runs.
```

```
2. Selection: Doing something only if a condition is true
What does it mean?

Selection means allowing a Python program to make decisions and choose which instructions to execute based on whether a condition is True or False.

Python uses three main keywords for selection:
- if: Executes a block of code when a condition is true.
- elif: Checks another condition when the previous condition is false.
- else: Executes a block when none of the previous conditions are true.

A. Understanding the if statement

Example:
```

````python
age = 20

if age >= 18:
    print("You can vote.")
````

```
How does this program work?
1. age = 20 stores 20 in the variable age.
2. if age >= 18: checks whether the age is greater than or equal to 18.
3. Since 20 is greater than 18, the condition becomes True.
4. Python executes the indented print() statement and displays the message.

If the age were 15, the condition would be False, so Python would skip the indented statement.
```

```
C. Understanding if, elif, and else together
What does it mean?

You can use if, elif, and else together when your program needs to check multiple conditions and select the appropriate action.

Example:
```

````python
temperature = 15

if temperature > 30:
    print("It's hot.")

elif temperature > 15:
    print("It's warm.")

else:
    print("It's cool.")
````

```
How does this program work?

1. temperature = 15 stores 15 in the variable.
2. if temperature > 30 checks whether the temperature is above 30; the condition is false.
3. elif temperature > 15 checks whether the temperature is above 15; this condition is also false.
4. Since both conditions are false, Python executes the else block.
5. The program displays It's cool.

Output:
It's cool.
```

```
Boolean logic allows you to combine or reverse conditions when making decisions.
```

```
3. Iteration: Doing something repeatedly
What does it mean?

Iteration means repeating a set of Python instructions multiple times instead of writing the same instructions again and again.

Python provides two main types of loops:
1. for loop
2. while loop

A. Understanding the for loop
What does it mean?

A for loop repeats a block of code once for each item in a collection, such as a list, or for each number in a sequence generated by range().

Example:
```

````python
for number in [10, 20, 30]:
    print(number)
````

```
How does this program work?
1. Python takes the first value, 10, from the list and stores it in number.
2. It executes print(number) and displays 10.
3. Python takes the second value, 20, and displays it.
4. Python takes the third value, 30, and displays it.
5. After processing all three values, the loop finishes.
```

```
Understanding how many times the loop runs:

Iteration	Value of number	   Output
First	         10	            10
Second	         20	            20
Third	         30	            30

The loop runs three times because the list contains three items.
```

```
B. Understanding the range() function:
range() generates a sequence of numbers that a for loop can use to repeat instructions a specific number of times.

Example:
```

````python
for number in range(1, 6):
    print(number)
````

```
How does this program work?

Python generates the numbers 1, 2, 3, 4, and 5, and executes the print() statement once for each number.

Important: range(1, 6) starts at 1 but stops before 6, so 6 is not included.
```

```
C. Understanding the while loop
What does it mean?

A while loop repeatedly executes a block of code as long as its specified condition remains True.

Example:
```

````python
count = 0

while count < 3:
    print(count)
    count = count + 1
````

```
How does this program work?
1. count = 0 initializes the counter with the value 0.
2. while count < 3 checks whether the counter is less than 3.
3. Since 0 is less than 3, Python executes the loop.
4. print(count) displays 0.
5. count = count + 1 increases the counter from 0 to 1.
6. Python checks the condition again and repeats the same process.
7. When count becomes 3, the condition count < 3 becomes false, and the loop stops.
```

```
Understanding the loop execution:
Iteration	   Count before printing	    Condition	         Action
First	               0	                 0 < 3 → True	      Prints 0
Second	               1	                 1 < 3 → True	      Prints 1
Third	               2	                 2 < 3 → True	      Prints 2
Fourth check	       3	                 3 < 3 → False	      Stops
```

```
D. Understanding an infinite loop
An infinite loop happens when a loop's condition never becomes false, causing Python to repeat the instructions continuously.

- Example of an infinite loop:
count = 0

while count < 3:
    print(count)

- Why does this happen?

- The variable count always remains 0 because we never increase its value, so count < 3 always remains true and the loop never stops on its own.

- Correct version:
count = 0

while count < 3:
    print(count)
    count = count + 1

- Here, increasing count eventually makes the condition false, allowing the loop to stop.
```

```
Key difference between for and while:

for loop:
- Repeats for each item in a collection or sequence.	
- Commonly used when the items or repetitions are known.	
- Automatically moves to the next item.

while loop:
- Repeats as long as a condition is true.
- Commonly used when repetitions depend on a condition.
- Often requires you to update a variable to eventually stop the loop.
```

```
4. Abstraction: Giving steps a name with functions

What does it mean?
- Abstraction means grouping related instructions under a meaningful name, usually a function, so you can reuse those instructions without writing the same code repeatedly.

A. Understanding functions
- A function is a named, reusable block of Python code designed to perform a particular task.

Example:
```

````python
def greet(name):
    print("Hello, " + name + "!")

greet("Ada")
greet("Grace")
````

```
How does this program work?
1. def greet(name): defines a function named greet.
2. name is a parameter that acts as a placeholder for the input value.
3. print("Hello, " + name + "!") is the instruction inside the function.
4. Defining the function does not execute its instructions immediately; it makes the function available for later use.
5. greet("Ada") calls the function and passes "Ada" as the input.
6. Python executes the function and displays Hello, Ada!.
7. greet("Grace") calls the same function again with "Grace".

Output:
Hello, Ada!
Hello, Grace!

The important idea: You define the instructions once and call the same function whenever you need to perform that task.
```

```
C. Understanding the return statement:

What does it mean?
- The return statement sends a calculated result from inside a function back to the part of the program that called the function.

Example:
```

````python
def add(a, b):
    return a + b

result = add(3, 4)

print(result)
````

```
How does this program work?
1. def add(a, b): defines a function named add that accepts two parameters.
2. add(3, 4) calls the function and passes 3 and 4 as inputs.
3. Inside the function, return a + b adds the two numbers.
4. The function sends the calculated result, 7, back to the calling code.
5. result = add(3, 4) stores the returned value 7 in the variable result.
6. print(result) displays the result.
```

```
D. Understanding the difference between print() and return
This is one of the most important concepts when learning Python functions.

print()
Displays information on the screen.	
Mainly used to show information to the user.
Does not automatically provide the displayed value to the calling code.

return
Sends a result back from a function.
Used when another part of the program needs the calculated result.
Makes the returned value available for further operations.

Example using print():
```

````python
def add(a, b):
    print(a + b)

result = add(3, 4)

print(result)
````

```
Why does this happen?
- The function displays 7 using print().
- However, it does not return anything.
- Python therefore gives the function a default return value of None.
- As a result, result stores None.

Example using return:
```

````python
def add(a, b):
    return a + b

result = add(3, 4)

print(result)
````

```
Here, the function returns 7, which gets stored in result and can be used in other calculations.
```

```
Example 01: Selection
```

````python
password = "hunter2"

if len(password) >= 8:
    print("Password length is OK.")
else:
    print("Password is too short.")
````

```
What does this program do?

This program checks whether a password contains at least 8 characters and displays a message depending on whether the password meets that length requirement.

Step-by-step execution:
1. password = "hunter2": Stores the text "hunter2" inside the variable password.
2. len(password): Uses Python's built-in len() function to count the number of characters in the password; here, the count is 7.
3. if len(password) >= 8:: Checks whether the password contains 8 or more characters; since 7 is less than 8, the condition evaluates to False.
4. print("Password length is OK."): Python skips this instruction because the if condition is false.
5. else:: Tells Python to execute the alternative instructions when the if condition is false.
6. print("Password is too short."): Displays the message because the password contains fewer than 8 characters.

How selection works in this program:
- If the password has 8 or more characters, Python displays Password length is OK.
- If the password has fewer than 8 characters, Python displays Password is too short.

Key takeaway: This is an example of selection because Python checks a condition and chooses one of two possible actions based on the result (True or False).
```

```
Example 2 — Iteration: Counting positive, negative, and zero numbers
```

````python
numbers = [4, -2, 0, 7, -9, 0, 1]

positive_count = 0
negative_count = 0
zero_count = 0

for number in numbers:
    if number > 0:
        positive_count = positive_count + 1
    elif number < 0:
        negative_count = negative_count + 1
    else:
        zero_count = zero_count + 1

print("Positive:", positive_count)
print("Negative:", negative_count)
print("Zero:", zero_count)
````

```
1. What does this program do:
This program takes a list of numbers, checks each number to determine whether it is positive, negative, or zero, counts how many numbers belong to each category, and finally displays the total count for all three categories.

2. Understanding the code:

Step 1: Creating a list of numbers
numbers = [4, -2, 0, 7, -9, 0, 1]

This creates a list named numbers containing seven values: 4, -2, 0, 7, -9, 0, and 1.

Step 2: Initializing three counters
positive_count = 0negative_count = 0zero_count = 0

Here, we create three variables to count positive numbers, negative numbers, and zeros, and initialize each counter with 0 because no numbers have been checked yet.

Step 3: Understanding the for loop
for number in numbers:

This for loop takes each number from the numbers list, one at a time, stores it temporarily in the variable number, and executes the indented instructions for every item in the list.

For example, Python processes the numbers in this order:

4 → -2 → 0 → 7 → -9 → 0 → 1

Step 4: Checking positive numbers using if
if number > 0:    positive_count = positive_count + 1

This condition checks whether the current number is greater than zero; if it is, Python increases positive_count by 1.

For example, when number = 4, the condition is true, so positive_count changes from 0 to 1.

Step 5: Checking negative numbers using elif
elif number < 0:    negative_count = negative_count + 1

If the first condition is false, Python checks whether the current number is less than zero; if it is, Python increases negative_count by 1.

For example, when number = -2, the condition is true, so negative_count changes from 0 to 1.

Step 6: Checking zero using else
else:    zero_count = zero_count + 1

If neither the if nor the elif condition is true, Python executes the else block and increases zero_count by 1, which happens when the current number is exactly zero.

For example, when number = 0, zero_count changes from 0 to 1.

Step 7: Displaying the final results
print("Positive:", positive_count)print("Negative:", negative_count)print("Zero:", zero_count)

After the for loop has processed all seven numbers, these three print() statements display the final number of positive values, negative values, and zeros.
```

```
3. Understanding how the loop processes each number
Let's see how the three counters change as Python processes the list.

Iteration	Current number	Category	Positive count	Negative count	Zero count
1	              4	        Positive	      1	               0	       0
2	             -2	        Negative	      1	               1	       0
3	              0	        Zero	          1	               1	       1
4	              7	        Positive	      2	               1	       1
5	             -9	        Negative	      2	               2	       1
6	              0	        Zero	          2	               2	       2
7	              1	        Positive	      3	               2	       2
```

```
Example 3 — Abstraction: Turning Repeated Logic into a Function
```

````python
def is_password_long_enough(password):
    return len(password) >= 8


def count_signs(numbers):
    positive_count = 0
    negative_count = 0
    zero_count = 0
    for number in numbers:
        if number > 0:
            positive_count = positive_count + 1
        elif number < 0:
            negative_count = negative_count + 1
        else:
            zero_count = zero_count + 1
    return positive_count, negative_count, zero_count


passwords_to_check = ["hunter2", "correcthorsebatterystaple", "abc"]
for password in passwords_to_check:
    if is_password_long_enough(password):
        print(password, "-> OK")
    else:
        print(password, "-> too short")

positives, negatives, zeros = count_signs([4, -2, 0, 7, -9, 0, 1])
print("Positive:", positives, "Negative:", negatives, "Zero:", zeros)
````

```
Program Explanation
This program performs two tasks:
1. Checks whether passwords contain at least 8 characters.
2. Counts positive, negative, and zero numbers in a list.
1. First function: Password length check
```

````python
def is_password_long_enough(password):
    return len(password) >= 8
````

```
- def creates a function named is_password_long_enough.
- password receives the password as input.
- len(password) counts the characters.
- >= 8 checks whether the password has at least 8 characters.
- return sends back True or False.

For example, "hunter2" has 7 characters, so the function returns False.

2. Second function: Count positive, negative, and zero numbers
```

````python
def count_signs(numbers):
    positive_count = 0
    negative_count = 0
    zero_count = 0
    for number in numbers:
        if number > 0:
            positive_count = positive_count + 1
        elif number < 0:
            negative_count = negative_count + 1
        else:
            zero_count = zero_count + 1
    return positive_count, negative_count, zero_count
````

```
- count_signs(numbers) receives a list of numbers.
- Three counters are initialized at 0.
- The for loop processes each number one by one.
- if number > 0 increases the positive counter.
- elif number < 0 increases the negative counter.
- else increases the zero counter.
- return sends all three counts back together.

For example, [4, -2, 0, 7, -9, 0, 1] produces:

- Positive numbers: 3
- Negative numbers: 2
- Zero values: 2

3. Check the passwords
```

````python
passwords_to_check = ["hunter2", "correcthorsebatterystaple", "abc"]

for password in passwords_to_check:
    if is_password_long_enough(password):
        print(password, "-> OK")
    else:
        print(password, "-> too short")
````

```
- Creates a list containing three passwords.
- The for loop checks each password one by one.
- Calls is_password_long_enough() to check its length.
- If the function returns True, it prints OK.

4. Call the counting function:
```

````python
positives, negatives, zeros = count_signs([4, -2, 0, 7, -9, 0, 1])

print("Positive:", positives, "Negative:", negatives, "Zero:", zeros)
````

```
- Calls count_signs() with a list of numbers.
- Receives three returned values.
- Stores them in three separate variables: positives, negatives, and zeros.
- Prints the results.
```
