```
Topic: Tuples and Unpacking

A tuple groups a small, fixed collection of related values that cannot be changed, while unpacking separates those values into individual variables in one line.

- Tuple: A collection of values that keeps their order and cannot be changed. Example: (3, 5)
- Immutable: Something that cannot be changed after it is created.
- Packing: Combining multiple values into a single tuple. Example: numbers = (10, 20, 30)
- Unpacking: Taking values from a tuple and assigning them to separate variables. Example: x, y = (3, 5)
- Single-item tuple: A tuple containing only one value, which requires a trailing comma. Example: (5,)
- Extended unpacking: Using *name to collect all remaining tuple values into a list. Example: first, *rest = (1, 2, 3, 4)
- Record: A small group of related values stored together as one unit. Example: ("Ada", 1815)
```

```
7. Function returning multiple values
A function can return multiple comma-separated values, which Python packs into a tuple.
```

````python
def min_max(numbers):
    return min(numbers), max(numbers)


result = min_max([10, 5, 30, 20])

print(result)
````

```
8. Unpacking function results
You can assign multiple values returned by a function to separate variables.
```

````python
def min_max(numbers):
    return min(numbers), max(numbers)


numbers = [10, 5, 30, 20]

lowest, highest = min_max(numbers)

print("Lowest:", lowest)
print("Highest:", highest)
````

```
7. Extended unpacking with *

1. Extended unpacking
Extended unpacking uses * to collect multiple remaining values into one variable.
```

````python
first, *rest = (1, 2, 3, 4)

print(first)
print(rest)
````

```
2. Middle unpacking
Using *middle collects all values between the first and last elements into a list.
```

````python
first, *middle, last = (1, 2, 3, 4, 5)

print(first)
print(middle)
print(last)
````

```
Appendix: Complete tuple Method Reference:
A Python tuple has exactly two public methods: count() and index(). Both methods read information from a tuple; neither changes its contents.

1. count(value)
The count() method returns how many times a particular value appears in a tuple.
```

````python
numbers = (1, 2, 2, 3, 2)

result = numbers.count(2)

print(result)
````

```
Explanation: count(2) counts how many times 2 appears in the tuple and returns 3.
- Does not modify the tuple.
- Returns an integer (int).

2. index(value)
The index() method returns the position of the first occurrence of a particular value in a tuple.
```

````python
colors = ("red", "green", "blue")

result = colors.index("blue")

print(result)
````

```
Example 1 — Representing a fixed point:
```

````python
point = (12, 7)

print(point)
print(f"x = {point[0]}, y = {point[1]}")

try:
    point[0] = 99
except TypeError as error:
    print("Could not change the point:", error)
````

```
Example 2 — A function returning related values, unpacked:
```

````python
def temperature_stats(readings):
    lowest = min(readings)
    highest = max(readings)
    average = sum(readings) / len(readings)
    return lowest, highest, average

readings = (18.5, 22.0, 19.5, 25.0, 20.5)
lowest, highest, average = temperature_stats(readings)

print(f"Lowest: {lowest}")
print(f"Highest: {highest}")
print(f"Average: {average:.1f}")
````

```
Example 3 — Iterating over a list of small records:
```

````python
birth_records = [("Ada", 1815), ("Grace", 1906), ("Alan", 1912)]

for name, year in birth_records:
    if year < 1900:
        print(f"{name} was born before 1900.")
    else:
        print(f"{name} was born in {year}.")
````
