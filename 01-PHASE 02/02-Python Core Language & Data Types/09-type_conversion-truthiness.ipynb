```
8. any() and all():

Both functions work with an iterable, such as a list, tuple, or set.

A. any(iterable)
any() returns True if at least one value in the iterable is truthy.
```

````python
values = [False, False, True]

print(any(values))  # True
````

```
Explanation:
- First value is False.
- Second value is False.
- Third value is True.
- At least one value is truthy, so any() returns True.

Another example:
```

````python
scores = [0, 85, 90]

print(any(scores))  # True
````

```
Why?
- 0 is falsy.
- 85 is truthy.
- 90 is truthy.
- At least one value is truthy.

B. all(iterable)
all() returns True only when every value in the iterable is truthy.
```

````python
values = [True, True, True]

print(all(values))  # True
````

````python
values = [True, False, True]

print(all(values))  # False
````

```
Why?

One value is False, so not every value is truthy.
```

```
C. Truthiness aggregation:
Truthiness aggregation means checking multiple values together using any() or all() instead of writing individual conditions.

Example:
```

````python
scores = [85, 90, 75]

print(any(scores))  # True
print(all(scores))  # True
````

````python
scores = [85, 0, 75]

print(any(scores))  # True
print(all(scores))  # False
````

```
Explanation:
- any(scores) returns True because at least one score is non-zero.
- all(scores) returns False because 0 is falsy.

Important additional behavior:
```

````python
print(any([]))  # False
print(all([]))  # True
````

```
An empty iterable has no truthy elements, so any() returns False; all() returns True because there is no element that violates the requirement that every element be truthy.
```

```
9. Collection type conversion:

Collection conversion means changing one collection type into another using list(), tuple(), set(), or dict().

A. list(iterable)
list() converts an iterable into a list.

Example 1 — String to list:
```

````python
language = "Python"

result = list(language)

print(result)
````

```
Explanation: Python converts each character of the string into an individual list element.

Example 2 — Tuple to list:
```

````python
numbers = (10, 20, 30)

result = list(numbers)

print(result)
````

```
The tuple is converted into a list.

B. tuple(iterable)
tuple() converts an iterable into a tuple.
```

````python
numbers = [10, 20, 30]

result = tuple(numbers)

print(result)
````

```
Explanation: The original list becomes a tuple.

C. set(iterable)
set() converts an iterable into a set and automatically removes duplicate values.
```

````python
numbers = [10, 20, 20, 30, 30]

result = set(numbers)

print(result)
````

```
D. dict(pairs)
dict() converts a collection of two-item pairs into a dictionary.
```

````python
data = [
    ("name", "Sovon"),
    ("age", 25),
    ("city", "Kolkata")
]

result = dict(data)

print(result)
````

```
E. Conversion between collection types
You can convert collections into other collection types.
```

````python
numbers = [10, 20, 30]

# List to tuple
tuple_data = tuple(numbers)

# List to set
set_data = set(numbers)

# Tuple to list
list_data = list(tuple_data)

print(tuple_data)  # (10, 20, 30)
print(set_data)    # {10, 20, 30}
print(list_data)   # [10, 20, 30]
````

```
10. isinstance():
A. isinstance(value, type)
isinstance() checks whether a value belongs to a specified data type and returns True or False.

Example:
```

````python
age = 25

print(isinstance(age, int))  # True
print(isinstance(age, str))  # False
````

```
B. Practical type validation
You can use isinstance() to validate data before processing it.
```

````python
age = 25

if isinstance(age, int):
    print("Age is a valid integer")
else:
    print("Age must be an integer")
````

```
C. type(value) == int versus isinstance(value, int)
Both can check the type of a value, but isinstance() is usually preferred.
```

````python
number = 42

print(type(number) == int)  # True
print(isinstance(number, int))  # True
````
