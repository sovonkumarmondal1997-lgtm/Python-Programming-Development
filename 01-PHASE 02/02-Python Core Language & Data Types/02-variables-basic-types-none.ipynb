```
Topics: Variables, Basic Types, and None

Every piece of data in Python has a type that tells Python what kind of value it is and what operations can be performed on it.

KEY_TERMS:

Type: Tells Python what kind of value something is and what you can do with it.
int: Stores whole numbers without decimal points, such as 7 or -3.
float: Stores numbers with decimal points, such as 3.14 or 7.0.
bool: Stores only two values: True or False.
str: Stores text written inside quotation marks, such as "hello".
None: Represents the absence of a meaningful value, such as information that is not available yet.
type(): Tells you the data type of a value, such as type(7) returning int.
```

```
Step-by-step explanation:

1. A quick recap: variables and assignment
- Variable: A name used to store or refer to a value, e.g. age = 25.
- Assignment (=): Connects a variable name to a value, e.g. age = 25.
- Object: The actual data stored in Python, e.g. the integer object 25.
- Reassignment: Assigning a new value to an existing variable, e.g. age = 30.
- Meaningful variable names: Use descriptive names so the purpose is clear, e.g. total_price instead of x.
- Snake_case: A naming style that uses lowercase words separated by underscores, e.g. customer_name.

2. What a type is, and why Python cares
- Type: Identifies what kind of value an object contains and what operations are allowed.
- Why Python cares: Python prevents incompatible operations, such as directly adding text to a number.
- Example: 10 + 5 works, but "Age" + 25 produces a TypeError.

3. int: whole numbers
- int: Stores whole numbers without decimal points, e.g. age = 25.
- Positive number: A number greater than zero, e.g. 10.
- Negative number: A number less than zero, e.g. -3.
- Zero: The number 0, which is neither positive nor negative.
- Use case: Store someone's age or the number of items, e.g. items_in_cart = 4.

4. float: numbers with a decimal point
- float: Stores floating-point numbers, e.g. price = 19.99.
- Decimal number: A number with a decimal point, e.g. 3.14.
- Scientific notation: A shorter way to write very large or small numbers, e.g. 1e3 means 1000.0.
- 7.0: This is a float, even though its numerical value is a whole number.
- Use case: Store measurements or prices, e.g. temperature = 36.6.
- Binary precision: Computers cannot represent some decimal numbers exactly, which can cause tiny calculation differences.
- decimal.Decimal: Python's decimal-number tool for calculations requiring exact decimal precision, such as financial calculations.

5. bool: exactly two values
- bool: Stores only True or False.
- True: Represents a yes or positive condition, e.g. is_subscribed = True.
- False: Represents a no or negative condition, e.g. is_completed = False.
- Capital letters: Python requires True and False to start with capital letters.
- Use case: Check whether a task is completed or a user is subscribed.
- Boolean subclass of int: Python treats True like 1 and False like 0 in some numeric operations.

6. str: text
- str: Stores text, e.g. first_name = "Amara".
- String literal: Text written inside quotation marks, e.g. "Hello".
- Single quotes: You can write text using single quotes, e.g. 'Hello'.
- Double quotes: You can also write text using double quotes, e.g. "Hello".
- "7": This is text containing the character 7, not the integer 7.
- Use case: Store names, messages, labels, and other text.

7. None: no meaningful value available
- None: Represents the absence of a meaningful value, e.g. middle_name = None.
- NoneType: The data type of the special value None.
- Use case: Represent information that has not been provided or does not apply.
- Empty string (""): Represents text containing zero characters, e.g. middle_name = "".
- Difference: None means no meaningful value is available, while "" is an actual empty text value.

8. Using type() to check a value's type
- type(): Checks the data type of a value, e.g. type(25) returns <class 'int'>.
- <class 'int'>: Means the value belongs to the integer type.
- Example: type(19.99) returns <class 'float'>.
- Example: type("hello") returns <class 'str'>.
- Use case: Check the type of a value when you are unsure what kind of data you are working with.

9. A variable's type can change after reassignment
- Dynamic typing: Python allows a variable to refer to values of different types.
- Example: signup_status = "not started" assigns a string.
- Reassignment: signup_status = True changes the value to a Boolean.
- Important: Python variables do not permanently lock into their first assigned type.

10. None is not 0, False, or ""
- None: Represents the absence of a meaningful value.
- 0: Represents the actual number zero.
- False: Represents a Boolean negative value.
- "": Represents an empty string.
- ==: Checks whether two values are equal.
- None == 0: Returns False because None and zero are different values.
- None == False: Returns False because None and False are different values.
- None == "": Returns False because None and an empty string are different values.

Key takeaway: None, 0, False, and "" are four different values with four different meanings.
```

```
Example 1 — Four types and None together, inspected with type():
```

````python
age = 25
price = 19.99
is_subscribed = True
name = "Amara"
middle_name = None

print(age, type(age))
print(price, type(price))
print(is_subscribed, type(is_subscribed))
print(name, type(name))
print(middle_name, type(middle_name))
````

```
Example 2 — None representing "not entered yet," then reassigned:
```

````python
phone_number = None
print(phone_number)
print(type(phone_number))

phone_number = "555-0142"
print(phone_number)
print(type(phone_number))
````
