```
Complete Python dict Method Reference:

A dictionary (dict) stores data as key-value pairs. Python provides 11 public dictionary methods. Below, each method is explained in simple terms with a coding example.

1. fromkeys()
Meaning: Creates a new dictionary using the given keys and assigns the same initial value to every key.

It is called using dict.fromkeys(), not on an existing dictionary.
```

````python
keys = ["a", "b", "c"]

new_dict = dict.fromkeys(keys, 0)

print(new_dict)
````

```
- Does not modify an existing dictionary.
- Returns a new dictionary.

Important: Mutable default value warning

If you use a list as the default value, all keys share the same list.
```

````python
keys = ["a", "b"]

data = dict.fromkeys(keys, [])

data["a"].append(1)

print(data)
````

```
Both values changed because they refer to the same list object.

To create separate lists, use a dictionary comprehension:
```

````python
keys = ["a", "b"]

data = {key: [] for key in keys}

data["a"].append(1)

print(data)
````

```
5. items()
Meaning: Returns all dictionary key-value pairs as a view, where each pair is represented as a tuple.
```

````python
scores = {"Ada": 95, "Grace": 88}

print(scores.items())
````

```
You can use it in a loop:
```

````python
scores = {"Ada": 95, "Grace": 88}

for name, score in scores.items():
    print(name, score)
````

```
8. popitem()
Meaning: Removes and returns the last inserted key-value pair as a tuple
```

````python
data = {"a": 1, "b": 2, "c": 3}

last_item = data.popitem()

print(last_item)
print(data)
````

```
- Mutates the original dictionary.
- Returns the removed key-value pair as a tuple.
- Raises KeyError if the dictionary is empty.

Difference between pop() and popitem():
- pop("a") removes the key you specify.
- popitem() removes the most recently inserted key-value pair.

9. setdefault()
Meaning: Returns the existing value if a key exists; otherwise, inserts the key with a default value and returns that value.
```

````python
person = {"name": "Ada"}

age = person.setdefault("age", 0)

print(age)
print(person)
````

```
10. update()
Meaning: Adds multiple key-value pairs to an existing dictionary or updates the values of existing keys.
```

````python
person = {"name": "Ada"}

person.update({
    "age": 30,
    "city": "London"
})

print(person)
````

```
It can also overwrite existing values:
```

````python
person = {"name": "Ada", "age": 25}

person.update({"age": 30})

print(person)
````

```
- Mutates the original dictionary.
- Returns None.

11. values()
Meaning: Returns a view containing all dictionary values.
```

````python
scores = {"Ada": 95, "Grace": 88}

print(scores.values())
````

```
You can use it in a loop:
```

````python
scores = {"Ada": 95, "Grace": 88}

for score in scores.values():
    print(score)
````

```
Built-in Functions and Operators Used With Dictionaries:

These are not dictionary methods. They are Python built-in functions and operators that work with dictionaries.

1. len()
Meaning: Counts the number of key-value pairs in a dictionary.
```

````python
scores = {"Ada": 95, "Grace": 88, "Alan": 76}

print(len(scores))
````

```
2. list()
Meaning: Converts a dictionary's keys into a list.
```

````python
scores = {"Ada": 95, "Grace": 88, "Alan": 76}

print(list(scores))
````

```
3. sorted()
Meaning: Returns a sorted list of dictionary keys by default.
```

````python
scores = {"Ada": 95, "Grace": 88, "Alan": 76}

print(sorted(scores))
````

```
4. == (Equality operator)
Meaning: Checks whether two dictionaries contain the same keys and corresponding values, regardless of insertion order.
```

````python
dict1 = {"Ada": 95, "Grace": 88}

dict2 = {"Grace": 88, "Ada": 95}

print(dict1 == dict2)
````

```
5. dict()
Meaning: Creates a dictionary from an iterable containing two-item pairs.
```

````python
pairs = [("a", 1), ("b", 2)]

new_dict = dict(pairs)

print(new_dict)
````

```
6. Dictionary merge operator |
Meaning: Combines two dictionaries and returns a new dictionary without modifying either original dictionary.

Available in Python 3.9 and later.
```

````python
defaults = {"debug": False, "timeout": 30}

overrides = {"timeout": 60}

merged = defaults | overrides

print(merged)
print(defaults)
print(overrides)
````

```
When both dictionaries contain the same key, the value from the right-hand dictionary takes priority.

Difference between update() and |:
```

````python
# update() modifies the existing dictionary
defaults.update(overrides)

# | creates a new dictionary
merged = defaults | overrides
````

```
update() modifies the original dictionary and returns None.

| creates and returns a new dictionary.
```
