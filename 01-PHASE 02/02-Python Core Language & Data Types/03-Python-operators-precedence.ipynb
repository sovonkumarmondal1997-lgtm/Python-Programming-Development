```
Topic: Operators and Precedence

Operators perform actions on values (such as calculations and comparisons), while precedence determines the order in which Python performs those actions.

- Operator: A symbol or keyword that performs an action on values, e.g. + adds numbers.
- Operand: A value on which an operator performs an action, e.g. 3 and 4 in 3 + 4.
- Arithmetic operator: Performs mathematical calculations, e.g. +, -, *.
- Comparison operator: Compares two values and returns True or False, e.g. 5 > 3.
- Boolean operator: Combines or reverses conditions using and, or, and not.
- Membership operator: Checks whether a value exists inside another value, e.g. "a" in "apple".
- Identity operator: Checks whether two variables refer to the exact same object, e.g. x is y.
- Operator precedence: Determines which operator Python evaluates first when multiple operators appear in an expression, e.g. 2 + 3 * 4 gives 14.
```

```
Step-by-step explanation:
    
1. Operators and operands:
- Operator: A symbol or keyword that performs an action, e.g. + adds numbers.
- Operand: A value on which an operator performs an action, e.g. 3 and 4 in 3 + 4.
- Expression: A combination of values and operators that produces a result, e.g. 3 + 4 = 7.

2. Arithmetic operators:

Operator	    Simple explanation	                                 Example → Result
+	            Adds two values.	                                 3 + 4 → 7
-	            Subtracts one value from another.	                 10 - 3 → 7
*	            Multiplies two values.	                             6 * 7 → 42
/	            Divides and returns a float.	                     7 / 2 → 3.5
//	            Divides and rounds down to the nearest whole number  7 // 2 → 3
%	            Returns the remainder after division.	             7 % 2 → 1
**	            Raises a number to a power.	                         2 ** 3 → 8

important differences: /, //, and %:
- / (Division): Returns the division result as a float, e.g. 8 / 2 → 4.0.
- // (Floor division): Returns the result rounded down, e.g. 7 // 2 → 3.
- % (Modulo): Returns what remains after division, e.g. 7 % 2 → 1.
- Negative floor division: Python rounds down toward negative infinity, e.g. -7 // 2 → -4.
- Negative modulo: Follows Python's floor-division rule, e.g. -7 % 2 → 1.
- Using // and % together: Finds complete groups and leftover items, e.g. 23 items in boxes of 4 gives 5 full boxes and 3 leftovers.
```

```
3. Comparison operators:

==	  Checks whether two values are equal   5 == 5 → True
!=	  Checks whether two values are different   5 != 3 → True
<	  Checks whether the left value is smaller   3 < 5 → True
>	  Checks whether the left value is greater   5 > 3 → True
<=	  Checks whether the left value is smaller than or equal to the right   5 <= 5 → True
>=	  Checks whether the left value is greater than or equal to the right   6 >= 5 → True


- Comparison result: Comparison operators return either True or False.
- = vs ==: = assigns a value (age = 25), while == checks equality (age == 25).
- Different types with ==: Python allows comparisons between different types, e.g. 1 == 1.0 → True.
- Integer vs string: 5 == "5" returns False because an integer and a string are different values.
- String comparison: Checks whether two text values match, e.g. "python" == "python" → True.
- Incompatible comparisons: Comparing a string and integer with <, e.g. "5" < 5, causes a TypeError.
- String concatenation: + joins two strings, e.g. "Hello" + " World" → "Hello World".
- Incompatible addition: "Age" + 25 causes a TypeError because Python cannot directly add text and an integer.
- Chained comparison: Combines multiple comparisons without repeating a variable, e.g. 18 <= age < 65.
- Chained comparison example: If age = 25, then 18 <= age < 65 → True.
```

```
4. Boolean operators: and, or, not
- and: Returns True only when both conditions are true, e.g. True and False → False.
- or: Returns True when at least one condition is true, e.g. True or False → True.
- not: Reverses a Boolean value, e.g. not True → False.

Example:
```

````python
has_ticket = True
is_on_time = False

print(has_ticket and is_on_time)  # False
print(has_ticket or is_on_time)   # True
print(not is_on_time)             # True
````

```
- Falsy values: Values Python treats as false in conditions, such as False, 0, None, and "".
- Truthy values: Values Python treats as true in conditions, such as "hello" and 10.
- and behavior: Returns the first falsy value, or the last value if all are truthy, e.g. 0 and "hello" → 0.
- or behavior: Returns the first truthy value, or the last value if all are falsy, e.g. 0 or "hello" → "hello".
- Short-circuiting: Python stops evaluating once the result is already determined, e.g. False and x never evaluates x.

More examples:
```

````python
print(0 and "hello")          # 0
print("hello" and "world")    # world
print(0 or "fallback")        # fallback
print("hello" or "fallback")  # hello
````

```
5. Membership operators: in and not in:

- in: Checks whether some text exists inside another text, e.g. "cat" in "concatenate" → True.
- not in: Checks whether some text does not exist inside another text, e.g. "dog" not in "concatenate" → True.
- Use case: Check whether a password contains a particular character, e.g. "2" in "hunter2" → True.
- Other collections: in and not in can also check whether an item exists in a list.

6. Identity operators: is and is not, and the None rule:

- is: Checks whether two variables refer to the exact same object.
- is not: Checks whether two variables do not refer to the same object.
- == vs is: == checks whether values are equal, while is checks whether they are the same object.
- Ordinary comparisons: Use == for comparing numbers, strings, and Boolean values.
- Checking None: Use is None to check whether a variable contains None.

Example:
```

````python
score = None

if score is None:
    print("No score has been recorded yet.")
````

```
- Why is None: Python has only one None object, so is None checks whether the variable refers to that object.
- Beginner rule: Use == for comparing values and is specifically for checking None.

7. Operator precedence: which operator runs first:

- Operator precedence: Determines the order in which Python evaluates operators in an expression.
- Parentheses (): Have the highest priority and allow you to control the calculation order.
- Exponent **: Calculates powers before multiplication and addition, e.g. 2 ** 3 → 8.
- Unary minus -x: Makes a number negative, e.g. -5.
- Multiplication, division, floor division, modulo: These operators run before addition and subtraction.
- Addition and subtraction: Run after multiplication and division.
- Comparisons: Check equality, inequality, greater-than, less-than, membership, and identity.
- not: Reverses a condition after comparisons.
- and: Evaluates after not.
- or: Evaluates after and.
```

```
Precedence order (highest to lowest):

Priority	           Operators
1	                   Parentheses ()
2	                   Exponent **
3	                   Unary minus -x
4	                   *, /, //, %
5	                   +, -
6	                   Comparisons, in, not in, is, is not
7	                   not
8	                   and
9	                   or

Important examples:
- Multiplication before addition: 2 + 3 * 4 → 14, not 20.
- Exponent evaluated right to left: 2 ** 3 ** 2 → 512, meaning 2 ** (3 ** 2).
- Negative number with exponent: -2 ** 2 → -4, because Python calculates the exponent first.
- Parentheses change the result: (-2) ** 2 → 4, because -2 is treated as the base.
- and before or: age >= 18 or age >= 13 and has_guardian evaluates the and condition first.
- Using parentheses: age >= 18 or (age >= 13 and has_guardian) makes the intended evaluation order clear.
- Best practice: Use parentheses when an expression is complex or its evaluation order might be unclear.
```

```
Example 1 — Arithmetic: splitting items into boxes:
```

````python
total_items = 23
box_size = 4

full_boxes = total_items // box_size
leftover_items = total_items % box_size

print(full_boxes)
print(leftover_items)
````

```
Example 2 — Comparison, Boolean logic, and precedence together:
```

````python
age = 15
has_guardian = True

is_eligible = age >= 18 or (age >= 13 and has_guardian)
print(is_eligible)
````
