```
11. Useful built-in functions for lists:

Example 1 — len()
len() returns the total number of elements in a list.
```

````python
numbers = [1, 3, 5, 7, 9]

print(len(numbers))
````

```
Example 2 — min()
min() returns the smallest value in a list.
```

````python
numbers = [5, 2, 9, 1, 7]

print(min(numbers))
````

```
Example 3 — max()
max() returns the largest value in a list.
```

````python
numbers = [5, 2, 9, 1, 7]

maximum_number = max(numbers)

print(maximum_number)
````

```
Example 4 — sum()
sum() adds all numerical elements in a list.
```

````python
numbers = [5, 2, 9, 1, 8]

print(sum(numbers))
````

```
sorted() returns a new list containing elements in sorted order without changing the original list.
```

````python
numbers = [5, 2, 9, 1, 7]

result = sorted(numbers)

print(result)
print(numbers)
````

```
- A for loop visits each list element one by one.
- A loop variable temporarily holds the current element during each loop iteration.
```

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

```
enumerate() provides both the index and value while looping through a list.
```

````python
fruits = ["apple", "banana", "cherry"]

for position, fruit in enumerate(fruits):
    print(position, fruit)
````

```
Appendix: Complete Python list Method Reference:

Python lists have 11 public methods that allow you to add, remove, search, sort, reverse, and copy elements.

1. append(item)
Adds one new element to the end of the list.
```

````python
fruits = ["apple", "banana"]
print(fruits)

fruits.append("cherry")
print(fruits)
````

```
2. extend(iterable)
Adds multiple elements individually to the end of the list.
```

````python
fruits = ["apple", "banana"]

fruits.extend(["cherry", "date"])

print(fruits)
````

```
Difference between append() and extend()
```

````python
fruits1 = ["apple", "banana"]

fruits1.append(["cherry", "date"])

print(fruits1)


fruits2 = ["apple", "banana"]

fruits2.extend(["cherry", "date"])

print(fruits2)
````

```
Remember: append() adds the entire list as one element, whereas extend() adds its elements individually.
```

```
3. insert(index, item)
Adds an element at a specific position in the list.
```

````python
fruits = ["apple", "cherry"]

fruits.insert(1, "banana")

print(fruits)

fruits.insert(3, "banana")

print(fruits)
````

```
Remember: insert(1, "banana") places "banana" at index 1.

4. remove(value)
Removes the first matching value from the list.
```

````python
fruits = ["apple", "banana", "cherry"]

fruits.remove("banana")

print(fruits)
````

```
Example — Removing a duplicate value
```

````python
numbers = [10, 20, 10, 30]

numbers.remove(10)

print(numbers)
````

```
Remember: remove() removes the first matching value, not an element by it's index.

5. pop(index=-1)
Removes an element from the list and returns the removed value; by default, it removes the last element:

Example 1 — Removing the last element
```

````python
fruits = ["apple", "banana", "cherry"]
print(f"original list:", fruits)

removed_item = fruits.pop()

print(f"last item of the list:", removed_item)
print(f"updated list:", fruits)

values = [1, 3, 5, 11, 14, 7, 2]
print(f"updated list:", values)

remove_value = values.pop()

print(f"last value of the list:", remove_value)
print(f"updated list:", values)
````

```
Example 2 — Removing an element using its index
```

````python
fruits = ["apple", "banana", "cherry"]

removed_item = fruits.pop(0)

print(removed_item)
print(fruits)
````

```
Remember: pop() removes an element by position and gives you the removed value.

6. clear()
Removes all elements from the list, making it empty.
```

````python
fruits = ["apple", "banana", "cherry"]

fruits.clear()

print(fruits)
````

```
7. index(value)
Returns the position of the first matching value in the list.
```

````python
fruits = ["apple", "banana", "cherry"]

position = fruits.index("cherry")

print(position)
````

```
Example — Finding the first occurrence
```

````python
numbers = [10, 20, 10, 30]

print(numbers.index(10))
````

```
Remember: index() returns the position of the first matching element.

8. count(value)
Counts how many times a particular value appears in a list.
```

````python
numbers = [10, 20, 10, 30, 10]

result = numbers.count(10)

print(result)
````

```
Remember: count(10) tells you how many times 10 appears in the list.

9. reverse()
Reverses the order of elements in the original list.
```

````python
numbers = [10, 20, 30, 40]

numbers.reverse()

print(numbers)
````

```
10. sort()
Sorts the original list in ascending order by default.
Example 1 — Ascending order
```

````python
numbers = [30, 10, 40, 20]

numbers.sort()

print(numbers)
````

```
Example 2 — Descending order
Use reverse=True to sort elements from largest to smallest.
```

````python
numbers = [30, 10, 40, 20]

numbers.sort(reverse=True)

print(numbers)
````

```
11. copy()
Creates a separate shallow copy of the original list.
```

````python
original = [10, 20, 30]

duplicate = original.copy()

duplicate.append(40)

print(original)
print(duplicate)
````
