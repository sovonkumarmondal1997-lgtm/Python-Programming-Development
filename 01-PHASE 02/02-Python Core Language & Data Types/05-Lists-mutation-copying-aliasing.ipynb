```
Topic: Lists: Mutation, Copying, and Aliasing

- A Python list stores multiple values in an ordered collection, and understanding mutation, copying, and aliasing helps you know how changes to one list affect another.

KEY_TERMS:

- List: A collection of values stored in order that you can change, written inside [].
- Element / item: A single value stored inside a list.
- Mutable: Something whose contents can be changed after creation.
- Index: A number used to access an item's position in a list.
- Zero-based indexing: Python starts counting list positions from 0 instead of 1.
- Negative index: A position counted from the end of a list, where -1 means the last item.
- IndexError: An error that occurs when you try to access a list position that does not exist.
- Slice: A way to extract selected items from a list using [start:stop] or [start:stop:step].
- Mutation: Changing the contents of an existing list without creating a new list.
- In-place: Changing the original object directly instead of creating a new one.
- Alias: A second variable name that points to the exact same list as another variable.
- Copy: Creating a separate list with the same starting elements as the original.
- Shallow copy: Copying only the outer list while keeping its nested lists shared.
- Deep copy: Creating a copy of the list and its nested objects so they are independent.
```

```
Indexing means accessing an element from a list using its position.
```

````python
fruits = ["apple", "banana", "cherry"]

print(fruits[0])
````

```
Example 2 — Zero-based indexing
Python starts counting list positions from 0, not 1.
```

````python
fruits = ["apple", "banana", "cherry"]

print(fruits[0])  # First element
print(fruits[1])  # Second element
print(fruits[2])  # Third element
````

```
Negative indexing allows you to access elements starting from the end of a list.
```

````python
fruits = ["apple", "banana", "cherry"]

print(fruits[-1])  # Last element
print(fruits[-2])  # Second-last element
print(fruits[-3])  # Third-last element
````

```

An IndexError occurs when you try to access an index that does not exist.
```

````python
fruits = ["apple", "banana", "cherry"]

print(fruits[10])
````

```
Use len() to check whether an index exists before accessing it.
```

````python
fruits = ["apple", "banana", "cherry"]

position = 2

if 0 <= position < len(fruits):
    print(fruits[position])
else:
    print("Invalid position")
````

````python
fruits = ["apple", "banana", "cherry"]

position = 5

if 0 <= position < len(fruits):
    print(fruits[position])
else:
    print("Invalid position")
````

```
8. Aliasing: why second_list = first_list is not a copy:

Example 1 — Variable reference
A variable name refers to a list object stored in memory.
```

````python
first_list = [10, 20, 30]

print(first_list)
print(id(first_list))
````

```
Example 2 — Aliasing
Aliasing happens when two variables refer to the exact same list.
```

````python
first_list = [10, 20, 30]

second_list = first_list

print(first_list)
print(second_list)

print(first_list is second_list)

print(id(first_list))
print(id(second_list))
````

```
Example 3 — Shared mutation
Changing the list through one variable also affects the other variable because both refer to the same list.
```

````python
first_list = [10, 20, 30]

second_list = first_list

second_list.append(40)

print(first_list)
print(second_list)
````

```
Example 4 — Reassignment
Reassignment makes a variable refer to a different list.
```

````python
first_list = [10, 20, 30]

second_list = first_list

first_list = [9, 9, 9]

print(first_list)
print(second_list)
````

```
9. Copying: making an independent list on purpose:

Example 1 — Copying
Copying creates a separate list so that changes to its outer elements do not affect the original.
```

````python
original_list = [10, 20, 30]

copied_list = original_list.copy()

copied_list.append(40)

print(original_list)
print(copied_list)
````

```
Example 2 — .copy()
The .copy() method creates a separate shallow copy of a list.
```

````python
original_list = [1, 2, 3]

copy_a = original_list.copy()

print(copy_a)
````

```
Example 3 — Slicing copy
Using [:] creates a separate list containing all elements.
```

````python
original_list = [1, 2, 3]

copy_b = original_list[:]

copy_b.append(4)

print(original_list)
print(copy_b)
````

```
Example 4 — list()
The list() function creates a separate list from an existing list.
```

````python
original_list = [1, 2, 3]

copy_c = list(original_list)

copy_c.append(4)

print(original_list)
print(copy_c)
````

```
Example 5 — Independent lists
Changing a copied list does not change the original list's elements.
```

````python
original_list = [10, 20, 30]

copy_a = original_list.copy()

copy_a.append(99)

print(original_list)
print(copy_a)
````

```
10. Shallow-copy awareness: lists that contain other lists:

Example 1 — Nested list
A nested list is a list that contains other lists as elements.
```

````python
matrix = [[1, 2], [3, 4]]

print(matrix)
````

```
Example 2 — Shallow copy
A shallow copy creates a separate outer list but shares the inner lists with the original.
```

````python
matrix = [[1, 2], [3, 4]]

shallow = matrix.copy()

print(matrix)
print(shallow)
print(matrix is shallow)
````

```
Example 3 — Shared inner list
Changing an inner list through the shallow copy also changes the original.
```

````python
matrix = [[1, 2], [3, 4]]

shallow = matrix.copy()

shallow[0].append(99)

print(matrix)
print(shallow)
````

```
Example 4 — Adding a new inner list
Adding a new inner list to the copied outer list does not change the original outer list.
```

````python
matrix = [[1, 2], [3, 4]]

shallow = matrix.copy()

shallow.append([5, 6])

print(matrix)
print(shallow)
````

```
Example 5 — Deep copy
A deep copy creates a separate outer list and separate copies of its nested lists.
```

````python
import copy

matrix = [[1, 2], [3, 4]]

deep = copy.deepcopy(matrix)

deep[0].append(99)

print(matrix)
print(deep)
````

```
Example 6 — import copy
import copy loads Python's built-in copy module so you can use copy.deepcopy().
```

````python
import copy

original = [[10, 20], [30, 40]]

deep_copy = copy.deepcopy(original)

print(deep_copy)
````
