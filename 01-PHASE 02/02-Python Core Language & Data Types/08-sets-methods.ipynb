```
Appendix: Complete set Method Reference in Python:

Python sets have 17 public methods. Some methods modify the original set, while others return a new set or a value without modifying the original.

Part 1: All 17 Set Methods

1. add(value)
Purpose: Adds a new value to a set; if the value already exists, nothing changes.
```

````python
colors = {"red", "green"}

colors.add("blue")

print(sorted(colors))
````

```
Another example:
```

````python
colors = {"red", "green"}

colors.add("red")

print(sorted(colors))
````

```
2. clear()
Purpose: Removes all values from a set, making it empty.
```

````python
colors = {"red", "green", "blue"}

colors.clear()

print(colors)
````

```
3. copy()
Purpose: Creates a new, independent copy of an existing set.
```

````python
original = {1, 2, 3}

duplicate = original.copy()

duplicate.add(4)

print(sorted(original))
print(sorted(duplicate))
````

```
4. difference(other)
Purpose: Returns values that exist in the first set but not in the second set.
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a.difference(b)

print(sorted(result))

print(sorted(b.difference(a)))
````

```
Explanation: Difference depends on which set comes first.

Equivalent operator: a - b

5. difference_update(other)
Purpose: Removes from the original set all values that also exist in another set
```

````python
a = {1, 2, 3}

a.difference_update({2, 3})

print(a)
````

```
Explanation: Values 2 and 3 are removed directly from a.

6. discard(value)
Purpose: Removes a value if it exists; does not raise an error if it is missing.
```

````python
numbers = {1, 2, 3}

numbers.discard(2)

print(sorted(numbers))
````

```
What happens if the value does not exist?
```

````python
numbers = {1, 2, 3}

numbers.discard(99)

print(sorted(numbers))
````

```
Explanation: Nothing happens because 99 is not present.

7. intersection(other)
Purpose: Returns only the values that exist in both sets.
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a.intersection(b)

print(sorted(result))
````

```
Explanation: 2 and 3 are common to both sets.

Equivalent operator: a & b

8. intersection_update(other)
Purpose: Keeps only the values that exist in both sets and removes the remaining values from the original set.
```

````python
a = {1, 2, 3}

a.intersection_update({2, 3, 4})

print(sorted(a))
````

```
Explanation: 1 is removed because it is not present in the other set.

9. isdisjoint(other)
Purpose: Checks whether two sets have no common values.
```

````python
a = {1, 2}
b = {3, 4}

print(a.isdisjoint(b))
````

```
Another example:
```

````python
a = {1, 2}
b = {2, 3}

print(a.isdisjoint(b))
````

```
Explanation: The second example returns False because both sets contain 2.

10. issubset(other)
Purpose: Checks whether every value in the first set also exists in the second set.
```

````python
a = {1, 2}
b = {1, 2, 3}

print(a.issubset(b))
````

```
Another example:
```

````python
a = {1, 2, 4}
b = {1, 2, 3}

print(a.issubset(b))
````

```
Explanation: 4 is missing from b, so a is not a subset of b.

Equivalent operator: a <= b

11. issuperset(other)
Purpose: Checks whether the first set contains every value of the second set.
```

````python
a = {1, 2, 3}
b = {1, 2}

print(a.issuperset(b))
````

```
Another example:
```

````python
a = {1, 2}
b = {1, 2, 3}

print(a.issuperset(b))
````

```
Explanation: a does not contain 3, so it is not a superset of b.

Equivalent operator: a >= b

12. pop()
Purpose: Removes and returns an arbitrary value from a set.
```

````python
numbers = {10, 20, 30}

removed = numbers.pop()

print("Removed:", removed)
print("Remaining:", numbers)
````

```
13. remove(value)
Purpose: Removes a specific value from a set; raises an error if the value does not exist.
```

````python
numbers = {1, 2, 3}

numbers.remove(2)

print(sorted(numbers))
````

```
What happens if the value does not exist?
```

````python
numbers = {1, 2, 3}

numbers.remove(99)
````

```
Important difference: remove() vs discard()
```

````python
numbers = {1, 2, 3}

numbers.discard(99)  # No error

numbers.remove(99)   # Raises KeyError
````

```
- discard() → Does nothing if the value is missing.
- remove() → Raises an error if the value is missing.

14. symmetric_difference(other)
Purpose: Returns values that exist in exactly one of the two sets, excluding common values.
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a.symmetric_difference(b)

print(sorted(result))
````

```
Explanation: 1 and 4 are included, but 2 and 3 are excluded because they exist in both sets.

Equivalent operator: a ^ b

15. symmetric_difference_update(other)
Purpose: Updates the original set to contain only values that exist in exactly one of the two sets.
```

````python
a = {1, 2, 3}

a.symmetric_difference_update({2, 3, 4})

print(sorted(a))
````

```
Explanation: Common values 2 and 3 are removed, and 4 is added.

16. union(other)
Purpose: Combines all values from two sets into a new set without duplicates.
```

````python
a = {1, 2}
b = {2, 3}

result = a.union(b)

print(sorted(result))
````

```
Explanation: All unique values from both sets are included.

Equivalent operator: a | b

17. update(other)
Purpose: Adds all values from another iterable directly into the original set.
```

````python
a = {1, 2}

a.update({3, 4})

print(sorted(a))
````

```
Explanation: Values from the second set are added directly to a.

You can also pass a list:
```

````python
numbers = {1, 2}

numbers.update([3, 4, 5])

print(sorted(numbers))
````
