```
Topic: Python Files, the Interpreter, the REPL, and Indentation

This topic teaches you how Python code is saved in files, how Python runs that code, how to test code interactively, and how proper spacing (indentation) organizes your code and prevents errors.

KEY_TERMS:

- Python file: A file ending in .py that contains Python code.
- Interpreter: A program that reads and executes Python code.
- Terminal:	A text-based window where you type commands to your computer.
- REPL:	An interactive Python environment where you write code and see results immediately.
- Comment: A note written using # that Python ignores when running the code.
- Indentation: Spaces at the beginning of a line that tell Python which code belongs together.
- Code block: A group of related lines of code that belong together.
- SyntaxError: An error caused by writing code that breaks Python's grammar rules.
- IndentationError:	An error caused by incorrect or inconsistent spacing in your code.
```

```
Step-by-step explanation:

1. What a Python file actually is
A Python file is a text file ending in .py that contains Python code, which stays saved until the interpreter runs it.

Example: hello.py contains print("Hello!").

2. The interpreter's job
The interpreter reads your Python code, checks its structure, and executes the instructions.

Example: python3 hello.py tells Python to run the code inside hello.py.

3. Checking Python from a terminal
Use this command to check whether Python is installed and see its version.
python3 --version
Example output: Python 3.12.4
```

```
4. Creating and running a simple .py file:

- Create a file, write Python code, save it, and execute it using the terminal.

Example:
print("Hello, Python!")

Run it:
python3 hello.py

Output: Hello, Python!

Important points:
- Running the same file again executes the code again.
- Every normal script execution starts with fresh Python memory.
- Variables from the previous execution are not retained.
```

```
5. The REPL: an interactive Python session
REPL allows you to write Python code directly into the terminal and see the result immediately without creating a .py file.

Example:
>>> 2 + 24
>>> print("hi")
hi

Understanding the prompts:
- $ = Terminal is waiting for your command.
- >>> = Python REPL is waiting for your code.
- ... = Python is waiting for the remaining lines of a multi-line statement.
When to use REPL: To quickly test small pieces of Python code.

When not to use REPL: When creating permanent programs that you want to save and reuse.

Exiting REPL: Type exit() and press Enter.

Difference between .py file and REPL:

.py file	                        REPL
Saves code permanently.	            Does not automatically save code into a file.
Runs the saved program.	            Runs code immediately as you type.
Used for real programs.	            Used for quick experiments.
```

```
6. Comments: notes that Python ignores
Comments are notes written using # that explain code to humans but are ignored by Python.

Example:
```

````python
# This program prints a greeting.
print("Hello!")
````

```
7. Indentation and code blocks
Indentation means adding spaces at the beginning of a line to tell Python which instructions belong together in a code block.

Example:
```

````python
age = 20

if age >= 18:
    print("You are an adult.")
    print("You can vote.")

print("This always runs.")
````

```
Important rules:
- Python uses indentation instead of {} to define code blocks.
- Use 4 spaces for each indentation level.
- Always maintain consistent indentation.
- Never mix tabs and spaces.

8. Common errors from getting this wrong:

1. SyntaxError: Occurs when your code breaks Python's grammar rules.

Example:
```

````python
if age >= 18
    print("Adult")
````

```
Here, the colon : is missing after the if condition.

2. IndentationError: Occurs when the required spacing at the beginning of a line is incorrect.

Example:
```

````python
if age >= 18:
print("Adult")
````

```
Here, print() should be indented inside the if block.

3. Runtime Error: Occurs when Python encounters a problem while executing a statement.

Example:
```

````python
print(username)
````

```
If username has not been defined, Python raises a NameError.

Important difference:
Error	                  Meaning
SyntaxError	              Code grammar is incorrect.
IndentationError	      Code indentation is incorrect.
NameError	              Python cannot find the requested variable or name.
```

```
Example 2 — Variables and multiple instructions running in order:
```

````python
# Store two pieces of information, then use them together.
name = "Asha"
favorite_number = 7
print("This program belongs to", name)
print("Their favorite number is", favorite_number)
````

```
Explanation (one line each):
1. name = "Asha" → Creates a variable named name and stores the text "Asha" in it.
2. favorite_number = 7 → Creates a variable named favorite_number and stores the number 7 in it.
3. print("This program belongs to", name) → Prints the text and the value stored in name, separated by a space.
4. print("Their favorite number is", favorite_number) → Prints the text and the value stored in favorite_number.
5. Sequence → Python executes instructions one after another, following their written order.
```

```
Example 3 — A basic if block: indentation decides what belongs together:
```

````python
# Decide what to print based on one piece of information.
age = 20

if age >= 18:
    print("You are old enough to vote.")
    print("Both of these lines are inside the if block.")

print("This line always runs, no matter what age is.")
````

```
Explanation (one line each):
1. age = 20 → Creates a variable named age and stores the value 20.
2. if age >= 18: → Checks whether age is greater than or equal to 18.
3. age >= 18 → A condition that returns True if the age is 18 or above.
4. print("You are old enough to vote.") → Runs only when the if condition is True.
5. print("Both of these lines are inside the if block.") → Also runs when the if condition is True because it belongs to the same indented block.
6. print("This line always runs, no matter what age is.") → Runs regardless of whether the if condition is True or False because it is outside the if block.
7. Indentation → The four spaces before print() tell Python that the statements belong inside the if block.
```
