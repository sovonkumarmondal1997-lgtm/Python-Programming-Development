```
Topic: Sets and Unique Values

A set is a collection that automatically removes duplicate values, stores unique items without a fixed order, and makes it easy to check membership or compare two groups of values.

- Set: A collection of unique values that does not maintain a fixed order.
- Set literal: A set created directly using curly braces {} with comma-separated values.
- Unordered: Values have no fixed positions, such as first, second, or third.
- Membership: Checking whether a particular value exists in a collection using in or not in.
- Hashable: A value that Python can store in a set or use as a dictionary key.
- Union: Combines all unique values from two sets using \|.
- Intersection: Finds only the values that exist in both sets using &.
- Difference: Finds values that exist in the first set but not in the second using -.
- Symmetric difference: Finds values that exist in either set but not in both using ^.
- Subset: A set in which every value also exists in another set.
- Superset: A set that contains all the values of another set.
- Disjoint: Two sets that have no common values.
- frozenset: An immutable set whose values cannot be added, removed, or changed after creation.

```
Step-by-step explanation:

1. What a set is, and when it is useful:
- Set: A collection of unique values without a fixed order, for example, colors = {"red", "green", "blue"}.
- When to use a set: Use a set when you want to store unique items or quickly check whether a value exists, for example, tracking unique website visitors.

2. Sets store unique values only and don't allow duplicates:
```

````python
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)  # {1, 2, 3}
````

```
- Automatic duplicate removal: A set automatically keeps only one copy of each value, for example, {1, 1, 2, 2} becomes {1, 2}.

3. Creating sets
```

````python
colors = {"red", "green", "blue"}
print(colors)
````

```
- Set literal: Create a set directly using curly braces {}, for example, numbers = {1, 2, 3}.
- Creating an empty set: Use set() to create an empty set, for example, empty_set = set().
- Empty dictionary: {} creates an empty dictionary, not a set, for example, empty_dict = {}.
- Checking the data type: Use type() to identify the collection's type, for example, type(set()) returns <class 'set'>.
- Important difference: {1, 2, 3} creates a set, but {} creates a dictionary.

4. Membership checks
```

````python
colors = {"red", "green", "blue"}

print("red" in colors)         # True
print("purple" in colors)      # False
print("purple" not in colors)  # True
````

```
- in: Checks whether a value exists in a set, for example, "red" in colors returns True.
- not in: Checks whether a value does not exist in a set, for example, "purple" not in colors returns True.

5. Sets are unordered:
- Unordered: A set does not maintain a fixed position or insertion order, for example, {3, 1, 2} has no guaranteed display order.
- Indexing: Sets do not support accessing values using indexes, for example, colors[0] raises a TypeError.
- Slicing: Sets do not support extracting values using slices, for example, numbers[0:2] raises a TypeError.
- Printed order: Never depend on the order in which Python displays set values because it is not guaranteed.
- sorted(): Sorts set values and returns a new list, for example, sorted({5, 1, 3}) returns [1, 3, 5].
- Mixed data types: Python cannot directly sort incomparable values such as integers and strings, for example, sorted({1, "hello"}) raises a TypeError.

6. Adding, removing, and clearing values:
```

````python
colors = {"red", "green"}

colors.add("blue")
print(sorted(colors))  # ['blue', 'green', 'red']
print(type(colors))

colors.remove("red")
print(sorted(colors))  # ['blue', 'green']
print(type(colors))

colors.clear()
print(colors)  # set()
````

```
- .add(): Adds a new value to a set, for example, colors.add("blue").
- .remove(): Removes an existing value from a set, for example, colors.remove("red").
- .clear(): Removes all values from a set, for example, colors.clear() produces set().
- Mutate: Changing the original set directly using methods such as .add(), .remove(), or .clear() is called mutation.
- .discard(): Removes a value if it exists without raising an error if it is missing, for example, colors.discard("blue").

7. Set Operations in Python:
Set operations allow us to compare two sets, combine their values, and find common or different values.

A. Union (.union() and |)
Meaning: Union combines all unique values from two sets into one new set.

1. Using .union() method
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a.union(b)

print(result)
````

```
Explanation:
- a.union(b) combines all values from sets a and b.
- Duplicate values (2 and 3) are included only once.
- The result is stored in a new set called result.
- Original sets a and b remain unchanged.

2. Using | operator
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a | b
new_result = b | a

print(result)
print(new_result)
````

```
Explanation:
- | performs the same operation as .union().
- It combines all unique values from both sets.
- Both original sets remain unchanged.

Remember: a.union(b) and a | b perform the same union operation.

B. Intersection (.intersection() and &)
Meaning: Intersection finds only the values that exist in both sets.

1. Using .intersection() method
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a.intersection(b)
new_result = b.intersection(a)

print(result)
print(new_result)
````

```
Explanation:
- a.intersection(b) compares both sets.
- 2 and 3 exist in both sets.
- 1 and 4 are excluded because they are not common to both sets.
- The original sets remain unchanged.

2. Using & operator
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a & b
new_result = b & a

print(result)
print(new_result)
````

```
Explanation:
- & performs the same operation as .intersection().
- It returns only the common values from both sets.

Remember: a.intersection(b) and a & b perform the same intersection operation.

C. Difference (.difference() and -)
Meaning: Difference finds values that exist in the first set but do not exist in the second set.

1. Using .difference() method
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a.difference(b)
new_result = b.difference(a)

print(result)
print(new_result)
````

```
Explanation:
- a.difference(b) checks which values exist in set a but not in set b.
- 2 and 3 are removed because they also exist in b.
- Only 1 remains.
- The original sets remain unchanged.

Important: The order of sets matters.
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

print(a.difference(b))
print(b.difference(a))
````

```
Explanation:
- a.difference(b) returns {1} because 1 exists only in a.
- b.difference(a) returns {4} because 4 exists only in b.

2. Using - operator
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a - b
new_result = b - a

print(result)
print(new_result)
````

```
Explanation:
- - performs the same operation as .difference().
- It returns values present in the first set but absent from the second set.
Remember: a.difference(b) and a - b perform the same difference operation.

D. Symmetric Difference (.symmetric_difference() and ^)
Meaning: Symmetric difference finds values that exist in either set but excludes values that exist in both sets.

1. Using .symmetric_difference() method
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a.symmetric_difference(b)
new_result = b.symmetric_difference(a)

print(result)
print(new_result)
````

```
Explanation:
- a.symmetric_difference(b) compares both sets.
- 1 exists only in set a, so it is included.
- 4 exists only in set b, so it is included.
- 2 and 3 are excluded because they exist in both sets.
- The original sets remain unchanged.

2. Using ^ operator
```

````python
a = {1, 2, 3}
b = {2, 3, 4}

result = a ^ b
new_result = b ^ a

print(result)
print(new_result)
````

```
Explanation:
- ^ performs the same operation as .symmetric_difference().
- It returns values that exist in exactly one of the two sets.

Remember: a.symmetric_difference(b) and a ^ b perform the same symmetric difference operation.

8. Hashable Values in Python Sets:

1. Hashable
A hashable value is a value Python can store inside a set or use as a dictionary key.

Examples: integers, strings, and tuples containing hashable values.
```

````python
numbers = {10, 20, 30}
languages = {"Python", "Java"}
coordinates = {(2, 3), (4, 5)}

print(numbers)
print(languages)
print(coordinates)
````

```
2. Numbers
Numbers are hashable, so Python allows them inside sets.
```

````python
numbers = {10, 20, 30}

print(numbers)
````

```
3. Strings
Strings are hashable, so Python allows them inside sets
```

````python
languages = {"Python", "Java", "C++"}

print(languages)
````

```
4. Booleans
Boolean values (True and False) are hashable, so they can be stored inside sets.
```

````python
values = {True, False}

print(values)
````

```
Important: True and 1 are treated as equal in sets because True == 1.
```

````python
values = {True, 1, False, 0}

print(values)
````

```
Python keeps only one value from each equal pair.

5. Tuples
Tuples can be stored inside sets if all their elements are hashable.
```

````python
coordinates = {(1, 2), (3, 4)}

print(coordinates)
````

```
A tuple containing a list cannot be stored inside a set because lists are unhashable.
```

````python
values = {(1, 2), (3, [4, 5])}

print(values)
````

```
6. Lists
Lists cannot be stored directly inside sets because lists are unhashable.
```

````python
numbers = {[1, 2], [3, 4]}

print(numbers)
````

```
Why? Lists are mutable, meaning their contents can change after creation.

7. Dictionaries
Dictionaries cannot be stored directly inside sets because dictionaries are unhashable.
```

````python
students = {
    {"name": "Sam"},
    {"name": "John"}
}

print(students)
````

```
Why? Dictionaries are mutable, meaning their key-value pairs can change.

8. Why Hashability Matters:
Python uses hash values to locate set elements efficiently, so every set member must be hashable.

We can check the hash value using Python's hash() function.
```

````python
print(hash(10))
print(hash("Python"))
print(hash((2, 3)))
````

```
Note: The exact hash numbers may differ between Python runs, particularly for strings.

You can also check whether a value is hashable by trying hash():
```

````python
print(hash(100))       # Works
print(hash("Hello"))   # Works

print(hash([1, 2]))    # Raises TypeError
````

9. Immutable vs. Hashable:

Immutable means a value cannot be changed after creation, while hashable means Python can use its hash value to store it in a set or use it as a dictionary key.

1. Immutable — Cannot be changed after creation
Meaning: An immutable object cannot be modified after it is created; for example, you cannot change individual characters inside a string.
```

````python
name = "Python"

name[0] = "J"  # TypeError
````

```
Reason: Strings are immutable, so Python does not allow you to change an individual character.

However, you can create a new string:
```

````python
name = "Python"
name = "J" + name[1:]

print(name)  # Jython
````

```
Reason: Python creates a new string "Jython" and assigns it to the variable name instead of modifying the original string.
```

```
2. Hashable — Can be used as a dictionary key or set element:

Meaning: A hashable object has a hash value that remains constant during its lifetime and can be used as a dictionary key or set element.
```

````python
name = "Python"

hash_value = hash(name)

print(f"Python hash_value is:", hash_value)
````

```
Output: A hash integer (the exact value can vary between Python processes).

Reason: Strings are hashable, so Python can calculate their hash value.

Example: Using a hashable object as a dictionary key
```

9. Converting between lists and sets
```

````python
numbers = [3, 1, 2, 3, 1, 4]

unique_numbers = set(numbers)
print(unique_numbers)  # Some order of {1, 2, 3, 4}

back_to_list = list(unique_numbers)
print(back_to_list)  # Order is not guaranteed
````

```
- set(list): Converts a list into a set and automatically removes duplicates, for example, set([1, 2, 2, 3]) produces {1, 2, 3}.
- list(set): Converts a set back into a list, for example, list({1, 2, 3}) produces a list containing those values.
- Order loss: Converting a list into a set loses its original order because sets do not maintain insertion order.
- Important: Converting a set back into a list does not guarantee that the original list order will be restored.

10. frozenset: An optional, immutable set
```

````python
frozen = frozenset({1, 2, 3})
print(frozen)  # frozenset({1, 2, 3})

frozen.add(4)  # AttributeError
````

```
frozenset: Creates an immutable set whose values cannot be changed after creation, for example, frozenset({1, 2, 3}).

Immutable: You cannot add, remove, or clear values from a frozenset after creating it.

.add() on frozenset: Calling .add() raises an AttributeError because frozensets do not support adding values.

Membership: You can still check whether a value exists in a frozenset, for example, 2 in frozen returns True.

Set operations: Frozensets support operations such as union and intersection without modifying the original frozenset.
