```
Topics: Dictionaries and Lookups
A list finds data using its position (index), whereas a dictionary finds data using a meaningful name (key), such as "email", "name", or "age".

KEY_TERMS:

- Dictionary (dict): A collection that stores data as key-value pairs and lets you find values using keys.
- Key: The name used to find a particular value in a dictionary.
- Value: The actual data stored under a key.
- Key-value pair: A combination of a key and its associated value, such as "name": "Ada".
- Mapping: A way of connecting keys with their corresponding values.
- Hashable: An object that Python can use as a dictionary key because its hash value remains stable.
- KeyError: An error Python raises when you try to access a dictionary key that does not exist.
- Dictionary view: A live view of a dictionary's keys, values, or key-value pairs that reflects changes made to the dictionary.
- Mutation: Changing an existing dictionary by adding, updating, or removing data.
- Alias: Another variable name that points to the exact same dictionary.
- Shallow copy: A separate dictionary whose inner mutable values, such as lists, are still shared with the original.
- Insertion order: The order in which key-value pairs were added to a dictionary, which Python preserves.
```

```
1. What a dictionary is:

Explanation: A dictionary stores data in key-value pairs, allowing you to find values using their keys instead of positions.
```

````python
person = {"name": "Ada", "age": 30}
````

```
- "name" is the key, and "Ada" is its value.
- "age" is the key, and 30 is its value.

2. Creating empty and populated dictionaries
Explanation: An empty dictionary contains no data, while a populated dictionary already contains key-value pairs.
```

````python
empty_dict = {}                         # Empty dictionary
print(empty_dict)

scores = {"Ada": 95, "Grace": 88}       # Populated dictionary
print(scores)
````

```
- {} creates an empty dictionary, not an empty set.
- scores contains two key-value pairs.

3. Keys and values:
Explanation: Keys identify stored data, while values contain the actual information associated with those keys.

A. Valid dictionary keys

Explanation: Dictionary keys must be hashable, meaning Python can calculate a stable hash value for them.
```

````python
data = {
    "name": "Ada",
    10: "Ten",
    3.5: "Decimal",
    True: "Boolean",
    (1, 2): "Tuple"
}

print(data)
````

```
- Strings, integers, floats, booleans, and tuples containing hashable elements can be dictionary keys.

B. Invalid dictionary keys

Explanation: Lists and dictionaries cannot be dictionary keys because they are unhashable.
```

````python
bad_dict = {[1, 2]: "value"}
````

```
C. Keys are matched by hashing and equality:

Explanation: Python treats keys as the same when they have equal values and matching hash values, even if their types look different.
```

````python
data = {1: "integer"}

print(data[True])   # integer
print(data[1.0])    # integer
````

```
- Python treats True, 1, and 1.0 as equal dictionary keys because they have equal values and hashes, so both True and 1.0 retrieve the value "integer" stored under key 1

D. Why lists and dictionaries cannot be used as keys:

Explanation: Lists and dictionaries can change after creation, so Python does not allow them as keys because dictionary keys need stable hash values.

E. Keys must be unique:

Explanation: A dictionary cannot keep multiple entries with the same key, so the last assigned value replaces the previous one.
```

````python
data = {"a": 1, "a": 2}

print(data)   # {'a': 2}
````

```
The second "a": 2 replaces the first "a": 1.

Different keys can have the same value.
```

````python
data = {"a": 10, "b": 10}
````

```
4. Looking up values:

A. Direct lookup
Explanation: Use square brackets with a key to retrieve its corresponding value.
```

````python
person = {"name": "Ada", "age": 30}

print(person["name"])
print(person["age"])
````

```
B. KeyError
Explanation: Python raises KeyError when you try to access a dictionary key that does not exist.
```

````python
person = {"name": "Ada"}

print(person["missing"])
````

```
C. Safer lookup with .get()
Explanation: .get() retrieves a value if the key exists; otherwise, it returns None instead of raising an error.
```

````python
person = {"name": "Ada"}

print(person.get("missing"))   # None
````

```
D. .get() with a default value
Explanation: You can provide a default value to .get() that Python returns when the requested key does not exist.
```

````python
person = {"name": "Ada"}

print(person.get("missing", "N/A"))   # N/A
````

```
- Use .get() when a missing key is expected.
- Use direct lookup when a missing key should be treated as an error.

5. Adding and updating key-value pairs:
Explanation: Use dictionary[key] = value to add a new key or update the value of an existing key
```

````python
person = {"name": "Ada"}

person["age"] = 30
print(person)   # {'name': 'Ada', 'age': 30}

person["age"] = 31
print(person)   # {'name': 'Ada', 'age': 31}
````

```
- person["age"] = 30 adds a new key-value pair.
- person["age"] = 31 updates the existing value.

6. Removing values safely:

A. Using del
Explanation: del removes a key-value pair from a dictionary but raises KeyError if the key does not exist.
```

````python
person = {"name": "Ada", "age": 30}

del person["age"]

print(person)   # {'name': 'Ada'}
````

```
B. Using .pop()

Explanation: .pop() removes a specified key and returns its value, or returns a default value if the key does not exist.
```

````python
person = {"name": "Ada"}

removed = person.pop("age", "not found")

print(removed)   # not found
print(person)    # {'name': 'Ada'}
````

```
7. Checking whether a key exists:
Explanation: Use in to check whether a key exists and not in to check whether it does not exist.
```

````python
person = {"name": "Ada", "age": 30}

print("name" in person)       # True
print("email" in person)      # False
print("email" not in person)  # True
````

```
- in returns True if the key exists.
- not in returns True if the key does not exist.
- Both check dictionary keys, not values.

Checking before direct lookup:
Explanation: Use in before accessing a key to prevent KeyError when the key might be missing.
```

````python
if "email" in person:
    print(person["email"])
````

```
Explanation: "email" in person checks whether the "email" key exists before accessing it, preventing a KeyError if the key is missing.

Output: No output, because "email" does not exist in the dictionary.
```

```
8. Iterating through a dictionary:

A. Looping directly through a dictionary
Explanation: A for loop directly over a dictionary visits its keys one by one.
```

````python
scores = {"Ada": 95, "Grace": 88, "Alan": 76}

for name in scores:
    print(name)
````

```
B. Using .keys()
Explanation: .keys() returns all dictionary keys, allowing you to loop through them explicitly.
```

````python
for name in scores.keys():
    print(name)
````

```
C. Using .values()
Explanation: .values() returns all dictionary values, allowing you to loop through them.
```

````python
for score in scores.values():
    print(score)
````

```
D. Using .items()
Explanation: .items() returns both keys and values together, allowing you to access them in the same loop.
```

````python
for name, score in scores.items():
    print(name, score)
````

```
- name receives the key.
- score receives its corresponding value.
- .items() is the clearest choice when you need both.

9. Dictionary ordering:
Explanation: Python 3.7+ preserves insertion order, meaning dictionary entries appear in the same order they were added, not automatically sorted.
```

````python
data = {}

data["z"] = 1
data["a"] = 2
data["m"] = 3

print(data)
````

```
- "z" was added first, so it appears first.
- "a" was added second.
- "m" was added third.
- Do not depend on insertion order for business logic unless it is intentionally part of your program's design.

10. Copying: aliasing versus .copy()

A. Aliasing
Explanation: Assigning one dictionary to another variable creates an alias, meaning both variables point to the same dictionary.
```

````python
original = {"a": 1}
alias = original

alias["b"] = 2

print(original)   # {'a': 1, 'b': 2}
print(alias)      # {'a': 1, 'b': 2}
````

```
- Changing alias also changes original because both refer to the same dictionary.

B. Using .copy()
Explanation: .copy() creates a separate dictionary, so adding or removing top-level entries in the copy does not affect the original.
```

````python
original = {"a": 1, "b": 2}

copy_dict = original.copy()
copy_dict["c"] = 3

print(original)    # {'a': 1, 'b': 2}
print(copy_dict)   # {'a': 1, 'b': 2, 'c': 3}
````

```
C. Shallow copying

Explanation: A shallow copy creates a separate outer dictionary, but mutable objects stored inside it, such as lists, are still shared.
```

````python
original = {"numbers": [1, 2, 3]}

copy_dict = original.copy()

copy_dict["numbers"].append(4)

print(original)    # {'numbers': [1, 2, 3, 4]}
print(copy_dict)   # {'numbers': [1, 2, 3, 4]}
````

```
- Both dictionaries share the same inner list, so changing that list affects both.

11. Practical use cases:

A. Product and quantity record
Explanation: Dictionaries store product names as keys and their available quantities as values.
```

````python
inventory = {"apples": 50, "bread": 12}

print(inventory["apples"])   # 50
````

```
B. User profile
Explanation: Dictionaries store a user's information using meaningful keys such as name, email, and age.
```

````python
user = {
    "name": "Ada",
    "email": "ada@example.com"
}

print(user["name"])   # Ada
````

```
C. Counting simple categories
Explanation: Dictionaries count how many times each category appears by storing the category as a key and its count as a value.
```

````python
purchases = ["fruit", "dairy", "fruit"]

counts = {}

for category in purchases:
    if category in counts:
        counts[category] += 1
    else:
        counts[category] = 1

print(counts)
# {'fruit': 2, 'dairy': 1}
````

```
D. Safe configuration lookup with defaults
Explanation: Dictionaries store application settings and .get() provides a default value when a setting is missing.
```

````python
config = {"debug": True, "max_retries": 3}

timeout = config.get("timeout", 30)

print(timeout)   # 30
````

```
- Since "timeout" does not exist, Python returns the default value 30.
```

```
Example 1 — A product and quantity record:
```

````python
inventory = {"apples": 50, "bananas": 30, "bread": 12}
print(f"Original inventory:", inventory)

inventory["apples"] -= 5
print(f"1st updated inventory:", inventory)

inventory["milk"] = 20
print(f"2nd updated inventory:", inventory)

print(inventory.get("eggs", 0))
````

```
Example 2 — Counting simple categories:
```

````python
purchases = ["fruit", "dairy", "fruit", "bakery", "dairy", "fruit"]

category_counts = {}

for category in purchases:
    if category in category_counts:
        category_counts[category] += 1
    else:
        category_counts[category] = 1

print(category_counts)
````

```
Example 3 — Safe configuration lookup with defaults:
```

````python
config = {"debug": True, "max_retries": 3}

timeout = config.get("timeout", 30)
max_retries = config.get("max_retries", 5)

print(f"Timeout: {timeout}")
print(f"Max retries: {max_retries}")

for setting, value in config.items():
    print(f"{setting} = {value}")
````

```
Explanation of each line:
- Creates a dictionary named config containing two settings: debug and max_retries.
- Searches for the "timeout" key and returns 30 as the default value because the key does not exist.
- Searches for the "max_retries" key and returns 3 because the key already exists, ignoring the default value 5.
- Prints the value stored in the timeout variable, which is 30.
- Prints the value stored in the max_retries variable, which is 3.
- Loops through the dictionary using items() to retrieve each key and its corresponding value.
- Prints each dictionary key and its corresponding value during every loop iteration.
```
