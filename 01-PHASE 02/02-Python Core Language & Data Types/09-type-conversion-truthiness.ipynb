```
Python: Type Conversion and Truthiness: PART 1

1. Type conversion:
Type conversion means changing a value from one data type into another data type.
For example, converting "25" from a string into the integer 25.

A. Implicit conversion:
Implicit conversion: Python automatically converts a value's type when it can safely combine compatible types.
```

````python
quantity = 3       # int
price = 2.5        # float

total = quantity * price

print(total)       # 7.5
print(type(total)) # <class 'float'>
````

```
Explanation:
- quantity is an integer.
- price is a floating-point number.
- Python automatically handles the numeric conversion during multiplication.
- The result is a float.

B. Explicit conversion:
Explicit conversion: You manually convert a value using a conversion function such as int(), float(), or str().
```

````python
age = "25"          # str

age = int(age)      # Convert str into int

print(age + 5)      # 30
print(type(age))    # <class 'int'>
````

```
Explanation:
- Initially, age contains the string "25".
- int(age) converts it into the integer 25.
- Now Python can perform numeric calculations with it.

C. Compatible and incompatible types:
Python can automatically combine some compatible types, but not every type.
```

````python
# Compatible types
number = 5
decimal = 2.5

print(number + decimal)  # 7.5

# Incompatible types
age = "25"

print(age + 5)           # TypeError
````

```
Explanation:
- int + float works because Python supports this numeric combination.
- str + int does not work because Python will not automatically guess how you want to convert the values.

To fix the second example:
```

````python
age = "25"

print(int(age) + 5)  # 30
````

```
2. Numeric conversion using int() and float():

A. int() conversion
int() converts suitable numeric text or a floating-point number into an integer.
```

````python
# Convert string into integer
number = int("42")

print(number)       # 42
print(type(number)) # <class 'int'>
````

```
Here, "42" is converted from str into int.

B. Converting a float into an integer
```

````python
price = 19.99

result = int(price)

print(result)  # 19
````

```
Python removes the decimal portion instead of rounding the number.

C. Truncation
Truncation means removing the decimal portion of a number without rounding it.
```

````python
print(int(5.9))   # 5
print(int(5.1))   # 5
print(int(9.99))  # 9
````

```
D. Truncation toward zero
Truncation toward zero means removing the decimal portion while moving the number toward zero, whether it is positive or negative.
```

````python
print(int(3.9))    # 3
print(int(-3.9))   # -3
print(int(-9.99))  # -9
````

```
It produces -3, not -4, because int() truncates toward zero rather than rounding down.

E. int() with a number base
A number base tells Python which number system the text represents.

For example, binary uses base 2.
```

````python
binary_number = "101"

result = int(binary_number, 2)

print(result)  # 5
````

```
Explanation:
- "101" is binary text.
- 2 tells Python to interpret it as a base-2 number.
- Its decimal equivalent is 5.

Another example:
```

````python
print(int("10", 2))  # 2
print(int("10", 8))  # 8
print(int("10", 16)) # 16
````

```
The same text produces different results because the number base changes.

F. float() conversion
float() converts suitable numeric text or an integer into a floating-point number.
```

````python
# String to float
price = float("19.99")

print(price)       # 19.99
print(type(price)) # <class 'float'>
````

````python
#Integer to float:
number = 7

result = float(number)

print(result)       # 7.0
print(type(result)) # <class 'float'>
````

```
5. Truthiness and falsiness:

A. Truthiness
Truthiness is Python's mechanism for treating a value as either True or False when making a decision.

A value does not have to be a Boolean to behave like one inside a condition.
```

````python
name = "Python"

if name:
    print("Value exists")
````

```
The string "Python" is truthy, so the if block executes.

B. Truthy values
Truthy values behave like True inside a condition.

Examples:
```

````python
print(bool("hello"))  # True
print(bool(10))       # True
print(bool(3.14))     # True
print(bool([1, 2]))   # True
print(bool((1, 2)))   # True
````

```
A few simple rules:
- Non-empty strings are truthy.
- Non-zero numbers are truthy.
- Non-empty built-in collections are truthy.
```

```
C. Falsy values:
- Falsy values behave like False inside a condition.

The standard falsy values covered in this lesson are:

Value             Example               Result
False             bool(False)           False
None              bool(None)            False
Integer zero      bool(0)               False
Float zero        bool(0.0)             False
Empty string      bool("")              False
Empty list        bool([])              False
Empty tuple       bool(())              False
Empty dictionary  bool({})              False
Empty set         bool(set())           False

Example:
```

````python
print(bool(False))
print(bool(None))
print(bool(0))
print(bool(0.0))
print(bool(""))
print(bool([]))
print(bool(()))
print(bool({}))
print(bool(set()))
````

```
D. Custom object truth-value behavior:

Custom Python objects can define their own truthiness behavior.

For example, a class can define the special method __bool__() to decide whether its objects should behave as True or False.
```

````python
class User:
    def __init__(self, active):
        self.active = active

    def __bool__(self):
        return self.active


user1 = User(True)
user2 = User(False)

print(bool(user1))  # True
print(bool(user2))  # False
````

```
Explanation:
- User is a custom class.
- __bool__() decides the truth value of its objects.
- user1 behaves as True.
- user2 behaves as False.

This is an example of how Python allows custom objects to define their own truthiness.
```

```
6. bool() conversion:

A. bool() conversion
bool() converts a value into either True or False according to Python's truthiness rules.
```

````python
print(bool(1))       # True
print(bool(0))       # False

print(bool("Hello")) # True
print(bool(""))      # False

print(bool([1, 2]))  # True
print(bool([]))      # False
````

```
B. Truthiness inside if statements
An if statement can directly evaluate a value without requiring an explicit Boolean comparison.

Example:
```

````python
name = "Sovon"

if name:
    print("Name is available")
else:
    print("Name is missing")
````

```
Why?
- name contains "Sovon".
- A non-empty string is truthy.
- Therefore, the if block executes.

Compare it with an empty string:
```

````python
name = ""

if name:
    print("Name is available")
else:
    print("Name is missing")
````

```
The empty string is falsy, so the else block executes.

Important: You do not need to write if bool(name):. Python already checks truthiness inside if.
```

```
7. Choosing between if value: and if value is None:

This is an important concept because both statements check different things.
A. Falsy-value ambiguity
Falsy-value ambiguity means different values can behave as False even though they have different meanings.

Consider:
```

````python
print(bool(0))       # False
print(bool(False))   # False
print(bool(None))    # False
print(bool(""))      # False
````

```
B. The problem with if value:
Consider this example:
```

````python
score = 0

if score:
    print("Score is available")
else:
    print("Score is missing")
````

```
But this is misleading.

The score is actually 0, not missing.

Why does this happen?

Because 0 is falsy, so Python executes the else block.
```

```
C. Explicit absence checking using is None
Use is None when you specifically want to check whether a value is missing.
```

````python
score = 0

if score is None:
    print("Score is missing")
else:
    print(f"Score is {score}")
````

```
D. Meaningful empty values
An empty value can be valid application data.

Example:
```

````python
username = ""

if username is None:
    print("Username was not provided")
else:
    print("Username value exists")
````

```
Here:
- "" means the value is an empty string.
- None means no value is available.
- They are different values, so the condition treats them differently.

Compare:
```

````python
username = None

if username is None:
    print("Username was not provided")
else:
    print("Username value exists")
````
