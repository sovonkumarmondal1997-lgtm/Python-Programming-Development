```
Python: Type Conversion and Truthiness:

1. Type conversion:
Type conversion means changing a value from one data type into another data type.
For example, converting "25" from a string into the integer 25.

A. Implicit conversion:
Implicit conversion: Python automatically converts a value's type when it can safely combine compatible types.
```

````python
quantity = 3       # int
price = 2.5        # float

total = quantity * price

print(total)       # 7.5
print(type(total)) # <class 'float'>
````

```
Explanation:
- quantity is an integer.
- price is a floating-point number.
- Python automatically handles the numeric conversion during multiplication.
- The result is a float.

B. Explicit conversion:
Explicit conversion: You manually convert a value using a conversion function such as int(), float(), or str().
```

````python
age = "25"          # str

age = int(age)      # Convert str into int

print(age + 5)      # 30
print(type(age))    # <class 'int'>
````

```
Explanation:
- Initially, age contains the string "25".
- int(age) converts it into the integer 25.
- Now Python can perform numeric calculations with it.

C. Compatible and incompatible types:
Python can automatically combine some compatible types, but not every type.
```

````python
# Compatible types
number = 5
decimal = 2.5

print(number + decimal)  # 7.5

# Incompatible types
age = "25"

print(age + 5)           # TypeError
````

```
Explanation:
- int + float works because Python supports this numeric combination.
- str + int does not work because Python will not automatically guess how you want to convert the values.

To fix the second example:
```

````python
age = "25"

print(int(age) + 5)  # 30
````

```
2. Numeric conversion using int() and float():

A. int() conversion
int() converts suitable numeric text or a floating-point number into an integer.
```

````python
# Convert string into integer
number = int("42")

print(number)       # 42
print(type(number)) # <class 'int'>
````

```
Here, "42" is converted from str into int.

B. Converting a float into an integer
```

````python
price = 19.99

result = int(price)

print(result)  # 19
````

```
Python removes the decimal portion instead of rounding the number.

C. Truncation
Truncation means removing the decimal portion of a number without rounding it.
```

````python
print(int(5.9))   # 5
print(int(5.1))   # 5
print(int(9.99))  # 9
````

```
D. Truncation toward zero
Truncation toward zero means removing the decimal portion while moving the number toward zero, whether it is positive or negative.
```

````python
print(int(3.9))    # 3
print(int(-3.9))   # -3
print(int(-9.99))  # -9
````

```
It produces -3, not -4, because int() truncates toward zero rather than rounding down.

E. int() with a number base
A number base tells Python which number system the text represents.

For example, binary uses base 2.
```

````python
binary_number = "101"

result = int(binary_number, 2)

print(result)  # 5
````

```
Explanation:
- "101" is binary text.
- 2 tells Python to interpret it as a base-2 number.
- Its decimal equivalent is 5.

Another example:
```

````python
print(int("10", 2))  # 2
print(int("10", 8))  # 8
print(int("10", 16)) # 16
````

```
The same text produces different results because the number base changes.

F. float() conversion
float() converts suitable numeric text or an integer into a floating-point number.
```

````python
# String to float
price = float("19.99")

print(price)       # 19.99
print(type(price)) # <class 'float'>
````

````python
#Integer to float:
number = 7

result = float(number)

print(result)       # 7.0
print(type(result)) # <class 'float'>
````

```
3. String conversion using str():

A. str() conversion
str() converts a value into its text representation.
```

````python
age = 25

result = str(age)

print(result)       # 25
print(type(result)) # <class 'str'>
````

```
Although the output looks like 25, its type is now str.

More examples:
```

````python
print(str(100))     # "100"
print(str(19.99))   # "19.99"
print(str(True))    # "True"
print(str(None))    # "None"
````

```
B. String conversion for concatenation:
Concatenation means joining strings together.
```

````python
age = 25

print("My age is " + str(age))
````
