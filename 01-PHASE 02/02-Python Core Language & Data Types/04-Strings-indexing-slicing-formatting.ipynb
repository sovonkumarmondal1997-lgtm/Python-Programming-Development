```
Topic: Strings: Indexing, Slicing, and Formatting

Strings (str) are used to store and work with text in Python, while indexing and slicing help you access specific characters or parts of that text, and formatting helps display it neatly.
```

```
Key Terms — Simple Explanations with Examples:

1. String (str): A string is text enclosed in quotation marks, such as "Hello".

2. Character: A character is a single letter, number, symbol, or space, such as "A".

3. Method: A method is an action performed on a value using a dot, such as "hello".upper().

4. Index: An index is a number that identifies a character's position in a string, such as text[1].

5. Zero-based indexing: Python starts counting string positions from 0, so "Python"[0] returns "P".

6. Negative index: A negative index counts positions from the end, so "Python"[-1] returns "n".

7. IndexError: An error that occurs when you access a character position that does not exist, such as "Cat"[5].

8. Slice: A slice extracts a specific part of a string, such as "Python"[0:3] returning "Pyt".

9. Step: A step controls how many positions Python skips while extracting characters, such as "Python"[::2] returning "Pto".

10. Immutable: Immutable means a string cannot be changed directly after creation, such as "Hello" remaining unchanged when you create a modified copy.

11. Escape sequence: An escape sequence uses a backslash to represent special characters, such as "\n" for a new line.

12. Concatenation: Concatenation means joining strings using +, such as "Hello" + " World".

13. Membership: Membership checks whether some text exists inside another string, such as "Py" in "Python" returning True.

14. f-string: An f-string allows you to insert Python values or expressions directly into text, such as f"Age: {25}".

15. Placeholder: A placeholder is {} inside a formatted string where a value is inserted, such as f"Hello, {name}".

16. Format specifier: A format specifier controls how a value is displayed, such as f"{3.14159:.2f}" returning "3.14".

17. Unicode: Unicode is a standard that assigns codes to characters from different languages and symbols, such as "নমস্কার" and "Hello".

18. Case folding: Case folding converts text into a form suitable for case-insensitive comparison, such as "HELLO".casefold() returning "hello".

19. Prefix: A prefix is the text at the beginning of a string, such as "Py" in "Python".

20. Suffix: A suffix is the text at the end of a string, such as "on" in "Python".
```

```
3. Strings Are Immutable:

1. Immutable
A string cannot be modified directly after it is created.
```

````python
name = "Sam"

name[0] = "J"

print(name)
````

```
You cannot directly replace a character inside a string.

2. String Methods
String methods return a new string without changing the original string.
```

````python
name = "hello"

result = name.upper()

print(result)
print(name)
````

```
upper() returns a new uppercase string; the original name remains unchanged.

3. Reassignment
You can store the returned string back in the same variable.
```

````python
name = "hello"

name = name.upper()

print(name)
````

```
Here, name now refers to the new string "HELLO".

4. Discarded Result
If you call a string method without storing or using its result, the returned string is discarded.
```

````python
name = "hello"

name.upper()

print(name)
````

```
pper() creates "HELLO", but because you did not store the result, name remains "hello".

Quick revision:

Immutable: You cannot directly modify a string.
String methods:	Return a new string.
Reassignment: Stores the returned string in a variable.
Discarded result: The returned string is not kept.
```

```
4. String Indexing:

1. Indexing
Indexing means accessing an individual character using its position.
```

````python
language = "python"

print(language[0])
````

```
2. Zero-Based Indexing
Python starts counting character positions from 0, not 1.
```

````python
language = "python"

print(language[0])
print(language[1])
print(language[2])
````

```
3. Positive Indexing
Positive indexing counts characters from the beginning of a string.
```

````python
language = "python"

print(language[2])
print(language[3])
print(language[4])
````

```
4. Negative Indexing
Negative indexing counts characters backward from the end of a string.
```

````python
language = "python"

print(language[-1])
print(language[-2])
print(language[-3])
````

```
7. IndexError
An IndexError occurs when you try to access a position that does not exist.
```

````python
language = "python"

print(language[10])
````

```
8. len() with Indexing
The len() function returns the total number of characters in a string.
```

````python
language = "python"

print(len(language))
````

```
You can use len() to identify the last valid positive index:
```

````python
language = "python"

print(language[len(language) - 1])
````

```
9. Safe Indexing
Safe indexing checks whether a position exists before accessing it.
```

````python
username = "Sovon"

position = 2

if 0 <= position < len(username):
    print(username[position])
else:
    print("Invalid position")
````

```
If the position does not exist:
```

````python
username = "Sovon"

position = 10

if 0 <= position < len(username):
    print(username[position])
else:
    print("Invalid position")
````

```
5. String Slicing:

1. Slicing
Slicing extracts a portion of a string using [start:stop].
```

````python
language = "python"

print(language[0:3])
````

```
2. Start
The start index specifies where the extraction begins.
```

````python
language = "python"

print(language[2:5])
````

```
3. Stop
The stop index specifies where extraction ends, but that position is excluded.
```

````python
language = "python"

print(language[0:3])
````

```
Extraction stops before index 3.

4. Omitting Stop
If you omit the stop index, Python extracts from the starting position to the end.
```

````python
language = "python"

print(language[2:])
````

```
5. Omitting Start
If you omit the start index, Python extracts from the beginning up to the stop position.
```

````python
language = "python"

print(language[:4])
````

```
6. Omitting Both
If you omit both start and stop, Python extracts the entire string.
```

````python
language = "python"

print(language[:])
````

```
7. Stop Exclusion
The stop index is never included in the extracted result.
```

````python
language = "python"

print(language[0:3])
````

```
Here, Python extracts indexes 0, 1, and 2, but excludes index 3.

8. Negative Slicing
Negative indexes allow you to extract characters using positions counted from the end.
```

````python
language = "python"

print(language[-3:])
````

```
9. Step
The step controls how many positions Python moves between extracted characters.

Syntax:
string[start:stop:step]

Example:
```

````python
language = "python"

print(language[::2])
````

```
10. Positive Step
A positive step moves forward through the string.
```

````python
language = "python"

print(language[::2])
print(language[::3])
````

```
- ::2 selects every second character.
- ::3 selects every third character.

11. Negative Step
A negative step moves backward through the string.
```

````python
language = "python"

print(language[::-1])
````

```
12. Reversing a String
You can reverse an entire string using the slicing step -1.
```

````python
message = "hello"

reversed_message = message[::-1]

print(reversed_message)
````

```
Quick Revision: String Slicing:

Expression	         Meaning	               Output
"python"[0:3]	     Extract index 0 to 2	   pyt
"python"[2:5]	     Start at index 2	       tho
"python"[2:]	     From index 2 to end	   thon
"python"[:4]	     Beginning to index 3	   pyth
"python"[:]	         Entire string	           python
"python"[-3:]	     Last 3 characters	       hon
"python"[::2]	     Every second character	   pto
"python"[::-1]	     Reverse the string	       nohtyp

Remember: start is included, stop is excluded, and step controls the movement between characters.
```

```
Format Specifiers:

5. Format Specifier (:)
Provides instructions for how a value should be displayed inside an f-string.
```

````python
price = 125.5678

print(f"{price:.2f}")
````

```
6. .2f
Displays a number with exactly two decimal places.
```

````python
price = 12.5

print(f"{price:.2f}")
````

```
7. ,.2f
Adds thousands separators and displays two decimal places.
```

````python
price = 1234.5

print(f"{price:,.2f}")
````

```
8. .1%
Converts a decimal into a percentage with one decimal place.
```

````python
score = 0.4567

print(f"{score:.1%}")
````

```
9. Right Alignment (>)
Aligns text to the right within a specified width.
```

````python
name = "Total"

print(f"{name:>10}")
````

```
10. Left Alignment (<)
Aligns text to the left within a specified width
```

````python
name = "Total"

print(f"{name:<10}")
````

```
11. Center Alignment (^)
Places text in the center within a specified width.
```

````python
name = "Total"

print(f"{name:^10}")
````

```
12. Width (10)
Specifies the minimum total display width of a value.
```

````python
message = "Hi"

print(f"{message:>10}")
````

```
Inserts values into {} placeholders using the .format() method.
```

````python
message = "Hello {}".format("Ada")

print(message)
````

```

Insert values according to their argument order.
```

````python
message = "{} is {} years old".format("Ada", 30)

print(message)
````

```
Insert values using matching parameter names.
```

```
16. % Formatting
An older method of inserting values into strings.
```

````python
name = "Ada"
age = 30

message = "%s is %d years old" % (name, age)

print(message)
````

```
17. %s
Represents a value formatted as text.
```

````python
message = "%s" % "Hello"

print(message)
````

```
18. %d
Represents a whole number (integer).
```

````python
age = 25

print("%d" % age)
````

```
Legacy formatting means older formatting techniques that you may encounter in existing Python code; prefer f-strings for new code.
```

````python
name = "Ada"
age = 30

# Old-style formatting
print("%s is %d years old" % (name, age))

# Modern f-string formatting
print(f"{name} is {age} years old")
````

```
8. Unicode Awareness:

1. Unicode
Unicode is a standard that allows Python strings to represent characters from different languages, symbols, and emojis.
```

````python
message = "Hello नमस्ते ☕"

print(message)
````

```
2. Unicode Code Point
A Unicode code point is a unique number assigned to a character.
```

````python
character = "A"

print(ord(character))
print(hex(ord(character)))
````

```
A has the Unicode code point U+0041.

3. len() with Unicode
The len() function counts Unicode code points, not necessarily the visible characters.
```

````python
text = "café"

print(len(text))
````

```
4. Combining Characters
Some visible characters are made from multiple Unicode code points.
```

````python
text = "e\u0301"

print(text)
print(len(text))
````

```
Although it looks like é, it contains two Unicode code points: e and the combining accent.

5. Indexing Unicode Strings
Indexing accesses individual Unicode code points in a string.
```

````python
text = "café"

print(text[0])
print(text[3])
````

```
6. Case Sensitivity
Python treats uppercase and lowercase characters as different.
```

````python
print("Café" == "café")
````

```
7. lower()
Converts uppercase letters into lowercase letters.
```

````python
text = "HELLO PYTHON"

print(text.lower())
````

```
8. casefold()
Converts text into a form better suited for case-insensitive comparisons.
```

````python
text = "straße"

print(text.casefold())
````

```
9. lower() vs. casefold()
lower() performs ordinary lowercase conversion, while casefold() is more suitable for international case-insensitive comparisons.
```

````python
text = "STRAẞE"

print(text.lower())
print(text.casefold())
````

```
10. Unicode Normalization
Unicode normalization converts equivalent text representations into a consistent form.
```

````python
import unicodedata

text = "e\u0301"

normalized = unicodedata.normalize("NFC", text)

print(normalized)
print(len(normalized))
````

```
11. unicodedata.normalize()
A Python function used to normalize Unicode text when different code-point sequences represent the same visible character.
```

````python
import unicodedata

text1 = "é"
text2 = "e\u0301"

normalized1 = unicodedata.normalize("NFC", text1)
normalized2 = unicodedata.normalize("NFC", text2)

print(normalized1 == normalized2)
````
