```python
# Strings and how to create them:
greeting_a = 'Hello'
greeting_b = "Hello"
message = """This is
a multi-line
string."""

print(message)
print(greeting_a)
print(greeting_b)
```

```python
# The backslash introduces escape sequences: \n newline, \t tab, \" double quote, \' single quote, and \\ literal backslash.
print("Line one\nLine two")
print("Name:\tAda")
print("She said \"hello\"")
print('It\'s sunny today')
print("Backslash: \\")
```

```python
# A raw string r"..." prevents ordinary escape processing, useful for Windows paths:
normal_path = "C:\\Users\\Ada"
raw_path = r"C:\Users\Ada"
print(normal_path)
print(raw_path)
```

```python
# Immutability and methods:
name = "sam"
name.upper()
print(name)       # sam
name = name.upper()
print(name)       # SAM
```

```python
#  Indexing and safe access:
word = "python"
print(word[0])    # p
print(word[-1])   # n
print(word[-2])   # o
print(word[10])   # IndexError: string index out of range
```

```python
# For a calculated non-negative position, check it before access:
# check whether a position is valid before accessing a character from a string.
username = "sovon" # There are 5 characters, so the valid positive indexes are 0 to 4.
# We create a variable named position and assign the integer 10, 
# which represents the character index we want to access.
position = 10 
if 0 <= position < len(username):
    print(username[position])
else:
    print("That position does not exist in this text.")
```

```python
# Slicing:
word = "python"
print(word[0:3])  # pyt
print(word[2:])   # thon
print(word[:4])   # pyth
print(word[:])    # python
print(word[-3:])  # hon
print(word[::2])  # pto
print(word[::-1]) # nohtyp (reverse)
```

```python
# String operations:
first = "Py"
second = "thon"
print(first + second)       # Python (concatenate)
print("ha" * 3)             # hahaha (repeat)
print(len("python"))        # 6
print("th" in "python")     # True
print("xy" not in "python") # True
```

```python
# Formatting:
name = "Ada"
age = 30
print(f"{name} is {age} years old.")
```

```python
price = 1234.5
print(f"{price:.2f}")    # 1234.50
print(f"{price:,.2f}")   # 1,234.50
ratio = 0.4567
print(f"{ratio:.1%}")    # 45.7%
label = "Total"
print(f"[{label:>10}]")  # right aligned
print(f"[{label:<10}]")  # left aligned
print(f"[{label:^10}]")  # centered
```

```python
# Old formatting style:
print("{} is {} years old.".format(name, age))
print("{n} is {a} years old.".format(n=name, a=age))
print("%s is %d years old." % (name, age))
```

```python
# Unicode and case comparison:
# Python str supports accented letters, other scripts, symbols and emoji:
greeting = "café ☕ こんにちは 🐍"
print(greeting)
print(len(greeting))  # 14
text = "e\u0301"
print(text)           # visually é
print(len(text))      # 2 code points
```

```python
word = "straße"
print(word.lower())     # straße
print(word.casefold())  # strasse
```
