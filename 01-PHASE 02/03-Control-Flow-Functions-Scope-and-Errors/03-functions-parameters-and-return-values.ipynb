Topic: Functions, Parameters, and Return Values

1. What a function is
- Function: A reusable block of code that performs a specific task.
- Less repetition: Write code once and reuse it whenever needed.
- Easier to read: A meaningful function name tells us what the code does.
- Easier to test and change: Fixing a function in one place updates its behavior everywhere it is called.
- Function definition: Creating a function using def without executing its body.
- Function call: Executing a function by writing its name followed by parentheses.

```

````python
def greet():
    print("Hello!")

greet()
````

````python
greet()  # Hello!
````

```
2. Basic function syntax:
- def: A Python keyword used to define a function.
- Function name: The name used to identify and call a function.
- Parentheses (): Used to define parameters or call a function.
- Colon :: Marks the beginning of the function body.
- Indentation: Spaces used to identify which statements belong to the function.
- Meaningful function names: Names that clearly describe the action performed by a function.
```

````python
def greet():
    print("Hello!")
````

3. Parameters and arguments
- Parameter: A variable written inside the function definition that receives input.
```

````python
def greet(name):
    print(name)
````

```
- Argument: The actual value passed to a function when calling it.
```

````python
greet("Ada")  # "Ada" is the argument
````

```
- Local parameter: A parameter that is available inside the function during its execution.
```

````python
def greet(name):
    print(name)
````

```
- Multiple parameters: A function can accept multiple inputs separated by commas.
```

````python
def add(a, b):
    print(a + b)

add(10, 20)  # 30
````

4. Return values
- return: Sends a value back to the caller and immediately ends the function.
```

````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)  # 30
````

```
- print(): Displays a value on the screen but does not return that value to the caller.
``

````python
def add(a, b):
    print(a + b)

result = add(10, 20)
print(result)  # None
````

```
- Difference between print() and return: print() displays a result, while return sends a result back for further use.
```

````python
def show():
    print(10)

def give():
    return 10

x = show()  # None
y = give()  # 10
````

```
- None: The special Python value representing the absence of a returned value.
```

````python
def greet():
    print("Hello!")

result = greet()
print(result)  # None
````

```
- Implicit return: Python automatically returns None when a function has no explicit return statement.
```

````python
def greet():
    print("Hello!")

print(greet())  # None
````

```
- Explicit return: Returning a value directly using the return keyword.
```

````python
def get_number():
    return 100

print(get_number())  # 100
````

```
- type(): Checks the data type of a returned value.
```

````python
def get_value():
    return None

print(type(get_value()))  # <class 'NoneType'>
````

```
- Storing a returned value: Assigning a function's returned result to a variable.
```

````python
def calculate():
    return 50

total = calculate()
print(total)  # 50
````

```
- Reusing a returned value: Using a returned result in another calculation.
```

````python
def calculate():
    return 100

total = calculate()
tax = total * 0.10
print(total + tax)  # 110.0
````

5. Positional arguments:
- Positional argument: An argument matched to a parameter according to its position in the function call.
```

````python
def greet(name, age):
    print(name, age)

greet("Ada", 25)  # Ada 25
````

```
- Argument order: Positional arguments must follow the parameter order defined in the function.
```

````python
def greet(name, age):
    print(name, age)

greet(25, "Ada")  # 25 Ada
````

```
- Wrong positional order: Swapping arguments can produce incorrect results even when Python accepts them.
```

````python
def describe(name, score):
    print(name, score)

describe(90, "Ada")  # 90 Ada
````

```
6. Keyword arguments:
- Keyword argument: An argument passed by explicitly specifying its parameter name.
```

````python
def greet(name, age):
    print(name, age)

greet(name="Ada", age=25)
````

```
- Order-independent keyword arguments: Keyword arguments can be supplied in any order because Python matches them by name.
```

````python
def greet(name, age):
    print(name, age)

greet(age=25, name="Ada")
````

```
- Mixing positional and keyword arguments: You can combine both, provided positional arguments come first.
```

````python
def greet(name, age):
    print(name, age)

greet("Ada", age=25)
````

```
- Positional argument after keyword argument: Python raises a SyntaxError when an ordinary positional argument follows a keyword argument.
```

````python
greet(name="Ada", 25)
````

```
7. Default arguments:
- Default argument: A predefined value automatically used when the caller does not provide an argument.
```

````python
def greet(name, message="Hello"):
    print(message, name)

greet("Ada")  # Hello Ada
````

```
- Overriding a default: Supplying another value replaces the default value for that function call.
- Positional default override: A default value can be overridden by passing another argument in its position.
```

````python
def greet(name, message="Hello"):
    print(message, name)

greet("Ada", "Welcome")
````

```
- Keyword default override: A default value can also be overridden using its parameter name.
```

````python
def greet(name, message="Hello"):
    print(message, name)

greet("Ada", message="Welcome")
````

```
- Immutable default values: Numbers, strings, booleans, and None are safe from mutable-default sharing problems.
- Default evaluation: Python evaluates default argument values once when the function is defined.
```

````python
def greet(message="Hello"):
    print(message)

greet()
````

```
8. Mutable default argument warning:
- Mutable default argument: A mutable default such as [] is created once and shared across calls that omit that argument.
```

````python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("apple"))   # ['apple']
print(add_item("banana"))  # ['apple', 'banana']
````

```
- Shared list: Multiple function calls modify the same default list instead of receiving separate lists.
```

````python
def add_item(item, items=[]):
    items.append(item)
    return items

a = add_item("apple")
b = add_item("banana")

print(a is b)  # True
print(a)
print(b)
````

````python
# - Aliasing: Two variables refer to the exact same object in memory.
a = [10, 20]
b = a

print(a is b)  # True
````

````python
# - Safe mutable default: Use None as the default and create a new list inside the function.
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item("apple"))   # ['apple']
print(add_item("banana"))  # ['banana']
````

````python
# - is None: Checks whether a variable refers to Python's None value.
items = None

if items is None:
    items = []

print(items)  # []
````

````python
# - Mutable values: Lists, dictionaries, and sets can be modified after creation.
items = []
items.append("apple")

print(items)  # ['apple']
````

```
9. Keyword-only arguments:
- Keyword-only parameter: A parameter that must be passed using its name rather than its position.
```

````python
def greet(name, *, uppercase=False):
    if uppercase:
        return name.upper()
    return name

print(greet("Ada", uppercase=True))  # ADA
````

````python
# - Asterisk *: Marks all parameters after it as keyword-only.
def calculate(price, *, tax):
    return price + tax

print(calculate(100, tax=10))  # 110
````

```
- Keyword-only argument: Makes function calls clearer by explicitly identifying optional settings.
```

````python
def format_text(text, *, uppercase=False):
    return text.upper() if uppercase else text

print(format_text("hello", uppercase=True))  # HELLO
````

```
- Passing keyword-only arguments positionally: Python raises a TypeError if you try to pass a keyword-only parameter by position.
```

````python
def greet(name, *, uppercase=False):
    return name.upper() if uppercase else name

greet("Ada", True)
````

10. Returning multiple values
- Returning multiple values: A function can return multiple comma-separated values together.
```

````python
def get_scores():
    return 80, 90

print(get_scores())  # (80, 90)
````

```
- Tuple packing: Python automatically packs multiple returned values into a single tuple.
```

````python
def get_values():
    return 10, 20

result = get_values()
print(type(result))  # <class 'tuple'>
````

````python
# - Unpacking: Assigning individual tuple elements to separate variables.
def get_values():
    return 10, 20

a, b = get_values()

print(a)  # 10
print(b)  # 20
````

````python
# - min(): Returns the smallest value from a collection.
numbers = [10, 5, 20]
print(min(numbers))  # 5
````

````python
# - max(): Returns the largest value from a collection.
numbers = [10, 5, 20]
print(max(numbers))  # 20
````

````python
# - Returning minimum and maximum: A function can return both the smallest and largest values together.
def get_min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = get_min_max([10, 5, 20])

print(lowest, highest)  # 5 20
````

11. Docstrings:
- Docstring: A string written as the first statement inside a function to describe its purpose.
```

````python
def greet():
    """Display a greeting message."""
    print("Hello!")
````

````python
# - Function description: A docstring explains what a function does.
def add(a, b):
    """Add two numbers."""
    return a + b
````

````python
# - Parameter description: A docstring can explain what each function parameter represents.
def greet(name):
    """Display a greeting for the given name."""
    print(name)
````

````python
# - Return description: A docstring can explain what value the function returns.
def add(a, b):
    """Return the sum of two numbers."""
    return a + b
````

````python
# - .__doc__: Accesses the documentation string stored inside a function.
def greet():
    """Display a greeting."""
    print("Hello!")

print(greet.__doc__)  # Display a greeting.
````

````python
# - help(): Displays documentation and information about a function.
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

help(add)
````

Example 1 — Greeting a user: print versus return:
```

````python
def greet_user(name):
    print(f"Hello, {name}! Welcome back.")

def build_greeting(name):
    return f"Hello, {name}! Welcome back."

greet_result = greet_user("Ada")
print("greet_user gave back:", greet_result)

message = build_greeting("Ada")
print("build_greeting gave back:", message)
````

Example 2 — Calculating an order total and reusing the result:
```

````python
def calculate_total(unit_price, quantity):
    return unit_price * quantity

order_total = calculate_total(12.50, 4)
shipping_fee = 5
final_amount = order_total + shipping_fee

print(f"Order total: ${order_total:.2f}")
print(f"Final amount (with shipping): ${final_amount:.2f}")
````

```
Example 3 — Calculating a discount with a default argument:
```

````python
def calculate_discounted_price(price, discount_rate=0.10):
    return price * (1 - discount_rate)

standard_price = calculate_discounted_price(80)
print(f"Standard discount price: ${standard_price:.2f}")

member_price = calculate_discounted_price(80, discount_rate=0.25)
print(f"Member discount price: ${member_price:.2f}")
````

Example 4 — Validating a score range
```

````python
def is_score_valid(score, minimum=0, maximum=100):
    return score >= minimum and score <= maximum

scores_to_check = [85, -5, 100, 101]

for score in scores_to_check:
    if is_score_valid(score):
        print(score, "-> valid")
    else:
        print(score, "-> invalid")
````

Example 5 — Returning the lowest and highest values
```

````python
def find_lowest_and_highest(numbers):
    return min(numbers), max(numbers)

exam_scores = [72, 95, 68, 88, 91]
lowest, highest = find_lowest_and_highest(exam_scores)

print(f"Lowest score: {lowest}")
print(f"Highest score: {highest}")
````

Example 6 — Safely adding items to a list:
```

````python
def add_item_safe(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

cart_one = add_item_safe("apples")
cart_two = add_item_safe("bread")

print("Cart one:", cart_one)
print("Cart two:", cart_two)

existing_cart = ["milk"]
updated_cart = add_item_safe("eggs", existing_cart)
print("Updated existing cart:", updated_cart)
````

```
Example 7 — Formatting a report with keyword-only options:
```

````python
def format_report(title, *, uppercase=False, border=False):
    """
    Build a simple one-line report title.

    title is the text to display.
    uppercase, when True, converts the title to capital letters.
    border, when True, surrounds the title with dashes.
    Returns the formatted title as a string.
    """
    formatted_title = title
    if uppercase:
        formatted_title = formatted_title.upper()
    if border:
        formatted_title = f"--- {formatted_title} ---"
    return formatted_title

print(format_report("monthly summary"))
print(format_report("monthly summary", uppercase=True))
print(format_report("monthly summary", uppercase=True, border=True))
````
