```
Topic: Functions, Parameters, and Return Values

1. What a function is
- Function: A reusable block of code that performs a specific task.
- Less repetition: Write code once and reuse it whenever needed.
- Easier to read: A meaningful function name tells us what the code does.
- Easier to test and change: Fixing a function in one place updates its behavior everywhere it is called.
- Function definition: Creating a function using def without executing its body.
- Function call: Executing a function by writing its name followed by parentheses.

```

````python
def greet():
    print("Hello!")

greet()
````

````python
greet()  # Hello!
````

```
2. Basic function syntax:
- def: A Python keyword used to define a function.
- Function name: The name used to identify and call a function.
- Parentheses (): Used to define parameters or call a function.
- Colon :: Marks the beginning of the function body.
- Indentation: Spaces used to identify which statements belong to the function.
- Meaningful function names: Names that clearly describe the action performed by a function.
```

````python
def greet():
    print("Hello!")
````

```
3. Parameters and arguments
- Parameter: A variable written inside the function definition that receives input.
```

````python
def greet(name):
    print(name)
````

```
- Argument: The actual value passed to a function when calling it.
```

````python
greet("Ada")  # "Ada" is the argument
````

```
- Local parameter: A parameter that is available inside the function during its execution.
```

````python
def greet(name):
    print(name)
````

```
- Multiple parameters: A function can accept multiple inputs separated by commas.
```

````python
def add(a, b):
    print(a + b)

add(10, 20)  # 30
````

```
4. Return values
- return: Sends a value back to the caller and immediately ends the function.
```

````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)  # 30
````

```
- print(): Displays a value on the screen but does not return that value to the caller.
``

````python
def add(a, b):
    print(a + b)

result = add(10, 20)
print(result)  # None
````

```
- Difference between print() and return: print() displays a result, while return sends a result back for further use.
```

````python
def show():
    print(10)

def give():
    return 10

x = show()  # None
y = give()  # 10
````

```
- None: The special Python value representing the absence of a returned value.
```

````python
def greet():
    print("Hello!")

result = greet()
print(result)  # None
````

```
- Implicit return: Python automatically returns None when a function has no explicit return statement.
```

````python
def greet():
    print("Hello!")

print(greet())  # None
````

```
- Explicit return: Returning a value directly using the return keyword.
```

````python
def get_number():
    return 100

print(get_number())  # 100
````

```
- type(): Checks the data type of a returned value.
```

````python
def get_value():
    return None

print(type(get_value()))  # <class 'NoneType'>
````

```
- Storing a returned value: Assigning a function's returned result to a variable.
```

````python
def calculate():
    return 50

total = calculate()
print(total)  # 50
````

```
- Reusing a returned value: Using a returned result in another calculation.
```

````python
def calculate():
    return 100

total = calculate()
tax = total * 0.10
print(total + tax)  # 110.0
````

```
5. Positional arguments:
- Positional argument: An argument matched to a parameter according to its position in the function call.
```

````python
def greet(name, age):
    print(name, age)

greet("Ada", 25)  # Ada 25
````

```
- Argument order: Positional arguments must follow the parameter order defined in the function.
```

````python
def greet(name, age):
    print(name, age)

greet(25, "Ada")  # 25 Ada
````

```
- Wrong positional order: Swapping arguments can produce incorrect results even when Python accepts them.
```

````python
def describe(name, score):
    print(name, score)

describe(90, "Ada")  # 90 Ada
````
