# 1. What iteration is, and why repeated work needs loops

| Concept | Simple Explanation | Example |
|---|---|---|
| Iteration | Repeating a block of code multiple times. | `for i in range(3): print("Hello")` |
| Repetition | Performing the same task again and again. | `print("Hello")` three times using a loop. |
| Loop | A programming structure that repeats code. | `for number in range(5): print(number)` |
| `_` (underscore) | A variable conventionally used when you don't need the loop value. | `for _ in range(3): print("Hello")` |

## 2. `for` Loops

| Concept | Simple Explanation | Example |
|---|---|---|
| `for` loop | Repeats code once for every item in an iterable. | `for fruit in fruits: print(fruit)` |
| Iterable | An object Python can go through one item at a time. | `["apple", "banana"]` |
| Loop variable | A variable that receives each item during a loop. | `for fruit in fruits:` |
| Loop body | The indented code that executes during each iteration. | `print(fruit)` |
| Iteration order | The order in which Python visits items. | First `"apple"`, then `"banana"`. |
| Looping through a string | Visits each character of a string individually. | `for letter in "cat": print(letter)` |
| Looping through a list | Visits each list element individually. | `for fruit in ["apple", "banana"]: print(fruit)` |
| Looping through a tuple | Visits each tuple element individually. | `for value in (10, 20, 30): print(value)` |
| Looping through a dictionary | Visits dictionary keys by default. | `for key in {"name": "Ada", "age": 25}: print(key)` |
| `.items()` | Returns dictionary key-value pairs for looping. | `for key, value in data.items(): print(key, value)` |
| Looping through a set | Visits each unique set element; order is not guaranteed. | `for color in {"red", "blue"}: print(color)` |
| `sorted()` | Returns items in sorted order. | `for color in sorted({"red", "blue"}): print(color)` |

````python
prices = {"apple": 0.50, "bread": 2.75, "milk": 1.90}

for product, price in prices.items():
    print(product, "costs", price)
````

# 3. `range()`

| Concept | Simple Explanation | Example |
|---|---|---|
| `range()` | Generates a sequence of integers. | `range(5)` |
| Start | The number where the range begins. | `range(2, 10)` starts at `2`. |
| Stop | The number where the range ends, but it is excluded. | `range(2, 5)` produces `2, 3, 4`. |
| Step | The amount by which the number increases or decreases. | `range(2, 10, 2)` |
| Default start | `range()` starts at `0` when start is not provided. | `range(5)` |
| Default step | `range()` increases by `1` when step is not provided. | `range(1, 5)` |
| Negative step | Counts backward instead of forward. | `range(5, 0, -1)` |
| Empty range | A range that produces no numbers. | `range(5, 5)` |
| Excluded stop value | The stop number is never included in the generated sequence. | `range(1, 5)` produces `1, 2, 3, 4`. |

````python
for number in range(2, 10, 2):
    print(number)
````

````python
for number in range(5, 0, -1):
    print(number)
````

````python
for number in range(0, 5):
    print(number)
````

# 4. `while` Loops

| Concept | Simple Explanation | Example |
|---|---|---|
| `while` loop | Repeats code as long as its condition is `True`. | `while count < 5:` |
| Loop condition | A Boolean expression that decides whether the loop continues. | `count < 5` |
| Condition checking | Python checks the condition before every iteration. | `while age < 18:` |
| State update | Changes a variable that controls the loop condition. | `count = count + 1` |
| Loop termination | The point when a loop stops executing. | `count < 5` becomes `False`. |
| `for` vs `while` | Use `for` for known iterations and `while` for condition-based repetition. | `for i in range(5)` vs `while count < 5` |

````python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown = countdown - 1

print("Liftoff!")
````

# 5. Infinite Loops

| Concept | Simple Explanation | Example |
|---|---|---|
| Infinite loop | A loop that never stops because its condition always remains `True`. | `while True: print("Hello")` |
| Accidental infinite loop | An unintended loop caused by incorrect loop logic. | Forgetting to update `count`. |
| Loop update | Changes the variable so the loop can eventually stop. | `count = count + 1` |
| `Ctrl + C` | Stops a running Python program from the terminal. | Press `Ctrl + C` when a loop runs endlessly. |
| Loop prevention | Ensure the loop has a reliable termination condition. | `while count < 5: count += 1` |

````python
count = 0

while count < 5:
    print(count)
    count = count + 1
````

# 6. `break`: Stopping a Loop Early

| Concept | Simple Explanation | Example |
|---|---|---|
| `break` | Immediately exits the loop completely. | `if number == 5: break` |
| Early termination | Stops a loop before all iterations are completed. | Stop searching once an item is found. |
| Search loop | Checks items until the required item is found. | `for name in names:` |
| `found` flag | A Boolean variable used to track whether something was found. | `found = False` |

````python
usernames = ["ada", "grace", "alan", "marie"]

for username in usernames:
    if username == "alan":
        print("User found!")
        break
````

# 7. `continue`: Skipping Just One Iteration

| Concept | Simple Explanation | Example |
|---|---|---|
| `continue` | Skips the remaining code in the current iteration and moves to the next iteration. | `if number == 3: continue` |
| Current iteration | The loop's current execution for one item. | Processing number `3`. |
| Skip condition | A condition that decides when an iteration should be skipped. | `if number < 0:` |
| `break` vs `continue` | `break` exits the loop; `continue` skips only the current iteration. | `break` stops; `continue` skips. |

````python
numbers = [1, 2, -3, 4, -5]

for number in numbers:
    if number < 0:
        continue

    print(number)
````

# 8. `enumerate()`: Getting Both a Position and a Value

| Concept | Simple Explanation | Example |
|---|---|---|
| `enumerate()` | Provides both the index and value of each item during iteration. | `enumerate(fruits)` |
| Position / Index | The numerical location of an item in a collection. | Index of `"apple"` is `0`. |
| Value | The actual item stored at a particular position. | `"apple"` |
| Default starting index | `enumerate()` starts counting from `0` by default. | `enumerate(tasks)` |
| `start=1` | Makes enumeration start counting from `1`. | `enumerate(tasks, start=1)` |
| Tuple unpacking | Assigns the two values from `enumerate()` to separate variables. | `for index, task in enumerate(tasks):` |

````python
tasks = ["Write report", "Review code", "Email client"]

for position, task in enumerate(tasks, start=1):
    print(position, task)
````

# 9. `zip()`: Looping Through Two Collections Together

| Concept | Simple Explanation | Example |
|---|---|---|
| `zip()` | Combines matching-position items from two or more iterables. | `zip(names, scores)` |
| Pairing | Combines related items based on their positions. | `"Ada"` with `92`. |
| Parallel iteration | Loops through multiple collections together. | `for name, score in zip(names, scores):` |
| Shortest iterable | `zip()` stops when the shortest iterable runs out of items. | 4 names and 3 scores produce 3 pairs. |
| Unequal lengths | Extra items in longer iterables are ignored by default. | The fourth name is not processed. |
| `strict=True` | Raises `ValueError` if the iterables have different lengths (Python 3.10+). | `zip(names, scores, strict=True)` |

````python
names = ["Ada", "Grace", "Alan"]
scores = [92, 88, 79]

for name, score in zip(names, scores):
    print(name, score)
````

````python
names = ["Ada", "Grace", "Alan", "Marie"]
scores = [92, 88, 79]

for name, score in zip(names, scores):
    print(name, score)
````

# 10. Choosing Between `for`, `while`, `break`, and `continue`

| Concept | When to Use It | Example |
|---|---|---|
| `for` | When you have a collection or a fixed number of iterations. | `for item in items:` |
| `while` | When repetition depends on a condition. | `while balance < 100:` |
| `break` | When you want to stop the entire loop immediately. | `if found: break` |
| `continue` | When you want to skip an item but continue the loop. | `if number < 0: continue` |

## Important Differences to Remember

| `break` | `continue` |
|---|---|
| Exits the entire loop. | Skips only the current iteration. |
| No further iterations happen. | Remaining iterations continue. |
| Used when the goal is reached. | Used when an item should be skipped. |

| `for` | `while` |
|---|---|
| Iterates over an iterable. | Repeats while a condition is `True`. |
| Commonly used when items or repetitions are known. | Commonly used when repetitions depend on a condition. |
| Example: `for i in range(5):` | Example: `while count < 5:` |
