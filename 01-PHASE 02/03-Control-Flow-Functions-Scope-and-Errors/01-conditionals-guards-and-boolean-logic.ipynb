```
Topic: Conditionals, Guards, and Boolean Logic

Conditionals and Boolean logic allow a Python program to make decisions and execute different instructions based on given conditions, while guards keep the code clean and match handles multiple fixed choices.

Why this topic matters
- Program decisions: Programs need to make decisions based on situations or conditions.
- Control flow: Controls which lines of code execute, skip, or repeat depending on the data.
- Guard conditions: Check invalid or special inputs early to keep code simple, flat, and readable.
- match statement: Allows a program to choose an action by matching a value against predefined options.
```

````python
# 2. Control flow: Controls which instructions Python executes, skips, or repeats.
age = 20

if age >= 18:
    print("Adult")
````

````python
# 5. if / elif / else statements
# 1. elif: Checks another condition when the previous if or elif condition is False.

score = 75

if score >= 90:
    print("Grade A")
elif score >= 70:
    print("Grade B")
else:
    print("Grade C")
````

````python
# D. Parentheses for clarity
# 1. Parentheses (()): Group conditions to make the intended evaluation order clear.
age = 16
has_guardian = True

if age >= 18 or (age >= 13 and has_guardian):
    print("Entry allowed")
````

````python
# 4. Safe condition checking: Check dangerous operations first to prevent errors.
denominator = 0

if denominator != 0 and 10 / denominator > 1:
    print("Valid result")
else:
    print("Cannot divide by zero")
````

````python
# 2. Nested if: An if statement placed inside another if statement.
age = 20

if age > 0:
    if age >= 18:
        print("Adult")
````

````python
# 3. Nested code: Code placed inside multiple levels of other code blocks.
age = 20
has_id = True

if age > 0:
    if age >= 18:
        if has_id:
            print("Access granted")
````

```
5. Guard clause: Checks a condition and exits the current code path early using return, raise, continue, or break.

Example using return:
```

````python
def check_age(age):
    if age < 0:
        return "Invalid age"

    return "Valid age"

print(check_age(-5))  # Invalid age
print(check_age(20))  # Valid age
````

````python
# 7. Early validation: Checks whether input is valid before allowing the program to continue.
def register_user(username):
    if not username:
        return "Username is required"

    return "Registration successful"

print(register_user(""))       # Username is required
print(register_user("Sovon"))  # Registration successful
````

```
8. Match-style branching:
1. match statement: Compares a value against multiple patterns and executes the first matching case.
```

````python
day = "Monday"

match day:
    case "Monday":
        print("Start of the week")
    case "Friday":
        print("Almost weekend")
    case "Sunday":
        print("Weekend")
````

```
2. case: Defines a possible matching option inside a match statement.
```

````python
color = "red"

match color:
    case "red":
        print("Red color")
    case "blue":
        print("Blue color")
````

```
3. Literal pattern: Matches an exact value.
```

````python
color = "red"

match color:
    case "red":
        print("You selected red")
    case "blue":
        print("You selected blue")
````

```
4. Wildcard pattern (_): Matches anything not matched by the previous cases, acting like else.
```

````python
color = "green"

match color:
    case "red":
        print("Red")
    case "blue":
        print("Blue")
    case _:
        print("Unknown color")
````

```
5. Default case: Handles all values that do not match the explicitly defined cases.
```

````python
status = "unknown"

match status:
    case "success":
        print("Operation successful")
    case "error":
        print("Something went wrong")
    case _:
        print("Unknown status")
````

````python
# Example 1 — Driving permit eligibility (if/elif/else with and):
age = 16
has_guardian_consent = True

if age >= 18:
    print("You can get a full driving permit.")
elif age >= 16 and has_guardian_consent:
    print("You can get a learner's permit with guardian consent.")
else:
    print("You are not yet eligible for a driving permit.")
````

````python
# Example 2 — Password rule checks with a guard chain:
candidate_passwords = ["", "short1", "longenoughbutnodigits", "hunter22"]

for password in candidate_passwords:
    has_min_length = len(password) >= 8

    has_digit = False
    for character in password:
        if character.isdigit():
            has_digit = True

    if password == "":
        print(password, "-> Error: password cannot be empty.")
    elif not has_min_length:
        print(password, "-> Error: password must be at least 8 characters long.")
    elif not has_digit:
        print(password, "-> Error: password must contain at least one digit.")
    else:
        print(password, "-> Password meets all the rules.")
````

````python
# Example 3 — Traffic light instructions with match:
light_colors = ["red", "yellow", "green", "flashing yellow"]

for light_color in light_colors:
    match light_color:
        case "red":
            instruction = "Stop"
        case "yellow":
            instruction = "Prepare to stop"
        case "green":
            instruction = "Go"
        case _:
            instruction = "Proceed with caution"
    print(light_color, "->", instruction)
````

````python
# Example 4 — An optional value with None:
customer_referral_code = None
bonus_points = 0

if customer_referral_code is None:
    print("No referral code was used.")
elif customer_referral_code == "":
    print("Referral code field was left blank on purpose.")
else:
    bonus_points = 50
    print("Referral code", customer_referral_code, "applied.")

print("Bonus points earned:", bonus_points)
````
